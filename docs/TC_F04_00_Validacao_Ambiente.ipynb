# 1.&nbsp;Contextualização e planejamento do projeto




## 1.1 Contexto e problema de negócio

Os marketplaces digitais disponibilizam informações sobre produtos, preços, descontos e avaliações de consumidores. A análise integrada desses atributos pode contribuir para identificar diferenças no portfólio, padrões de engajamento, níveis de satisfação e oportunidades de melhoria.

O Tech Challenge — Fase 04 propõe o desenvolvimento de uma solução analítica para apoiar decisões em uma empresa de e-commerce, utilizando o Amazon Sales Dataset.

O problema central definido no enunciado consiste em identificar quais produtos, categorias e estratégias comerciais merecem maior atenção e quais oportunidades podem ser reconhecidas a partir dos dados disponíveis.

A investigação considera quatro dimensões relacionadas ao negócio:

- **Portfólio:** características e diferenças entre produtos e categorias.
- **Estratégia comercial:** preços originais, preços com desconto e percentuais de desconto.
- **Engajamento e satisfação:** quantidade de avaliações e ratings atribuídos aos produtos.
- **Experiência do consumidor:** opiniões, percepções e problemas relatados nos comentários.

A finalidade é transformar os dados em informações relevantes para a liderança, combinando indicadores quantitativos, interpretação dos comentários e recomendações fundamentadas.

**Referência:** FIAP. *Tech Challenge — Amazon Product Performance & Customer Insights*, pp. 2–4.

---



## 1.2 Objetivos do projeto

### Objetivo geral

Desenvolver uma solução de Data Analytics, com dashboard interativo no Microsoft Power BI, para analisar o portfólio de produtos, os preços, os descontos, o engajamento e a satisfação dos consumidores, produzindo insights e recomendações úteis à tomada de decisão.

### Objetivos específicos

1. Caracterizar e avaliar a qualidade do conjunto de dados utilizado.
2. Investigar produtos e categorias, identificando diferenças e pontos de atenção.
3. Examinar preços, descontos e suas associações com indicadores de avaliação dos produtos.
4. Analisar engajamento e satisfação a partir das informações disponíveis.
5. Explorar comentários dos consumidores para identificar temas, percepções e problemas recorrentes.
6. Desenvolver indicadores e visualizações interativas para apoiar a interpretação dos resultados.
7. Formular recomendações de negócio fundamentadas nas evidências produzidas.
8. Avaliar a viabilidade de uma análise preditiva complementar, sem comprometer os requisitos obrigatórios.

O objetivo preditivo constitui uma ampliação do escopo acadêmico adotada pelo grupo. Sua implementação depende da viabilidade metodológica e da qualidade dos dados.

## 1.6 Premissas, limitações e estratégia metodológica

### Premissas

A análise será desenvolvida com base nos atributos efetivamente disponíveis no conjunto de dados.

Os procedimentos serão documentados de maneira reproduzível, com preservação da fonte original, descrição das transformações, validação dos resultados e registro das decisões metodológicas.

### Limitações preliminares

Antes da análise, devem ser consideradas as seguintes restrições:

- A quantidade de avaliações não equivale à quantidade de unidades vendidas.
- Preços de produtos não representam receita de vendas.
- Associações estatísticas não demonstram, isoladamente, relações causais.
- Não será presumida a existência de séries temporais sem variáveis de data adequadas.
- A representatividade das avaliações dos consumidores precisa ser considerada na interpretação dos resultados.
- A granularidade dos registros e a unicidade dos identificadores exigem verificação.
- Modelos preditivos deverão considerar riscos de vazamento de dados e dependência entre observações.

Essas limitações orientam o trabalho, mas não constituem diagnóstico de problemas já comprovados na base.

### Estratégia metodológica

O projeto seguirá uma sequência de investigação:

1. Preparação e validação do ambiente de trabalho.
2. Preservação e verificação da integridade dos dados originais.
3. Diagnóstico estrutural e avaliação da qualidade dos dados.
4. Tratamentos justificados e documentados, quando necessários.
5. Análise exploratória e desenvolvimento de indicadores.
6. Exploração dos comentários e percepções dos consumidores.
7. Avaliação da viabilidade de modelos preditivos complementares.
8. Desenvolvimento e validação do dashboard em Power BI.
9. Consolidação dos insights e recomendações.
10. Elaboração da apresentação executiva.

A aplicação de técnicas e transformações dependerá das características efetivamente observadas nos dados.

O desenvolvimento deverá priorizar a validade das análises, a clareza das conclusões e a utilidade das recomendações para o negócio.

## 2.2. Organização da estrutura de diretórios

### Objetivo

Organizar o armazenamento do Tech Challenge
em diretórios específicos, permitindo separar
os dados originais dos arquivos derivados,
códigos, modelos, relatórios e evidências.

### Fundamentação metodológica

A organização dos arquivos contribui para
a rastreabilidade das fontes, preservação dos
dados originais e reprodutibilidade das análises.

A estrutura foi definida conforme as necessidades
do projeto, como decisão técnica da equipe.

### Metodologia

A criação e a verificação dos diretórios foram
realizadas no Windows, utilizando o PowerShell.

O procedimento identificou a pasta principal,
verificou a existência das subpastas previstas
e criou os diretórios que ainda não existiam.

**Ferramenta:** Windows PowerShell

**Diretório principal:**

`G:\Meu Drive\FIAP_TC_FASE04_AMAZON`

### Comando de organização

O código abaixo corresponde ao comando utilizado
como referência para a organização do projeto.

~~~powershell
$raiz = "G:\Meu Drive\FIAP_TC_FASE04_AMAZON"

$pastas = @(
    "00_GOVERNANCA",
    "01_DADOS_ORIGINAIS",
    "02_DADOS_TRATADOS",
    "03_NOTEBOOKS_COLAB",
    "04_MODELOS_PREDITIVOS",
    "05_POWER_BI",
    "06_EVIDENCIAS_LOGS",
    "07_ENTREGAVEIS",
    "08_BACKUPS"
)

if (-not (Test-Path -LiteralPath $raiz -PathType Container)) {
    throw "Pasta principal nao encontrada: $raiz"
}

foreach ($pasta in $pastas) {
    $destino = Join-Path $raiz $pasta

    if (-not (Test-Path -LiteralPath $destino)) {
        New-Item -ItemType Directory -Path $destino | Out-Null
    }
}

Get-ChildItem -LiteralPath $raiz -Directory |
    Select-Object -ExpandProperty Name
~~~

### Resultados obtidos

A verificação realizada no PowerShell retornou

~~~

## 2.3. Configuração e validação do armazenamento persistente

### Objetivo

Verificar a integração entre o ambiente Google Colab
e o Google Drive, assegurando o acesso à estrutura de
diretórios definida para o Tech Challenge — Fase 04.

### Fundamentação metodológica

O ambiente de execução do Google Colab é temporário.
Por isso, foi adotado o Google Drive como repositório
persistente para notebooks, bases de dados, resultados,
documentação e demais artefatos do projeto.

Essa configuração integra a estratégia técnica de
organização e reprodutibilidade da análise.

### Metodologia

A verificação foi implementada em Python, utilizando:

- `google.colab.drive` para montagem do Google Drive;
- `pathlib.Path` para identificação dos diretórios;
- `is_dir()` para verificar a existência das pastas.

Foram examinadas a pasta principal do projeto e as
nove subpastas previstas na estrutura de trabalho.

### Critérios de validação

1. Montagem bem-sucedida do Google Drive.
2. Localização da pasta principal do projeto.
3. Identificação das nove subpastas esperadas.
4. Resultado consolidado de 9/9.
5. Ausência de erros durante a execução.

### Escopo

O procedimento verifica o acesso aos diretórios,
sem realizar leitura analítica, transformação ou
alteração da base Amazon.


```python

# Tech Challenge FIAP - Fase 04
# Procedimento 2.3 - Verificação do Google Drive
# Nenhum dado será alterado.

from google.colab import drive
from pathlib import Path

# Montar o Google Drive
drive.mount('/content/drive')

# Caminho principal do projeto
raiz = Path(
    '/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON'
)

# Subpastas esperadas
pastas_esperadas = [
    '00_GOVERNANCA',
    '01_DADOS_ORIGINAIS',
    '02_DADOS_TRATADOS',
    '03_NOTEBOOKS_COLAB',
    '04_MODELOS_PREDITIVOS',
    '05_POWER_BI',
    '06_EVIDENCIAS_LOGS',
    '07_ENTREGAVEIS',
    '08_BACKUPS'
]

print('\nVERIFICAÇÃO DO PROJETO')

if raiz.is_dir():
    print('Pasta principal: ENCONTRADA')

    for nome in pastas_esperadas:
        pasta = raiz / nome
        status = 'OK' if pasta.is_dir() else 'AUSENTE'
        print(f'{nome}: {status}')

    total = sum(
        (raiz / nome).is_dir()
        for nome in pastas_esperadas
    )

    print(f'\nSubpastas validadas: {total}/9')
else:
    print('ERRO: Pasta principal não encontrada')
```

## 3.2. Verificação de integridade da base original — SHA-256

### Objetivo

Estabelecer uma referência de integridade para o arquivo
original do Amazon Sales Dataset, mediante identificação
de seu tamanho exato e cálculo do hash SHA-256.

### Fundamentação metodológica

A preservação dos dados originais é importante para
assegurar a rastreabilidade e a reprodutibilidade de
um projeto analítico.

O algoritmo SHA-256 produz um resumo criptográfico
do conteúdo de um arquivo. A comparação desse valor
em diferentes momentos permite identificar eventuais
alterações em seus bytes.

A utilização desse mecanismo integra os controles
técnicos de integridade definidos para o projeto.

**Referência ao problema acadêmico:**

FIAP. *Tech Challenge — Amazon Product Performance
& Customer Insights*, pp. 3 e 7.

O enunciado identifica a fonte de dados e estabelece
a preparação dos dados como critério de avaliação.
O uso específico de SHA-256 constitui uma escolha
metodológica da equipe.

### Arquivo verificado

**Nome:** `01.amazon.csv`

**Formato:** CSV

**Localização:**

`G:\Meu Drive\FIAP_TC_FASE04_AMAZON\01_DADOS_ORIGINAIS\01.amazon.csv`

### Metodologia

A verificação foi realizada no Windows PowerShell,
sem utilização de operações de escrita no arquivo.

Foram executadas as seguintes verificações:

1. Confirmação da existência do arquivo.
2. Identificação do caminho completo.
3. Obtenção do tamanho exato em bytes.
4. Cálculo do hash com algoritmo SHA-256.
5. Registro dos resultados como referência de integridade.

**Ferramenta:** Windows PowerShell

### Comando utilizado

O comando a seguir documenta a verificação realizada
no ambiente Windows. Não deve ser executado como
código Python no Google Colab.

~~~powershell
$arquivo = "G:\Meu Drive\FIAP_TC_FASE04_AMAZON\01_DADOS_ORIGINAIS\01.amazon.csv"

if (-not (Test-Path -LiteralPath $arquivo -PathType Leaf)) {
    Write-Host "ERRO: Arquivo nao encontrado."
    return
}

$item = Get-Item -LiteralPath $arquivo
$hash = Get-FileHash -LiteralPath $arquivo -Algorithm SHA256

Write-Host "=== INTEGRIDADE DA BASE ORIGINAL ==="
Write-Host "Nome:" $item.Name
Write-Host "Caminho:" $item.FullName
Write-Host "Tamanho exato (bytes):" $item.Length
Write-Host "Algoritmo:" $hash.Algorithm
Write-Host "SHA-256:" $hash.Hash
Write-Host "=== FIM DA VERIFICACAO ==="
~~~

### Resultados obtidos

| Indicador | Resultado |
|---|---|
| Arquivo | `01.amazon.csv` |
| Tamanho exato | 4.744.481 bytes |
| Algoritmo | SHA-256 |
| Execução | Sem erros relatados |

**Hash SHA-256 registrado:**

~~~text
4BA126C4BA8EDD35E62E94CE1C853073D832DE380184AB4F5F97D1E314882C77
~~~

O hash possui 64 caracteres hexadecimais,
correspondentes à representação de 256 bits.

### Interpretação dos resultados

O procedimento estabeleceu uma referência
criptográfica de integridade para o arquivo
original armazenado no projeto.

Em verificações posteriores, o cálculo de
SHA-256 sobre o mesmo arquivo poderá ser

A correspondência dos hashes constitui uma

4. A integridade deverá ser novamente verificada


## 3.3. Diagnóstico inicial da estrutura dos dados

### Objetivo

Caracterizar a estrutura do Amazon Sales Dataset,
identificando as dimensões, os nomes das variáveis,
a regularidade dos registros e os padrões aparentes
de representação dos valores.

### Fundamentação acadêmica

O enunciado do Tech Challenge — Fase 04 estabelece
a análise de produtos, categorias, preços, descontos,
avaliações e comentários dos consumidores.

A disciplina Análise de Negócios, na Aula 03 —
Geração de Insights, apresenta a exploração e a
preparação dos dados como componentes do processo
de identificação e validação de padrões.

O diagnóstico estrutural constitui uma etapa
preliminar para compreender a organização dos
dados antes da definição de transformações e
indicadores.

**Referências:**
- FIAP. *Tech Challenge — Amazon Product Performance
  & Customer Insights*, p. 3.
- KIRIYAMA, Edgard Joseph. *Análise de Negócios —
  Aula 03: Geração de Insights*. FIAP, p. 3.

### Fonte de dados

**Arquivo:** `01.amazon.csv`

**Diretório:** `01_DADOS_ORIGINAIS`

O arquivo original foi preservado e possui
referência de integridade SHA-256 documentada
na seção 3.2.

### Metodologia

O procedimento foi implementado em Python,
utilizando as bibliotecas `csv`, `pathlib` e `re`.

Foram realizadas:

1. Leitura do cabeçalho original do CSV.
2. Contagem de registros e colunas.
3. Identificação dos nomes das variáveis.
4. Verificação da quantidade de campos por registro.
5. Identificação de cabeçalhos duplicados.
6. Inspeção preliminar de padrões textuais e
   aparentemente numéricos em até 500 registros.

A leitura dos valores não envolveu conversão
de tipos nem modificação do arquivo original.

### Critérios de validação

- Arquivo localizado e acessível.
- Dimensões identificadas.
- Cabeçalhos examinados.
- Regularidade estrutural quantificada.
- Padrões aparentes identificados.
- Ausência de alterações no arquivo original.

### Escopo

O procedimento limita-se ao diagnóstico estrutural.

Não compreende limpeza, imputação, exclusão,
normalização, conversão ou agregação dos dados.



```python

# FIAP - TECH CHALLENGE FASE 04
# Procedimento 3.3 - Diagnostico estrutural
# SOMENTE LEITURA - SEM TRATAMENTO DE DADOS

import csv
import re
from pathlib import Path
from collections import Counter
from google.colab import drive

# 1. Verificar conexão com Google Drive
if not Path("/content/drive/MyDrive").is_dir():
    drive.mount("/content/drive")

arquivo = Path(
    "/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON/"
    "01_DADOS_ORIGINAIS/01.amazon.csv"
)

if not arquivo.is_file():
    raise FileNotFoundError(f"Arquivo nao encontrado: {arquivo}")

# 2. Identificar padroes sem converter valores
def padrao_aparente(valores):
    validos = [v.strip() for v in valores if v.strip()]

    if not validos:
        return "Indeterminado (amostra vazia)"

    if all(re.fullmatch(r"[+-]?\d+", v) for v in validos):
        return "Inteiro aparente (texto)"

    if all(
        re.fullmatch(r"[+-]?(?:\d+(?:[.,]\d+)?|[.,]\d+)", v)
        for v in validos
    ):
        return "Numerico aparente (texto)"

    return "Texto ou misto"

# 3. Ler o CSV preservando os valores como texto
with arquivo.open(
    mode="r",
    encoding="utf-8-sig",
    newline=""
) as f:

    leitor = csv.reader(
        f,
        delimiter=",",
        quotechar='"',
        strict=True
    )

    cabecalho = next(leitor)
    total_colunas = len(cabecalho)

    total_registros = 0
    registros_regulares = 0
    registros_irregulares = 0
    exemplos_irregulares = []

    amostras = [[] for _ in range(total_colunas)]

    for linha in leitor:
        total_registros += 1

        if len(linha) != total_colunas:
            registros_irregulares += 1

            if len(exemplos_irregulares) < 10:
                exemplos_irregulares.append(
                    (total_registros, len(linha))
                )
            continue

        registros_regulares += 1

        if registros_regulares <= 500:
            for i, valor in enumerate(linha):
                amostras[i].append(valor)

# 4. Conferir nomes duplicados
contagem = Counter(cabecalho)

duplicados = [
    nome for nome, qtd in contagem.items()
    if qtd > 1
]

# 5. Exibir resultados
print("=" * 55)
print("DIAGNOSTICO ESTRUTURAL AMAZON")
print("=" * 55)

print(f"Arquivo: {arquivo.name}")
print(f"Registros de dados: {total_registros}")
print(f"Colunas: {total_colunas}")
print(f"Registros estruturalmente regulares: {registros_regulares}")
print(f"Registros estruturalmente irregulares: {registros_irregulares}")

print("\nCOLUNAS E TIPOS APARENTES")
print("-" * 55)

for i, nome in enumerate(cabecalho):
    print(
        f"{i+1:02d}. {nome} | "
        f"Fisico: texto CSV | "
        f"Padrao: {padrao_aparente(amostras[i])}"
    )

print("\nVERIFICACAO ESTRUTURAL")
print("-" * 55)

print(f"Cabecalhos duplicados: {duplicados}")
print(f"Registros com quantidade irregular de campos: {registros_irregulares}")

if exemplos_irregulares:
    print("Exemplos (registro, numero de campos):")
    for exemplo in exemplos_irregulares:
        print(exemplo)

print("\nAmostra de registros regulares examinados:",
      min(registros_regulares, 500))

print("\nNenhum tratamento de dados foi realizado.")
print("=" * 55)
```


### 3.3.1. Resultados e discussão

O diagnóstico estrutural identificou 1.465
registros distribuídos em 16 variáveis.

Não foram observadas irregularidades na quantidade
de campos por registro nem duplicidades nos nomes
dos cabeçalhos.

#### Resumo dos resultados

|---|---:|
| Registros analisados | 1.465 |
| Variáveis identificadas | 16 |
| Registros estruturalmente regulares | 1.465 |
| Registros estruturalmente irregulares | 0 |

#### Caracterização preliminar

Esses formatos exigem investigação específica
antes da definição de possíveis conversões.

#### Interpretação

## 3.4. Diagnóstico preliminar da qualidade dos dados

### Objetivo

Investigar características de qualidade do Amazon Sales Dataset,
identificando valores ausentes, possíveis marcadores de ausência,
duplicidades integrais, repetições de identificadores e padrões
de representação dos atributos numéricos.

### Fundamentação acadêmica

O Tech Challenge — Fase 04 requer uma solução analítica
fundamentada em dados preparados e indicadores confiáveis.

A avaliação preliminar da qualidade antecede eventuais
transformações, permitindo identificar problemas e definir
critérios técnicos para sua investigação.

**Referências:**
- FIAP. Tech Challenge — Amazon Product Performance
  & Customer Insights, pp. 3, 6–7.
- FIAP. Conexões e Tratamento de Dados, Aula 02, p. 3.

### Metodologia

O diagnóstico será realizado em Python, com leitura
do arquivo original em modo não destrutivo.

Serão examinadas as seguintes dimensões:

1. DQ01 — Valores vazios ou compostos apenas por espaços.
2. DQ02 — Possíveis marcadores textuais de ausência.
3. DQ03 — Registros integralmente duplicados.
4. DQ04 — Repetições exatas dos identificadores.
5. DQ05 — Formatos aparentes dos atributos numéricos.

A verificação compreenderá todos os registros da base.

Antes da leitura, serão conferidos o tamanho do arquivo
e seu hash SHA-256 em relação à referência registrada.

### Critérios de validação

- Integridade SHA-256 confirmada.
- Dimensões compatíveis com o diagnóstico estrutural.
- Resultados de DQ01 a DQ05 apresentados.
- Ausência de operações de escrita ou tratamento.
- Preservação do arquivo CSV original.

### Limitações

Identificadores concatenados serão examinados como
valores textuais integrais, sem desmembramento.

A repetição de um identificador não será considerada
automaticamente uma inconsistência.

Os formatos dos atributos numéricos serão classificados
sem conversão de valores.

A interpretação dos resultados dependerá das
características efetivamente observadas.


```python

# FIAP - TECH CHALLENGE FASE 04
# Grupo 165
# Procedimento 3.4 - Diagnostico preliminar de qualidade
# SOMENTE LEITURA: nenhum dado sera tratado ou gravado.

from pathlib import Path
from collections import Counter
import csv
import hashlib
import re

# 1. Localizacao da base original
arquivo = Path(
    "/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON/"
    "01_DADOS_ORIGINAIS/01.amazon.csv"
)

hash_referencia = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

tamanho_referencia = 4744481

if not arquivo.is_file():
    raise FileNotFoundError(
        f"Arquivo nao encontrado: {arquivo}"
    )

# 2. Integridade: verificar bytes e SHA-256
tamanho_atual = arquivo.stat().st_size

if tamanho_atual != tamanho_referencia:
    raise ValueError(
        "Tamanho divergente. Diagnostico interrompido."
    )

sha256 = hashlib.sha256()

with arquivo.open("rb") as f:
    for bloco in iter(lambda: f.read(1024 * 1024), b""):
        sha256.update(bloco)

hash_atual = sha256.hexdigest().upper()

if hash_atual != hash_referencia:
    raise ValueError(
        "SHA-256 divergente. Diagnostico interrompido."
    )

print("=== CONTROLE DE INTEGRIDADE ===")
print("Arquivo:", arquivo.name)
print("Tamanho em bytes:", tamanho_atual)
print("SHA-256: CONFERIDO")

# 3. Leitura textual do CSV, sem conversoes
with arquivo.open(
    "r", encoding="utf-8-sig", newline=""
) as f:
    leitor = csv.reader(f)
    colunas = next(leitor)
    registros = [tuple(linha) for linha in leitor]

total = len(registros)
n_colunas = len(colunas)

if total != 1465 or n_colunas != 16:
    raise ValueError(
        "Dimensoes divergentes do diagnostico 3.3."
    )

if any(len(linha) != n_colunas for linha in registros):
    raise ValueError(
        "Encontrados registros estruturalmente irregulares."
    )

print("\n=== DIMENSOES ===")
print("Registros:", total)
print("Colunas:", n_colunas)

# 4. DQ01 e DQ02 - Ausencias e marcadores potenciais
marcadores = {"null", "none", "nan", "n/a", "na"}

print("\n=== DQ01/DQ02: AUSENCIAS ===")

for indice, coluna in enumerate(colunas):
    valores = [linha[indice] for linha in registros]

    vazios = sum(
        valor.strip() == ""
        for valor in valores
    )

    marcadores_potenciais = Counter(
        valor.strip().lower()
        for valor in valores
        if valor.strip().lower() in marcadores
    )

    percentual_vazios = 100 * vazios / total

    print(
        f"{coluna}: "
        f"vazios={vazios} ({percentual_vazios:.2f}%) | "
        f"marcadores={dict(marcadores_potenciais)}"
    )

# 5. DQ03 - Duplicidades integrais
frequencias = Counter(registros)

grupos_duplicados = sum(
    freq > 1
    for freq in frequencias.values()
)

registros_excedentes = sum(
    freq - 1
    for freq in frequencias.values()
    if freq > 1
)

print("\n=== DQ03: DUPLICIDADES INTEGRAIS ===")
print("Grupos de registros repetidos:", grupos_duplicados)
print("Registros excedentes:", registros_excedentes)
print(
    "Percentual excedente:",
    f"{100 * registros_excedentes / total:.2f}%"
)

# 6. DQ04 - Unicidade dos identificadores
print("\n=== DQ04: IDENTIFICADORES ===")

for identificador in [
    "product_id", "user_id", "review_id"
]:
    if identificador not in colunas:
        print(f"{identificador}: coluna ausente")
        continue

    indice = colunas.index(identificador)

    valores = [
        linha[indice]
        for linha in registros
        if linha[indice].strip() != ""
    ]

    frequencias_id = Counter(valores)

    grupos_repetidos = sum(
        freq > 1
        for freq in frequencias_id.values()
    )

    excedentes = sum(
        freq - 1
        for freq in frequencias_id.values()
        if freq > 1
    )

    print(
        f"{identificador}: "
        f"preenchidos={len(valores)} | "
        f"distintos={len(frequencias_id)} | "
        f"grupos_repetidos={grupos_repetidos} | "
        f"ocorrencias_excedentes={excedentes}"
    )

# 7. DQ05 - Padroes dos campos numericos
# Classificacao textual, sem conversao numerica.
def classificar_formato(valor):
    texto = valor.strip()

    if not texto:
        return "Vazio"

    if texto.lower() in marcadores:
        return "Marcador potencial de ausencia"

    if texto.startswith("₹"):
        conteudo = texto[1:].strip()

        if re.fullmatch(
            r"\d+(?:,\d{2,3})*(?:\.\d+)?",
            conteudo
        ):
            return "Moeda em rupias (texto)"

        return "Moeda com formato diverso"

    if re.fullmatch(
        r"[+-]?\d+(?:\.\d+)?%",
        texto
    ):
        return "Percentual (texto)"

    if re.fullmatch(r"[+-]?\d+", texto):
        return "Inteiro aparente"

    if re.fullmatch(r"[+-]?\d+\.\d+", texto):
        return "Decimal aparente"

    if re.fullmatch(
        r"[+-]?\d{1,3}(?:,\d{2,3})+(?:\.\d+)?",
        texto
    ):
        return "Numero com separadores (texto)"

    return "Outro formato"

campos_numericos = [
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating",
    "rating_count"
]

print("\n=== DQ05: FORMATOS NUMERICOS ===")

for campo in campos_numericos:
    if campo not in colunas:
        print(f"\n{campo}: coluna ausente")
        continue

    indice = colunas.index(campo)
    valores = [linha[indice] for linha in registros]

    categorias = Counter(
        classificar_formato(valor)
        for valor in valores
    )

    print(f"\n{campo}")

    for categoria, quantidade in sorted(
        categorias.items()
    ):
        percentual = 100 * quantidade / total
        print(
            f"  {categoria}: "
            f"{quantidade} ({percentual:.2f}%)"
        )

    # Exemplos limitados aos formatos inesperados.
    # Nao sao exibidos identificadores pessoais.
    exemplos = list(dict.fromkeys(
        valor
        for valor in valores
        if classificar_formato(valor)
        in {"Outro formato", "Moeda com formato diverso"}
    ))[:5]

    if exemplos:
        print("  Exemplos para investigar:", exemplos)

print("\n=== FIM DO DIAGNOSTICO 3.4 ===")
print("Arquivo original: nenhuma alteracao realizada.")
print("Tratamentos: nenhum.")
```

## 3.5. Investigação da granularidade e das repetições de identificadores

### Objetivo

Investigar a unidade de observação do Amazon Sales Dataset
e compreender as repetições identificadas nos campos
`product_id`, `user_id` e `review_id`.

### Fundamentação metodológica

A identificação da granularidade constitui uma etapa
necessária à definição de indicadores analíticos.

A existência de identificadores repetidos não representa,
isoladamente, duplicidade indevida ou inconsistência.

Sua interpretação depende da estrutura dos registros,
da variação dos atributos e das relações entre os campos.

### Metodologia

O procedimento será realizado em Python, utilizando
exclusivamente a base original.

Serão investigados:

1. Frequências de repetição dos identificadores.
2. Variação de atributos em registros com o mesmo produto.
3. Repetições de combinações de identificadores.
4. Indícios de delimitadores em campos textuais.

Os campos serão examinados integralmente, sem
desmembramento, normalização ou conversão.

### Critérios de validação

- Integridade SHA-256 confirmada.
- Dimensões compatíveis com os diagnósticos anteriores.
- Resultados GR01 a GR04 apresentados.
- Nenhuma modificação da base original.
- Conclusões limitadas às evidências observadas.

### Limitações

A presença de vírgulas ou outros caracteres em
identificadores não comprova, isoladamente, a existência
de múltiplos identificadores concatenados.

Não serão realizadas inferências sobre a quantidade real
de avaliações individuais sem validação da estrutura
dos campos.

Este procedimento não contempla tratamento dos dados.


```python

# FIAP - TECH CHALLENGE FASE 04
# P3.5 - Granularidade e identificadores
# Somente leitura; nenhuma transformacao ou gravacao.

from pathlib import Path
from collections import Counter, defaultdict
import csv
import hashlib

arquivo = Path(
    "/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON/"
    "01_DADOS_ORIGINAIS/01.amazon.csv"
)

hash_referencia = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

# 1. Conferencia da integridade
if not arquivo.is_file():
    raise FileNotFoundError(f"Arquivo nao encontrado: {arquivo}")

if arquivo.stat().st_size != 4744481:
    raise ValueError("Tamanho do arquivo divergente.")

sha = hashlib.sha256()

with arquivo.open("rb") as f:
    for bloco in iter(lambda: f.read(1024 * 1024), b""):
        sha.update(bloco)

if sha.hexdigest().upper() != hash_referencia:
    raise ValueError("SHA-256 divergente. Execucao interrompida.")

print("=== CONTROLE DE INTEGRIDADE ===")
print("SHA-256: CONFERIDO")
print("Tamanho: 4744481 bytes")

# 2. Leitura sem conversao de tipos
with arquivo.open("r", encoding="utf-8-sig", newline="") as f:
    leitor = csv.DictReader(f)
    colunas = leitor.fieldnames
    registros = list(leitor)

if colunas is None:
    raise ValueError("Cabecalho nao identificado.")

if len(registros) != 1465 or len(colunas) != 16:
    raise ValueError("Dimensoes divergentes do P3.4.")

if any(
    None in registro or None in registro.values()
    for registro in registros
):
    raise ValueError("Estrutura de registro inesperada.")

identificadores = ["product_id", "user_id", "review_id"]

if not all(campo in colunas for campo in identificadores):
    raise ValueError("Identificadores esperados nao encontrados.")

print("\n=== DIMENSOES ===")
print("Registros:", len(registros))
print("Colunas:", len(colunas))

# GR01 - Distribuicao das repeticoes
print("\n=== GR01: REPETICOES DE IDENTIFICADORES ===")

for campo in identificadores:
    frequencias = Counter(
        registro[campo] for registro in registros
    )

    grupos_repetidos = sum(
        n > 1 for n in frequencias.values()
    )
    excedentes = sum(
        n - 1 for n in frequencias.values() if n > 1
    )

    distribuicao = Counter(
        n for n in frequencias.values() if n > 1
    )

    print(f"\n{campo}")
    print("Valores distintos:", len(frequencias))
    print("Grupos repetidos:", grupos_repetidos)
    print("Ocorrencias excedentes:", excedentes)
    print(
        "Distribuicao das repeticoes:",
        dict(sorted(distribuicao.items()))
    )

# GR02 - Variacao entre registros do mesmo produto
grupos_produto = defaultdict(list)

for registro in registros:
    grupos_produto[registro["product_id"]].append(registro)

produtos_repetidos = [
    grupo
    for grupo in grupos_produto.values()
    if len(grupo) > 1
]

atributos = [
    "product_name",
    "category",
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating",
    "rating_count",
    "user_id",
    "review_id"
]

print("\n=== GR02: ATRIBUTOS EM PRODUTOS REPETIDOS ===")
print("Grupos de produtos repetidos:", len(produtos_repetidos))

for atributo in atributos:
    if atributo not in colunas:
        print(f"{atributo}: coluna ausente")
        continue

    grupos_com_variacao = sum(
        len({registro[atributo] for registro in grupo}) > 1
        for grupo in produtos_repetidos
    )

    print(
        f"{atributo}: "
        f"{grupos_com_variacao} grupos com variacao"
    )

# GR03 - Combinacoes e relacionamentos
print("\n=== GR03: RELACOES ENTRE IDENTIFICADORES ===")

pares_produto_review = Counter(
    (r["product_id"], r["review_id"])
    for r in registros
)

pares_usuario_review = Counter(
    (r["user_id"], r["review_id"])
    for r in registros
)

def resumir_pares(nome, frequencias):
    repetidos = sum(
        n > 1 for n in frequencias.values()
    )
    excedentes = sum(
        n - 1 for n in frequencias.values()
        if n > 1
    )

    print(f"\n{nome}")
    print("Pares distintos:", len(frequencias))
    print("Pares com repeticao:", repetidos)
    print("Ocorrencias excedentes:", excedentes)

resumir_pares(
    "product_id + review_id",
    pares_produto_review
)

resumir_pares(
    "user_id + review_id",
    pares_usuario_review
)

usuarios_reviews = defaultdict(set)
reviews_usuarios = defaultdict(set)

for r in registros:
    usuarios_reviews[r["user_id"]].add(r["review_id"])
    reviews_usuarios[r["review_id"]].add(r["user_id"])

print(
    "\nuser_id associado a mais de um review_id:",
    sum(len(v) > 1 for v in usuarios_reviews.values())
)

print(
    "review_id associado a mais de um user_id:",
    sum(len(v) > 1 for v in reviews_usuarios.values())
)

# GR04 - Indicios de delimitadores
# Apenas presenca de caracteres; nenhum campo e separado.
print("\n=== GR04: CARACTERES EM IDENTIFICADORES ===")

for campo in ["user_id", "review_id"]:
    valores = [r[campo] for r in registros]

    com_virgula = sum("," in valor for valor in valores)
    com_barra = sum("|" in valor for valor in valores)

    print(f"\n{campo}")
    print("Registros contendo virgula:", com_virgula)
    print("Registros contendo barra vertical:", com_barra)

print("\n=== FIM DO PROCEDIMENTO 3.5 ===")
print("Identificadores preservados integralmente.")
print("Nenhuma transformacao ou gravacao realizada.")
```

A verificação de integridade confirmou o tamanho de

#### GR01 — Distribuição das repetições

Considerando os valores textuais completos:

Os resultados indicam correspondência um-para-um

O total de 1.361 pares distintos de produto e

A análise da presença de caracteres apresentou:

### Interpretação da granularidade

A unidade física observada corresponde a uma
linha do arquivo CSV contendo atributos de produto
e campos relacionados a usuários e avaliações.

O diagnóstico demonstrou que essa linha não
corresponde necessariamente a um produto único,
pois existem 1.351 identificadores distintos
distribuídos em 1.465 registros.

Também não foi possível estabelecer que cada
linha represente uma avaliação individual.

Dessa forma, a granularidade lógica e o significado
das repetições ainda dependem de investigação.

### Implicações para a análise de negócio

A definição dos indicadores deverá respeitar
a unidade de observação apropriada.

Em particular:

1. A contagem de linhas não deve ser apresentada
   automaticamente como quantidade de produtos únicos.

2. A quantidade de avaliações individuais não pode
   ser inferida diretamente da contagem de linhas
   ou da contagem de valores distintos dos campos
   potencialmente concatenados.

3. Diferenças em preços, descontos e `rating_count`
   exigem critérios explícitos antes de eventual
   consolidação por produto.

4. As repetições poderão afetar distribuições,
   estatísticas agregadas e procedimentos de
   treinamento e validação de modelos.

5. A base não permite inferir quantidade de vendas,
   faturamento ou evolução temporal apenas a partir
   dos campos examinados.

### Limitações metodológicas

- Os identificadores foram analisados como textos
  completos, sem desmembramento.
- A presença de vírgulas não valida o delimitador.
- Não foram examinadas individualmente as linhas
  dos produtos com atributos divergentes.
- Não foi determinada a causa das diferenças entre
  registros associados ao mesmo produto.
- Não foram realizadas agregações nem selecionadas
  regras para escolha de registros.
- Não foi estabelecida uma relação individual
  validada entre usuários, avaliações e produtos.

### Conclusão

O procedimento demonstrou que o conjunto de dados
contém repetições de identificadores de produtos,
usuários e avaliações, além de variações em alguns
atributos de produtos repetidos.

Foram identificados indícios de concatenação nos
campos `user_id` e `review_id`, mas sua estrutura
interna permanece não validada.

A granularidade lógica da base ainda não pode ser
considerada definitivamente estabelecida.

Nenhuma repetição foi automaticamente classificada
como erro, e nenhum tratamento foi realizado.

produto, preservando os dados originais.

## 3.6. Validação da estrutura dos campos potencialmente concatenados

### Objetivo

Investigar a estrutura dos campos `user_id`, `user_name`
e `review_id`, avaliando a hipótese de que a vírgula
seja utilizada para representar múltiplos elementos
em uma mesma célula do arquivo original.

### Fundamentação metodológica

O diagnóstico anterior identificou a presença de
vírgulas em 1.461 dos 1.465 registros, tanto em
`user_id` quanto em `review_id`.

Entretanto, a presença de um caractere separador
não comprova, isoladamente, sua função como
delimitador de identificadores individuais.

A validação da estrutura interna desses campos
é necessária antes de definir a granularidade
das informações relacionadas aos consumidores
e às avaliações.

### Metodologia

O procedimento será realizado em Python, com
leitura da base original e verificação prévia
de sua integridade por SHA-256.

Serão examinados:

1. DC01 — Frequências de vírgulas e quantidades
   de elementos aparentes.
2. DC02 — Compatibilidade quantitativa entre
   `user_id`, `user_name` e `review_id`.
3. DC03 — Elementos aparentes vazios e espaços
   nas extremidades.
4. DC04 — Registros excepcionais e consistência
   dos padrões observados.

A separação por vírgula, quando utilizada no
diagnóstico, ocorrerá apenas temporariamente
em memória, sem gerar arquivos derivados
ou modificar o CSV original.

### Critérios de validação

- Integridade SHA-256 confirmada.
- Dimensões compatíveis com os diagnósticos anteriores.
- Todos os 1.465 registros examinados.
- Resultados DC01–DC04 apresentados.
- Exceções identificadas sem exposição de dados pessoais.
- Nenhuma transformação persistida.

### Limitações

A correspondência das quantidades de elementos
aparentes não comprova o alinhamento individual
entre nomes, usuários e avaliações.

Vírgulas em nomes podem integrar o próprio texto,
sem funcionar como delimitadores.

O procedimento não estabelecerá, isoladamente,
a quantidade definitiva de usuários ou de
avaliações individuais.

Nenhum tratamento será realizado.


```python

# FIAP - TECH CHALLENGE FASE 04
# Procedimento P3.6 - Estrutura de campos potencialmente concatenados
# SOMENTE LEITURA. Nenhum arquivo sera criado ou alterado.

from pathlib import Path
from collections import Counter
import csv
import hashlib

# 1. Fonte original e referencia de integridade

arquivo = Path(
    "/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON/"
    "01_DADOS_ORIGINAIS/01.amazon.csv"
)

hash_referencia = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

if not arquivo.is_file():
    raise FileNotFoundError(f"Arquivo nao encontrado: {arquivo}")

tamanho = arquivo.stat().st_size

if tamanho != 4744481:
    raise ValueError("Tamanho divergente. Diagnostico interrompido.")

sha256 = hashlib.sha256()

with arquivo.open("rb") as f:
    for bloco in iter(lambda: f.read(1024 * 1024), b""):
        sha256.update(bloco)

if sha256.hexdigest().upper() != hash_referencia:
    raise ValueError("SHA-256 divergente. Diagnostico interrompido.")

print("=== CONTROLE DE INTEGRIDADE ===")
print("Arquivo:", arquivo.name)
print("Tamanho:", tamanho, "bytes")
print("SHA-256: CONFERIDO")

# 2. Leitura textual da base, sem alteracao

with arquivo.open(
    "r", encoding="utf-8-sig", newline=""
) as f:
    leitor = csv.DictReader(f)
    colunas = leitor.fieldnames
    registros = list(leitor)

if colunas is None:
    raise ValueError("Cabecalho nao encontrado.")

if len(registros) != 1465 or len(colunas) != 16:
    raise ValueError("Dimensoes divergentes dos diagnosticos anteriores.")

if len(set(colunas)) != len(colunas):
    raise ValueError("Cabecalhos duplicados.")

if any(
    None in registro or None in registro.values()
    for registro in registros
):
    raise ValueError("Registro com estrutura inesperada.")

campos = ["user_id", "user_name", "review_id"]

if not all(campo in colunas for campo in campos):
    raise ValueError("Campos necessarios nao encontrados.")

print("\n=== DIMENSOES ===")
print("Registros:", len(registros))
print("Colunas:", len(colunas))

# 3. Caracterizacao somente em memoria.
# O split nao produz novas colunas ou arquivos.

def examinar_campo(texto):
    if texto.strip() == "":
        return {
            "virgulas": 0,
            "elementos": 0,
            "vazios": 0,
            "com_espacos": 0,
        }

    partes = texto.split(",")

    return {
        "virgulas": texto.count(","),
        "elementos": len(partes),
        "vazios": sum(
            parte.strip() == ""
            for parte in partes
        ),
        "com_espacos": sum(
            parte != parte.strip()
            for parte in partes
        ),
    }

diagnosticos = [
    {
        campo: examinar_campo(registro[campo])
        for campo in campos
    }
    for registro in registros
]

# DC01 - Distribuicoes por campo

print("\n=== DC01: ESTRUTURA APARENTE DOS CAMPOS ===")

for campo in campos:
    virgulas = Counter(
        linha[campo]["virgulas"]
        for linha in diagnosticos
    )

    elementos = Counter(
        linha[campo]["elementos"]
        for linha in diagnosticos
    )

    registros_com_virgula = sum(
        linha[campo]["virgulas"] > 0
        for linha in diagnosticos
    )

    print(f"\nCampo: {campo}")
    print("Registros com virgula:", registros_com_virgula)
    print(
        "Percentual:",
        f"{100 * registros_com_virgula / len(registros):.2f}%"
    )
    print(
        "Distribuicao de elementos aparentes:",
        dict(sorted(elementos.items()))
    )
    print(
        "Distribuicao de quantidades de virgulas:",
        dict(sorted(virgulas.items()))
    )

# DC02 - Compatibilidade quantitativa

print("\n=== DC02: COMPATIBILIDADE ENTRE CAMPOS ===")

pares = [
    ("user_id", "review_id"),
    ("user_id", "user_name"),
    ("review_id", "user_name"),
]

for campo_a, campo_b in pares:
    compativeis = sum(
        linha[campo_a]["elementos"] == linha[campo_b]["elementos"]
        for linha in diagnosticos
    )

    divergentes = len(registros) - compativeis

    print(f"\n{campo_a} x {campo_b}")
    print("Registros com contagens iguais:", compativeis)
    print("Registros com contagens diferentes:", divergentes)
    print(
        "Compatibilidade quantitativa:",
        f"{100 * compativeis / len(registros):.2f}%"
    )

# DC03 - Elementos vazios e espacos nas extremidades

print("\n=== DC03: EXCECOES DE FORMATACAO ===")

for campo in campos:
    registros_com_vazios = sum(
        linha[campo]["vazios"] > 0
        for linha in diagnosticos
    )

    total_elementos_vazios = sum(
        linha[campo]["vazios"]
        for linha in diagnosticos
    )

    registros_com_espacos = sum(
        linha[campo]["com_espacos"] > 0
        for linha in diagnosticos
    )

    print(f"\nCampo: {campo}")
    print(
        "Registros com elementos aparentes vazios:",
        registros_com_vazios
    )
    print(
        "Total de elementos aparentes vazios:",
        total_elementos_vazios
    )
    print(
        "Registros com espacos nas extremidades de elementos:",
        registros_com_espacos
    )

# DC04 - Posicoes de registros excepcionais
# Sao exibidos somente numeros de posicao, sem dados pessoais.

print("\n=== DC04: REGISTROS EXCEPCIONAIS ===")

divergencias = []

for posicao, linha in enumerate(diagnosticos, start=1):
    contagens = {
        campo: linha[campo]["elementos"]
        for campo in campos
    }

    if len(set(contagens.values())) > 1:
        divergencias.append((posicao, contagens))

print(
    "Registros com divergencia entre os tres campos:",
    len(divergencias)
)

print("Primeiras 15 divergencias (posicao e contagens):")

for posicao, contagens in divergencias[:15]:
    print(
        f"Registro {posicao}: "
        f"user_id={contagens['user_id']}, "
        f"user_name={contagens['user_name']}, "
        f"review_id={contagens['review_id']}"
    )

sem_virgula_ids = [
    posicao
    for posicao, linha in enumerate(diagnosticos, start=1)
    if (
        linha["user_id"]["virgulas"] == 0
        and linha["review_id"]["virgulas"] == 0
    )
]

print(
    "\nRegistros sem virgula em ambos os identificadores:",
    len(sem_virgula_ids)
)
print(
    "Primeiras 15 posicoes:",
    sem_virgula_ids[:15]
)

padroes = Counter(
    (
        linha["user_id"]["elementos"],
        linha["user_name"]["elementos"],
        linha["review_id"]["elementos"],
    )
    for linha in diagnosticos
)

print("\nPadroes de contagens mais frequentes")
print("Ordem: user_id, user_name, review_id")

for padrao, quantidade in padroes.most_common(15):
    print(f"{padrao}: {quantidade} registros")

print("\n=== FIM DO PROCEDIMENTO 3.6 ===")
print("Campos originais preservados.")
print("Nenhum arquivo derivado criado.")
print("Nenhuma transformacao persistida.")
```

## 3.7. Investigação direcionada das divergências em nomes

### Objetivo

Examinar as 12 divergências de contagem identificadas
entre `user_name`, `user_id` e `review_id`, investigando
a hipótese de vírgulas internas nos nomes.

### Metodologia

A análise utiliza a base original, preservada e
verificada por SHA-256.

Serão realizadas comparações temporárias, em memória,
entre os elementos aparentes de `user_name` e as
associações observadas nas linhas com contagens
compatíveis.

O procedimento não altera os dados e não apresenta
nomes ou identificadores individuais na saída.

### Limitações

As correspondências utilizadas são evidências
exploratórias, não uma validação definitiva
do alinhamento individual dos campos.

Não serão executadas correções, separações
persistentes ou exclusões de registros.


```python

# FIAP - Tech Challenge Fase 04
# P3.7 - Investigacao das divergencias em user_name
# Somente leitura. Nenhum arquivo sera modificado.

from pathlib import Path
from collections import defaultdict, Counter
import hashlib
import csv

arquivo = Path(
    "/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON/"
    "01_DADOS_ORIGINAIS/01.amazon.csv"
)

hash_esperado = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

# 1. Integridade
if not arquivo.is_file():
    raise FileNotFoundError(arquivo)

if arquivo.stat().st_size != 4744481:
    raise ValueError("Tamanho divergente.")

sha = hashlib.sha256()
with arquivo.open("rb") as f:
    for bloco in iter(lambda: f.read(1024 * 1024), b""):
        sha.update(bloco)

if sha.hexdigest().upper() != hash_esperado:
    raise ValueError("SHA-256 divergente.")

# 2. Leitura textual
with arquivo.open("r", encoding="utf-8-sig", newline="") as f:
    leitor = csv.DictReader(f)
    colunas = leitor.fieldnames
    registros = list(leitor)

if colunas is None or len(colunas) != 16 or len(registros) != 1465:
    raise ValueError("Dimensoes inesperadas.")

campos = {"user_id", "user_name", "review_id"}
if not campos.issubset(colunas):
    raise ValueError("Campos esperados ausentes.")

if any(
    None in r or None in r.values()
    for r in registros
):
    raise ValueError("Estrutura CSV inesperada.")

print("=== INTEGRIDADE ===")
print("SHA-256: CONFERIDO")
print("Registros:", len(registros))

# 3. Separacao temporaria, exclusivamente em memoria
def elementos(registro, campo):
    return registro[campo].split(",")

regulares = []
excecoes = []

for posicao, registro in enumerate(registros, start=1):
    ids = elementos(registro, "user_id")
    nomes = elementos(registro, "user_name")
    reviews = elementos(registro, "review_id")

    if len(ids) == len(nomes) == len(reviews):
        regulares.append((posicao, ids, nomes, reviews))
    else:
        excecoes.append((posicao, ids, nomes, reviews))

print("\n=== P37-01: CLASSIFICACAO ===")
print("Linhas com contagens compativeis:", len(regulares))
print("Linhas divergentes:", len(excecoes))

padroes = Counter(
    (len(ids), len(nomes), len(reviews))
    for _, ids, nomes, reviews in excecoes
)

print("Padroes de divergencia:", dict(padroes))

# 4. Referencias exploratorias:
# associações observadas nas linhas de contagem compatível.
referencias = defaultdict(set)

for _, ids, nomes, reviews in regulares:
    for uid, nome in zip(ids, nomes):
        if uid.strip() and nome.strip():
            referencias[uid].add(nome)

# 5. Investigar se unir dois fragmentos adjacentes
# de nomes restabelece a contagem esperada.
# Nenhuma juncao sera persistida ou gravada.

print("\n=== P37-02: HIPOTESE DE VIRGULA INTERNA ===")
print("Posicoes de juncao sao contadas a partir de 1.")

resultados = []

for posicao, ids, nomes, reviews in excecoes:
    alternativas = []

    if len(ids) == len(reviews) and len(nomes) == len(ids) + 1:
        for j in range(len(nomes) - 1):
            candidato = (
                nomes[:j]
                + [nomes[j] + "," + nomes[j + 1]]
                + nomes[j + 2:]
            )

            cobertura = sum(
                uid in referencias for uid in ids
            )

            acertos = sum(
                uid in referencias
                and nome in referencias[uid]
                for uid, nome in zip(ids, candidato)
            )

            alternativas.append(
                (j + 1, cobertura, acertos)
            )

    if alternativas:
        maior_acerto = max(a[2] for a in alternativas)
        melhores = [
            a for a in alternativas
            if a[2] == maior_acerto
        ]
        posicoes = [a[0] for a in melhores]
        cobertura = melhores[0][1]
    else:
        maior_acerto = 0
        cobertura = 0
        posicoes = []

    resultados.append(
        (posicao, cobertura, maior_acerto, len(posicoes))
    )

    print(
        f"Registro {posicao}: "
        f"referencias_disponiveis={cobertura}/{len(ids)} | "
        f"melhor_correspondencia={maior_acerto}/{len(ids)} | "
        f"posicoes_candidatas={posicoes}"
    )

# 6. Consolidacao, sem revelar dados identificaveis
print("\n=== P37-03: CONSOLIDACAO ===")

print(
    "Excecoes com uma unica posicao candidata:",
    sum(qtd == 1 for _, _, _, qtd in resultados)
)

print(
    "Excecoes com multiplas posicoes candidatas:",
    sum(qtd > 1 for _, _, _, qtd in resultados)
)

print(
    "Excecoes sem referencias comparativas:",
    sum(cobertura == 0 for _, cobertura, _, _ in resultados)
)

# Contagem de sequencias distintas, sem divulgar valores.
sequencias_distintas = {
    (
        tuple(ids),
        tuple(nomes),
        tuple(reviews)
    )
    for _, ids, nomes, reviews in excecoes
}

print(
    "Sequencias distintas entre as excecoes:",
    len(sequencias_distintas)
)

print("\n=== FIM DO P3.7 ===")
print("Nenhum identificador ou nome individual exibido.")
print("Nenhuma transformacao persistida.")
```

## 4.1. Tratamento T4.1 — Rating inválido

### Problema identificado

O diagnóstico P3.4 identificou um valor `|`
na coluna `rating`, incompatível com sua
interpretação como avaliação numérica.

A ocorrência corresponde a 1 dos 1.465
registros (0,07%).

### Decisão de tratamento

Foi aprovada a alternativa de criar uma
coluna derivada `rating_numerico`, preservando
integralmente a coluna original `rating`.

Os valores numéricos serão convertidos,
enquanto o único valor `|` será representado
como ausente na coluna derivada.

Não haverá exclusão de registros nem
imputação de avaliações.

### Método

O procedimento será realizado em Python,
utilizando pandas, com leitura do CSV original,
verificação prévia do SHA-256 e geração de
um novo arquivo no diretório de dados tratados.

O arquivo original não será sobrescrito.

### Critérios de validação

1. Integridade da fonte original confirmada.
2. Total de 1.465 registros preservado.
3. As 16 colunas originais preservadas.
4. Uma nova coluna `rating_numerico` criada.
5. Exatamente 1.464 valores numéricos válidos.
6. Exatamente um valor ausente na coluna derivada.
7. Nenhuma exclusão de registro.
8. Arquivo derivado salvo e reaberto com sucesso.

### Limitações

A transformação corrige a representação numérica
do atributo, mas não verifica a plausibilidade
estatística ou a consistência de todos os ratings.

A avaliação de valores extremos e das faixas
admissíveis será realizada separadamente,
caso necessária.

O tratamento não altera os campos de preços,
descontos, contagens de avaliações ou
identificadores.

```python

# FIAP - TECH CHALLENGE FASE 04
# Grupo 165
# T4.1 - Tratamento aprovado do rating invalido

from pathlib import Path
import hashlib
import re
import pandas as pd
import numpy as np

# 1. Caminhos
raiz = Path("/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON")

origem = raiz / "01_DADOS_ORIGINAIS" / "01.amazon.csv"

destino = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_01_rating_v01.csv"
)

hash_referencia = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

def calcular_sha256(caminho):
    sha = hashlib.sha256()
    with caminho.open("rb") as arquivo:
        for bloco in iter(
            lambda: arquivo.read(1024 * 1024), b""
        ):
            sha.update(bloco)
    return sha.hexdigest().upper()

# 2. Controle antes do tratamento
if not origem.is_file():
    raise FileNotFoundError(f"Fonte nao encontrada: {origem}")

if not destino.parent.is_dir():
    raise FileNotFoundError(
        f"Diretorio de destino nao encontrado: {destino.parent}"
    )

if destino.exists():
    raise FileExistsError(
        "Arquivo derivado ja existe. "
        "Execucao interrompida para evitar sobrescrita."
    )

if origem.stat().st_size != 4744481:
    raise ValueError("Tamanho da base original divergente.")

if calcular_sha256(origem) != hash_referencia:
    raise ValueError("SHA-256 original divergente.")

print("=== T4.1 - CONTROLE INICIAL ===")
print("SHA-256 original: CONFERIDO")

# 3. Leitura integral como texto
# Evita conversoes involuntarias nos demais campos.
base = pd.read_csv(
    origem,
    dtype=str,
    keep_default_na=False,
    encoding="utf-8-sig"
)

if base.shape != (1465, 16):
    raise ValueError(
        f"Dimensoes inesperadas: {base.shape}"
    )

if "rating" not in base.columns:
    raise ValueError("Coluna rating nao encontrada.")

colunas_originais = base.columns.tolist()

# 4. Validacao da anomalia conhecida
padrao_numerico = re.compile(r"[+-]?\d+(?:\.\d+)?")

valido = base["rating"].map(
    lambda valor: bool(
        padrao_numerico.fullmatch(valor.strip())
    )
)

invalidos = base.loc[~valido, "rating"]

if len(invalidos) != 1 or invalidos.iloc[0] != "|":
    raise ValueError(
        "Anomalias diferentes das aprovadas. "
        "Tratamento interrompido."
    )

# 5. Transformacao aprovada
# Cria apenas uma coluna derivada.
tratada = base.copy()

tratada["rating_numerico"] = pd.to_numeric(
    tratada["rating"],
    errors="coerce"
)

# 6. Validacoes antes de salvar
assert tratada.shape == (1465, 17)
assert tratada["rating_numerico"].notna().sum() == 1464
assert tratada["rating_numerico"].isna().sum() == 1

pd.testing.assert_frame_equal(
    base,
    tratada[colunas_originais],
    check_dtype=True
)

print("\n=== T4.1 - RESULTADOS EM MEMORIA ===")
print("Registros antes:", len(base))
print("Registros depois:", len(tratada))
print("Colunas antes:", len(base.columns))
print("Colunas depois:", len(tratada.columns))
print(
    "Ratings numericos validos:",
    tratada["rating_numerico"].notna().sum()
)
print(
    "Ratings ausentes na coluna derivada:",
    tratada["rating_numerico"].isna().sum()
)
print("Registros excluidos:", len(base) - len(tratada))

# 7. Salvamento sem sobrescrita
# Modo 'x': cria apenas se o arquivo nao existir.
with destino.open(
    "x", encoding="utf-8", newline=""
) as arquivo_saida:
    tratada.to_csv(arquivo_saida, index=False)

# 8. Reabertura e testes da versao salva
reaberta = pd.read_csv(
    destino,
    dtype=str,
    keep_default_na=False,
    encoding="utf-8"
)

if reaberta.shape != (1465, 17):
    raise AssertionError("Dimensoes da versao salva divergentes.")

pd.testing.assert_frame_equal(
    base,
    reaberta[colunas_originais],
    check_dtype=True
)

assert (
    reaberta["rating_numerico"].eq("").sum() == 1
)

assert (
    reaberta.loc[
        reaberta["rating_numerico"].eq(""), "rating"
    ].tolist() == ["|"]
)

numericos_reabertos = pd.to_numeric(
    reaberta["rating_numerico"],
    errors="coerce"
)

assert np.allclose(
    tratada["rating_numerico"].to_numpy(dtype=float),
    numericos_reabertos.to_numpy(dtype=float),
    equal_nan=True
)

# 9. Confirmacao final da fonte original
if calcular_sha256(origem) != hash_referencia:
    raise AssertionError("Integridade da fonte alterada.")

print("\n=== T4.1 - VALIDACAO FINAL ===")
print("Arquivo derivado:", destino.name)
print("Registros preservados:", len(reaberta))
print("Colunas preservadas:", len(colunas_originais))
print("Nova coluna: rating_numerico")
print("Ratings validos:", numericos_reabertos.notna().sum())
print("Ratings ausentes:", numericos_reabertos.isna().sum())
print("SHA-256 original: PRESERVADO")
print("SHA-256 derivado:", calcular_sha256(destino))
print("Reabertura e comparacao: APROVADAS")
print("=== FIM DO TRATAMENTO T4.1 ===")
```

## 4.2. Tratamento T4.2 — Valores ausentes em rating_count

### Objetivo

Representar adequadamente os valores ausentes no atributo
`rating_count`, preservando as informações originalmente
disponíveis e evitando a atribuição artificial de contagens.

A ausência de informação não permite concluir que

### Decisão metodológica

Foi aprovada a criação da variável derivada
`rating_count_tratado`, com representação dos
dois valores vazios como ausentes.

Os 1.463 valores preenchidos permanecerão
textualmente inalterados.

A coluna original `rating_count` será preservada.

Não haverá imputação, exclusão de registros
ou conversão numérica neste procedimento.

### Fonte de entrada

Arquivo derivado do tratamento T4.1:

`02_DADOS_TRATADOS/amazon_t4_01_rating_v01.csv`

A base original permanece em:

`01_DADOS_ORIGINAIS/01.amazon.csv`

### Metodologia

O procedimento será implementado em Python,
utilizando pandas e controles de integridade SHA-256.

Serão realizadas:

1. Verificação da integridade da base original.
2. Verificação da integridade da versão T4.1.
3. Identificação dos dois valores vazios.
4. Criação de `rating_count_tratado`.
5. Representação dos vazios como valores ausentes.
6. Preservação dos valores preenchidos.
7. Salvamento em nova versão independente.
8. Reabertura e validação do arquivo derivado.

### Critérios de validação

- 1.465 registros preservados.
- 17 colunas anteriores preservadas.
- 18 colunas na versão derivada.
- Exatamente dois valores ausentes na nova coluna.
- 1.463 valores preenchidos sem modificações.
- Nenhuma exclusão ou imputação.
- Integridade dos arquivos de origem preservada.
- Arquivo derivado salvo e reaberto com sucesso.

### Limitações

A variável derivada permanecerá textual nesta etapa.

A conversão de contagens com separadores de milhares
para valores inteiros dependerá de tratamento específico.

Os valores ausentes serão serializados como campos
vazios no CSV e deverão ser reconhecidos como nulos
na leitura posterior e no Power BI.


```python

# FIAP - TECH CHALLENGE FASE 04
# Grupo 165
# T4.2 - Tratamento aprovado de ausencias em rating_count

from pathlib import Path
import hashlib
import pandas as pd

# 1. Caminhos e referencias de integridade

raiz = Path("/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON")

original = raiz / "01_DADOS_ORIGINAIS" / "01.amazon.csv"

entrada = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_01_rating_v01.csv"
)

saida = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_02_rating_count_v01.csv"
)

hash_original = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

hash_t41 = (
    "283410DA811C5C29B1BF37AAFBBDAE2"
    "7BF5CDD153E5F9F87058E401BED78C6DE"
)

def calcular_sha256(caminho):
    sha = hashlib.sha256()

    with caminho.open("rb") as f:
        for bloco in iter(
            lambda: f.read(1024 * 1024), b""
        ):
            sha.update(bloco)

    return sha.hexdigest().upper()

# 2. Verificacoes antes do tratamento

for caminho in [original, entrada]:
    if not caminho.is_file():
        raise FileNotFoundError(
            f"Arquivo nao encontrado: {caminho}"
        )

if saida.exists():
    raise FileExistsError(
        "Arquivo de saida ja existe. "
        "Execucao interrompida para evitar sobrescrita."
    )

if calcular_sha256(original) != hash_original:
    raise ValueError("SHA-256 da base original divergente.")

if calcular_sha256(entrada) != hash_t41:
    raise ValueError("SHA-256 do T4.1 divergente.")

print("=== T4.2 - CONTROLE INICIAL ===")
print("SHA-256 original: CONFERIDO")
print("SHA-256 T4.1: CONFERIDO")

# 3. Leitura textual da versao T4.1

base = pd.read_csv(
    entrada,
    dtype=str,
    keep_default_na=False,
    encoding="utf-8"
)

if base.shape != (1465, 17):
    raise ValueError(
        f"Dimensoes inesperadas: {base.shape}"
    )

if "rating_count" not in base.columns:
    raise ValueError("Coluna rating_count ausente.")

if "rating_numerico" not in base.columns:
    raise ValueError("Coluna do T4.1 ausente.")

if "rating_count_tratado" in base.columns:
    raise ValueError("Coluna derivada ja existente.")

colunas_anteriores = base.columns.tolist()

# 4. Validacao da ausencia previamente diagnosticada

vazios = base["rating_count"].eq("")

if int(vazios.sum()) != 2:
    raise ValueError(
        "Quantidade de vazios diferente do P3.4. "
        "Tratamento interrompido."
    )

if base["rating_numerico"].eq("").sum() != 1:
    raise ValueError(
        "Resultado do T4.1 diferente do esperado."
    )

# 5. Tratamento aprovado

tratada = base.copy(deep=True)

tratada["rating_count_tratado"] = (
    base["rating_count"]
    .astype("string")
    .mask(vazios, pd.NA)
)

# 6. Testes antes do salvamento

if tratada.shape != (1465, 18):
    raise AssertionError("Dimensoes incorretas.")

if tratada["rating_count_tratado"].isna().sum() != 2:
    raise AssertionError("Quantidade de nulos incorreta.")

if tratada["rating_count_tratado"].notna().sum() != 1463:
    raise AssertionError("Valores preenchidos divergentes.")

pd.testing.assert_frame_equal(
    base,
    tratada[colunas_anteriores],
    check_dtype=True
)

preenchidos = ~vazios

if not (
    tratada.loc[preenchidos, "rating_count_tratado"]
    .astype(str)
    .equals(base.loc[preenchidos, "rating_count"])
):
    raise AssertionError(
        "Valores preenchidos foram modificados."
    )

print("\n=== T4.2 - RESULTADOS EM MEMORIA ===")
print("Registros antes:", len(base))
print("Registros depois:", len(tratada))
print("Colunas antes:", len(base.columns))
print("Colunas depois:", len(tratada.columns))
print("Vazios originais:", int(vazios.sum()))
print(
    "Nulos em rating_count_tratado:",
    int(tratada["rating_count_tratado"].isna().sum())
)
print("Valores preenchidos preservados: 1463")
print("Registros excluidos: 0")
print("Imputacoes realizadas: 0")

# 7. Salvamento de nova versao, sem sobrescrita

with saida.open(
    "x", encoding="utf-8", newline=""
) as f:
    tratada.to_csv(
        f,
        index=False,
        na_rep=""
    )

# 8. Reabertura preservando literalmente as colunas
# anteriores, inclusive os campos vazios do T4.1.

reaberta = pd.read_csv(
    saida,
    dtype=str,
    keep_default_na=False,
    encoding="utf-8"
)

if reaberta.shape != (1465, 18):
    raise AssertionError("Dimensoes salvas divergentes.")

pd.testing.assert_frame_equal(
    base,
    reaberta[colunas_anteriores],
    check_dtype=True
)

if reaberta["rating_count_tratado"].eq("").sum() != 2:
    raise AssertionError("Ausencias salvas divergentes.")

if not (
    reaberta.loc[preenchidos, "rating_count_tratado"]
    .equals(base.loc[preenchidos, "rating_count"])
):
    raise AssertionError(
        "Valores preenchidos divergentes apos reabertura."
    )

# 9. Teste de reconhecimento dos nulos na releitura

reaberta_com_nulos = pd.read_csv(
    saida,
    dtype={"rating_count_tratado": "string"},
    keep_default_na=False,
    na_values={"rating_count_tratado": [""]},
    encoding="utf-8"
)

if (
    reaberta_com_nulos["rating_count_tratado"]
    .isna().sum() != 2
):
    raise AssertionError(
        "Falha na interpretacao dos valores ausentes."
    )

# 10. Integridade apos o tratamento

if calcular_sha256(original) != hash_original:
    raise AssertionError("Base original alterada.")

if calcular_sha256(entrada) != hash_t41:
    raise AssertionError("Arquivo T4.1 alterado.")

print("\n=== T4.2 - VALIDACAO FINAL ===")
print("Arquivo derivado:", saida.name)
print("Registros preservados:", len(reaberta))
print("Colunas anteriores preservadas:", len(colunas_anteriores))
print("Nova coluna: rating_count_tratado")
print("Valores preenchidos: 1463")
print("Valores ausentes reconhecidos: 2")
print("SHA-256 original: PRESERVADO")
print("SHA-256 T4.1: PRESERVADO")
print("SHA-256 T4.2:", calcular_sha256(saida))
print("Reabertura e comparacao: APROVADAS")
print("Nenhuma conversao numerica realizada.")
print("=== FIM DO TRATAMENTO T4.2 ===")
```


### 4.2.1. Resultados, validação e discussão do tratamento T4.2

## 4.3. Tratamento T4.3 — Conversão numérica de rating_count

### Objetivo

Converter os valores preenchidos de `rating_count_tratado`
para uma variável numérica inteira e anulável, preservando
a informação original e os valores ausentes.

### Problema identificado

O diagnóstico P3.4 identificou:

- 326 registros com inteiros aparentes sem separadores;
- 1.137 registros com separadores numéricos;
- 2 registros com valores ausentes.

Os valores preenchidos ainda estavam armazenados como
texto, impedindo seu uso numérico direto.

### Decisão metodológica

Foi aprovada a criação da coluna `rating_count_numerico`,
com tipo inteiro anulável (`Int64`).

As regras estabelecidas foram:

1. Preservar todas as colunas existentes.
2. Validar os formatos numéricos antes da conversão.
3. Interpretar as vírgulas validadas como separadores
   de agrupamento numérico.
4. Converter somente os 1.463 valores preenchidos.
5. Manter os dois valores ausentes como nulos.
6. Não excluir, imputar, agregar ou alterar registros.

### Fonte e versionamento

**Entrada:**
`02_DADOS_TRATADOS/amazon_t4_02_rating_count_v01.csv`

**Saída prevista:**
`02_DADOS_TRATADOS/amazon_t4_03_rating_count_numerico_v01.csv`

### Metodologia

O tratamento será realizado em Python, utilizando
pandas, expressões regulares e controles SHA-256.

A conversão será interrompida caso o arquivo de entrada,
as dimensões ou os formatos encontrados sejam
incompatíveis com os critérios definidos.

O arquivo derivado será salvo em versão independente,
sem sobrescrever versões anteriores.

### Critérios de validação

- Preservação dos 1.465 registros.
- Preservação das 18 colunas anteriores.
- Criação de uma única coluna numérica.
- Conversão de 1.463 valores para inteiros.
- Manutenção de dois valores ausentes.
- Ausência de exclusões e imputações.
- Integridade dos arquivos anteriores preservada.
- Reabertura e comparação dos dados aprovadas.

### Limitações

A conversão não valida, isoladamente, a plausibilidade
das contagens ou sua relação com avaliações individuais.

A existência de produtos repetidos impede interpretar
automaticamente a soma de `rating_count_numerico`
como quantidade de avaliações únicas.

O procedimento não altera a granularidade dos registros.


```python

# FIAP - TECH CHALLENGE FASE 04
# T4.3 - Verificacao do arquivo preexistente
# DIAGNOSTICO: nenhuma escrita ou alteracao.

from pathlib import Path
import hashlib
import pandas as pd

raiz = Path("/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON")

original = raiz / "01_DADOS_ORIGINAIS" / "01.amazon.csv"

entrada = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_02_rating_count_v01.csv"
)

saida = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_03_rating_count_numerico_v01.csv"
)

def sha256_arquivo(caminho):
    sha = hashlib.sha256()
    with caminho.open("rb") as f:
        for bloco in iter(lambda: f.read(1024 * 1024), b""):
            sha.update(bloco)
    return sha.hexdigest().upper()

hash_original = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

hash_t42 = (
    "0AEF9520C8B3CC19DA22F79B6536B7E4"
    "A500CF103691ADFABAEACADD3C36C6A0"
)

print("=== VERIFICACAO DO T4.3 ===")

print("Arquivo existente:", saida.is_file())

if not saida.is_file():
    print("RESULTADO: arquivo T4.3 nao encontrado.")
else:
    print("Tamanho em bytes:", saida.stat().st_size)
    print("SHA-256 T4.3:", sha256_arquivo(saida))

    try:
        integridade_original = (
            original.is_file()
            and sha256_arquivo(original) == hash_original
        )

        integridade_t42 = (
            entrada.is_file()
            and sha256_arquivo(entrada) == hash_t42
        )

        print("SHA-256 original confere:", integridade_original)
        print("SHA-256 T4.2 confere:", integridade_t42)

        if not (integridade_original and integridade_t42):
            print("VALIDACAO INTERROMPIDA: verificar fontes.")
        else:
            base = pd.read_csv(
                entrada,
                dtype=str,
                keep_default_na=False,
                encoding="utf-8"
            )

            gerada = pd.read_csv(
                saida,
                dtype=str,
                keep_default_na=False,
                encoding="utf-8"
            )

            print("\n=== ESTRUTURA ===")
            print("Dimensoes T4.2:", base.shape)
            print("Dimensoes T4.3:", gerada.shape)

            estrutura_ok = (
                base.shape == (1465, 18)
                and gerada.shape == (1465, 19)
                and gerada.columns.tolist()
                == base.columns.tolist() + ["rating_count_numerico"]
            )

            print("Estrutura esperada:", estrutura_ok)

            if estrutura_ok:
                preservadas = base.equals(
                    gerada[base.columns]
                )

                print("18 colunas preservadas:", preservadas)

                nova = gerada["rating_count_numerico"]

                inteiros_validos = nova.map(
                    lambda x: x.isdecimal() if x else False
                )

                print(
                    "Valores numericos preenchidos:",
                    int(inteiros_validos.sum())
                )

                print(
                    "Valores vazios:",
                    int(nova.eq("").sum())
                )

                # Compara a conversao esperada com o CSV salvo.
                fonte = base["rating_count_tratado"]

                formato_fonte_ok = fonte.map(
                    lambda x: (
                        x.replace(",", "").isdecimal()
                        if x else True
                    )
                ).all()

                if formato_fonte_ok:
                    esperado = fonte.map(
                        lambda x: (
                            str(int(x.replace(",", "")))
                            if x else ""
                        )
                    )

                    valores_corretos = nova.equals(esperado)
                else:
                    valores_corretos = False

                print(
                    "Conversao corresponde a fonte:",
                    valores_corretos
                )

                aprovado = (
                    preservadas
                    and int(inteiros_validos.sum()) == 1463
                    and int(nova.eq("").sum()) == 2
                    and valores_corretos
                )

                print("\n=== RESULTADO ===")
                print("TESTES DE CONTEUDO APROVADOS:", aprovado)

            else:
                print(
                    "TESTES DE CONTEUDO APROVADOS: False"
                )

    except Exception as erro:
        print("\nFalha na verificacao:", type(erro).__name__)
        print("Detalhe:", str(erro))

print("\nNenhum arquivo foi alterado.")
print("=== FIM DA VERIFICACAO ===")
```

As 18 colunas anteriores foram preservadas.

O arquivo existente foi mantido sem sobrescrita
e seu hash SHA-256 foi registrado.

**Status:** Resultado do tratamento T4.3 validado.

**Ressalva de auditoria:** A criação original do
arquivo preexistente não foi integralmente
observada; a validação foi realizada por testes
posteriores sobre seu conteúdo.



## 4.4. Tratamento T4.4 — Conversão dos preços para rupias indianas

### Objetivo

Disponibilizar variáveis monetárias numéricas para
análise dos preços originais e dos preços com desconto
dos produtos do Amazon Sales Dataset.

### Problema identificado

O diagnóstico P3.4 identificou 1.465 valores em
formato monetário textual, com símbolo de rupia
indiana (₹), em cada um dos atributos:

- `discounted_price`;
- `actual_price`.

Não foram identificados valores vazios nessas colunas.

A representação textual exige conversão controlada
para utilização em operações quantitativas.

### Decisão metodológica

Foi aprovada a alternativa de preservar os campos
originais e criar duas variáveis derivadas:

- `discounted_price_inr`;
- `actual_price_inr`.

A conversão manterá a moeda original em rupias
indianas (INR), sem conversão cambial.

Serão utilizados valores do tipo `Decimal`,
preservando a precisão decimal necessária às
operações monetárias.

### Regras de tratamento

1. Preservar os 1.465 registros.
2. Preservar integralmente as 19 colunas anteriores.
3. Verificar a presença do símbolo ₹.
4. Validar os formatos numéricos e separadores.
5. Interpretar os separadores de agrupamento,
   sem modificar os campos originais.
6. Criar as duas colunas monetárias derivadas.
7. Registrar os valores com duas casas decimais.
8. Interromper a execução diante de formatos
   inesperados, sem correção automática.
9. Não excluir, imputar, agregar ou corrigir preços.

### Fonte e versionamento

**Entrada:**

`02_DADOS_TRATADOS/amazon_t4_03_rating_count_numerico_v01.csv`

**Saída prevista:**

`02_DADOS_TRATADOS/amazon_t4_04_precos_inr_v01.csv`

### Metodologia

O procedimento utilizará Python, pandas,
expressões regulares, Decimal e SHA-256.

A validação contemplará formatos monetários
sem agrupamento e formatos agrupados segundo
padrões ocidentais e indianos.

Os valores preenchidos serão interpretados
sem conversão cambial e sem utilização de
números de ponto flutuante.

### Critérios de validação

- Integridade dos arquivos anteriores confirmada.
- 1.465 registros preservados.
- 19 colunas anteriores preservadas.
- Duas novas variáveis monetárias criadas.
- 1.465 valores convertidos em cada variável.
- Ausência de exclusões ou imputações.
- Correspondência entre valores originais
  e representações monetárias derivadas.
- Arquivo salvo e reaberto com sucesso.

### Limitações

A conversão monetária não demonstra a
consistência comercial dos preços.

Não serão avaliados neste procedimento:

- Coerência entre preço original e descontado;
- Cálculo ou validação de descontos percentuais;
- Outliers e valores extremos;
- Variações entre registros do mesmo produto;
- Receita, faturamento ou volume de vendas.

O formato CSV não preserva o tipo Decimal.
A interpretação numérica deverá ser validada
na importação posterior para o Power BI.


```python

# FIAP - TECH CHALLENGE FASE 04
# Grupo 165
# T4.4 - Conversao monetaria aprovada para INR

from pathlib import Path
from decimal import Decimal, InvalidOperation
import hashlib
import re
import pandas as pd

# 1. Caminhos e referencias de integridade

raiz = Path("/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON")

original = raiz / "01_DADOS_ORIGINAIS" / "01.amazon.csv"

entrada = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_03_rating_count_numerico_v01.csv"
)

saida = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_04_precos_inr_v01.csv"
)

hash_original = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

hash_t43 = (
    "512FF5FE51998C313799D80F5B3D6A77"
    "84F2A74E657F057B26856E24DB614CBF"
)

def calcular_sha256(caminho):
    sha = hashlib.sha256()

    with caminho.open("rb") as f:
        for bloco in iter(
            lambda: f.read(1024 * 1024), b""
        ):
            sha.update(bloco)

    return sha.hexdigest().upper()

# 2. Verificacoes iniciais

for caminho in [original, entrada]:
    if not caminho.is_file():
        raise FileNotFoundError(
            f"Arquivo nao encontrado: {caminho}"
        )

if saida.exists():
    raise FileExistsError(
        "Arquivo T4.4 ja existe. "
        "Execucao interrompida para evitar sobrescrita."
    )

if original.stat().st_size != 4744481:
    raise ValueError("Tamanho da base original divergente.")

if calcular_sha256(original) != hash_original:
    raise ValueError("SHA-256 original divergente.")

if calcular_sha256(entrada) != hash_t43:
    raise ValueError("SHA-256 T4.3 divergente.")

print("=== T4.4 - CONTROLE INICIAL ===")
print("SHA-256 original: CONFERIDO")
print("SHA-256 T4.3: CONFERIDO")

# 3. Leitura como texto, preservando as colunas anteriores

base = pd.read_csv(
    entrada,
    dtype=str,
    keep_default_na=False,
    encoding="utf-8"
)

if base.shape != (1465, 19):
    raise ValueError(
        f"Dimensoes inesperadas: {base.shape}"
    )

campos = {
    "discounted_price": "discounted_price_inr",
    "actual_price": "actual_price_inr",
}

for campo_original, campo_novo in campos.items():
    if campo_original not in base.columns:
        raise ValueError(
            f"Coluna original ausente: {campo_original}"
        )

    if campo_novo in base.columns:
        raise ValueError(
            f"Coluna derivada ja existe: {campo_novo}"
        )

colunas_anteriores = base.columns.tolist()

# 4. Definicao dos formatos monetarios aceitos
# Exemplos aceitos:
# ₹999
# ₹1,999
# ₹12,34,567
# ₹1,299.50
# ₹129.99
#
# Formatos inesperados nao serao corrigidos
# automaticamente.

padrao_simples = re.compile(
    r"[0-9]+(?:\.[0-9]{1,2})?"
)

padrao_ocidental = re.compile(
    r"[0-9]{1,3}(?:,[0-9]{3})+(?:\.[0-9]{1,2})?"
)

padrao_indiano = re.compile(
    r"[0-9]{1,3}(?:,[0-9]{2})+,[0-9]{3}"
    r"(?:\.[0-9]{1,2})?"
)

centavo = Decimal("0.01")

def converter_preco(valor):
    texto = valor.strip()

    if not texto.startswith("₹"):
        return None, "Simbolo monetario ausente ou divergente"

    numero_texto = texto[1:].strip()

    formatos_validos = (
        padrao_simples.fullmatch(numero_texto)
        or padrao_ocidental.fullmatch(numero_texto)
        or padrao_indiano.fullmatch(numero_texto)
    )

    if not formatos_validos:
        return None, "Formato numerico inesperado"

    try:
        valor_decimal = Decimal(
            numero_texto.replace(",", "")
        )

        if not valor_decimal.is_finite():
            return None, "Valor nao finito"

        if valor_decimal < 0:
            return None, "Preco negativo"

        valor_decimal = valor_decimal.quantize(centavo)

        return valor_decimal, "Valido"

    except (InvalidOperation, ValueError):
        return None, "Falha de conversao decimal"

# 5. Diagnostico de todos os valores monetarios
# Antes de criar qualquer arquivo derivado.

convertidos = {}
problemas = {}

print("\n=== T4.4 - VALIDACAO DOS FORMATOS ===")

for campo_original, campo_novo in campos.items():
    valores = []
    invalidos = []

    for posicao, valor in enumerate(
        base[campo_original], start=1
    ):
        numero, situacao = converter_preco(valor)

        valores.append(numero)

        if numero is None:
            invalidos.append(
                (posicao, situacao, valor)
            )

    convertidos[campo_novo] = valores
    problemas[campo_original] = invalidos

    print(f"\nCampo: {campo_original}")
    print("Registros examinados:", len(valores))
    print(
        "Valores validos:",
        sum(numero is not None for numero in valores)
    )
    print("Formatos inesperados:", len(invalidos))

    if invalidos:
        print("Primeiras ocorrencias para investigacao:")

        for posicao, situacao, valor in invalidos[:5]:
            print(
                f"  Registro {posicao}: "
                f"{situacao} | valor={valor!r}"
            )

# Nao prosseguir diante de formatos inesperados.

if any(problemas.values()):
    raise ValueError(
        "Foram identificados formatos monetarios "
        "nao aprovados. Nenhum arquivo foi criado."
    )

# 6. Criacao das colunas numericas derivadas
# Cada valor e um objeto Decimal, nao float.

tratada = base.copy(deep=True)

for campo_novo, valores in convertidos.items():
    tratada[campo_novo] = pd.Series(
        valores,
        index=base.index,
        dtype=object
    )

# 7. Validacoes antes da gravacao

if tratada.shape != (1465, 21):
    raise AssertionError("Dimensoes incorretas.")

pd.testing.assert_frame_equal(
    base,
    tratada[colunas_anteriores],
    check_dtype=True
)

for campo_novo in campos.values():
    coluna = tratada[campo_novo]

    if not coluna.map(
        lambda valor: isinstance(valor, Decimal)
    ).all():
        raise AssertionError(
            f"Tipo Decimal nao preservado: {campo_novo}"
        )

    if coluna.isna().sum() != 0:
        raise AssertionError(
            f"Valores ausentes inesperados: {campo_novo}"
        )

    if not coluna.map(
        lambda valor: valor >= Decimal("0.00")
    ).all():
        raise AssertionError(
            f"Preco negativo: {campo_novo}"
        )

    if not coluna.map(
        lambda valor: (
            valor == valor.quantize(centavo)
        )
    ).all():
        raise AssertionError(
            f"Precisao decimal divergente: {campo_novo}"
        )

print("\n=== T4.4 - RESULTADOS EM MEMORIA ===")
print("Registros antes:", len(base))
print("Registros depois:", len(tratada))
print("Colunas antes:", len(base.columns))
print("Colunas depois:", len(tratada.columns))

for campo_novo in campos.values():
    print(
        f"{campo_novo}:",
        len(tratada[campo_novo]),
        "valores Decimal"
    )

print("Registros excluidos: 0")
print("Imputacoes realizadas: 0")
print("Conversoes cambiais: 0")

# 8. Gravacao protegida
# Decimal sera serializado como numero com duas casas.
# A fonte original e a versao T4.3 nao sao alteradas.

with saida.open(
    "x", encoding="utf-8", newline=""
) as f:
    tratada.to_csv(
        f,
        index=False,
        na_rep=""
    )

# 9. Reabertura e comparacao

reaberta = pd.read_csv(
    saida,
    dtype=str,
    keep_default_na=False,
    encoding="utf-8"
)

if reaberta.shape != (1465, 21):
    raise AssertionError(
        "Dimensoes divergentes apos reabertura."
    )

pd.testing.assert_frame_equal(
    base,
    reaberta[colunas_anteriores],
    check_dtype=True
)

for campo_novo in campos.values():
    valores_esperados = tratada[campo_novo].map(str)

    if not reaberta[campo_novo].equals(
        valores_esperados
    ):
        raise AssertionError(
            f"Valores divergentes: {campo_novo}"
        )

    # Recuperacao exata com Decimal
    recuperados = reaberta[campo_novo].map(Decimal)

    if not recuperados.equals(
        tratada[campo_novo]
    ):
        raise AssertionError(
            f"Falha na recuperacao Decimal: {campo_novo}"
        )

# 10. Conferencia final de integridade

if calcular_sha256(original) != hash_original:
    raise AssertionError(
        "Integridade da base original alterada."
    )

if calcular_sha256(entrada) != hash_t43:
    raise AssertionError(
        "Integridade da versao T4.3 alterada."
    )

print("\n=== T4.4 - VALIDACAO FINAL ===")
print("Arquivo derivado:", saida.name)
print("Registros preservados:", len(reaberta))
print("Colunas anteriores preservadas:", len(colunas_anteriores))
print("Total de colunas:", len(reaberta.columns))
print("Precos convertidos em cada coluna: 1465")
print("Moeda: INR (rupia indiana)")
print("Precisao: duas casas decimais")
print("SHA-256 original: PRESERVADO")
print("SHA-256 T4.3: PRESERVADO")
print("SHA-256 T4.4:", calcular_sha256(saida))
print("Reabertura e comparacao: APROVADAS")
print("=== FIM DO TRATAMENTO T4.4 ===")
```

## 4.5. Tratamento T4.5 — Conversão do percentual de desconto

### Objetivo

Criar uma representação numérica para os percentuais
de desconto informados no Amazon Sales Dataset,
possibilitando cálculos e análises comerciais.

### Problema identificado

O diagnóstico P3.4 identificou que os 1.465
registros apresentavam `discount_percentage`
em formato textual, contendo o símbolo `%`.

Não foram identificados valores vazios nesse campo.

### Decisão metodológica

Foi aprovada a criação da variável `discount_rate`,
preservando a coluna original `discount_percentage`.

A nova variável será representada em escala de
0 a 1, pela divisão do percentual informado por 100.

Exemplo ilustrativo:

- Valor original: `64%`
- Valor derivado: `0.64`
- Exibição percentual: `64%`

Não serão realizadas correções dos percentuais
com base nos preços dos produtos.

### Regras de tratamento

1. Preservar os 1.465 registros.
2. Preservar as 21 colunas anteriores.
3. Validar a representação textual dos percentuais.
4. Verificar a faixa de 0% a 100%.
5. Converter os valores utilizando `Decimal`.
6. Criar exclusivamente a coluna `discount_rate`.
7. Não excluir, imputar ou agregar registros.
8. Interromper a execução diante de valores inesperados.

### Fonte e versionamento

**Arquivo de entrada:**

`02_DADOS_TRATADOS/amazon_t4_04_precos_inr_v01.csv`

**Arquivo de saída previsto:**

`02_DADOS_TRATADOS/amazon_t4_05_desconto_pct_v01.csv`

### Metodologia

A transformação será realizada em Python,
utilizando pandas, Decimal, expressões regulares
e verificações de integridade SHA-256.

Os valores originais serão preservados e
os percentuais convertidos sem utilização
de números de ponto flutuante.

A representação numérica será armazenada
em CSV e validada após reabertura.

### Critérios de validação

- Integridade dos arquivos anteriores confirmada.
- 1.465 registros preservados.
- 21 colunas anteriores preservadas.
- 22 colunas na versão derivada.
- 1.465 percentuais convertidos.
- Valores derivados entre 0 e 1.
- Correspondência exata com os percentuais originais.
- Nenhuma exclusão, imputação ou correção.
- Arquivo derivado salvo e reaberto com sucesso.

### Limitações

A conversão não comprova a coerência entre o
percentual informado e os preços do produto.

Não serão avaliados neste procedimento:

- Divergências entre percentuais e preços;
- Valores extremos de descontos por categoria;
- Descontos em produtos repetidos;
- Relações entre descontos e avaliações;
- Efeitos causais dos descontos sobre satisfação.

O formato CSV não preserva nativamente o tipo
Decimal do Python.

A interpretação numérica e o formato percentual
deverão ser conferidos posteriormente no Power BI.


```python

# FIAP - TECH CHALLENGE FASE 04
# Grupo 165
# T4.5 - Conversao aprovada do percentual de desconto

from pathlib import Path
from decimal import Decimal, InvalidOperation
import hashlib
import re
import pandas as pd

# 1. Caminhos e referencias de integridade

raiz = Path("/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON")

original = raiz / "01_DADOS_ORIGINAIS" / "01.amazon.csv"

entrada = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_04_precos_inr_v01.csv"
)

saida = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_05_desconto_pct_v01.csv"
)

hash_original = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

hash_t44 = (
    "F891CF8DE4D5BC940F9AFD143E8A0AC3"
    "88BC2FB1587FF4C159DB1E9367818313"
)

def calcular_sha256(caminho):
    sha = hashlib.sha256()

    with caminho.open("rb") as f:
        for bloco in iter(
            lambda: f.read(1024 * 1024), b""
        ):
            sha.update(bloco)

    return sha.hexdigest().upper()

# 2. Controles iniciais

for caminho in [original, entrada]:
    if not caminho.is_file():
        raise FileNotFoundError(
            f"Arquivo nao encontrado: {caminho}"
        )

if saida.exists():
    raise FileExistsError(
        "Arquivo T4.5 ja existe. "
        "Execucao interrompida para evitar sobrescrita."
    )

if original.stat().st_size != 4744481:
    raise ValueError("Tamanho da base original divergente.")

if calcular_sha256(original) != hash_original:
    raise ValueError("SHA-256 original divergente.")

if calcular_sha256(entrada) != hash_t44:
    raise ValueError("SHA-256 T4.4 divergente.")

print("=== T4.5 - CONTROLE INICIAL ===")
print("SHA-256 original: CONFERIDO")
print("SHA-256 T4.4: CONFERIDO")

# 3. Leitura textual da versao T4.4

base = pd.read_csv(
    entrada,
    dtype=str,
    keep_default_na=False,
    encoding="utf-8"
)

if base.shape != (1465, 21):
    raise ValueError(
        f"Dimensoes inesperadas: {base.shape}"
    )

coluna_original = "discount_percentage"
coluna_nova = "discount_rate"

if coluna_original not in base.columns:
    raise ValueError("Coluna discount_percentage ausente.")

if coluna_nova in base.columns:
    raise ValueError("Coluna discount_rate ja existente.")

colunas_anteriores = base.columns.tolist()

# 4. Validacao rigorosa dos percentuais
# Aceita exemplos como 0%, 10%, 64% e 12.5%.
# Espacos externos sao tolerados, mas o dado
# original permanece integralmente preservado.

padrao_percentual = re.compile(
    r"([0-9]+(?:\.[0-9]+)?)%"
)

cem = Decimal("100")
zero = Decimal("0")
um = Decimal("1")

def converter_percentual(valor):
    texto = valor.strip()

    correspondencia = padrao_percentual.fullmatch(texto)

    if correspondencia is None:
        return None, "Formato percentual inesperado"

    try:
        percentual = Decimal(correspondencia.group(1))

        if not percentual.is_finite():
            return None, "Valor nao finito"

        if not (zero <= percentual <= cem):
            return None, "Percentual fora da faixa 0 a 100"

        taxa = percentual / cem

        return taxa, "Valido"

    except (InvalidOperation, ValueError):
        return None, "Falha de conversao decimal"

# 5. Verificacao de todos os valores antes de salvar

valores_convertidos = []
problemas = []

for posicao, valor in enumerate(
    base[coluna_original], start=1
):
    taxa, situacao = converter_percentual(valor)

    valores_convertidos.append(taxa)

    if taxa is None:
        problemas.append((posicao, situacao, valor))

print("\n=== T4.5 - VALIDACAO DOS FORMATOS ===")
print("Registros examinados:", len(valores_convertidos))
print(
    "Percentuais validos:",
    sum(valor is not None for valor in valores_convertidos)
)
print("Formatos ou valores inesperados:", len(problemas))

if problemas:
    print("\nPrimeiras ocorrencias para investigacao:")

    for posicao, situacao, valor in problemas[:10]:
        print(
            f"Registro {posicao}: "
            f"{situacao} | valor={valor!r}"
        )

    raise ValueError(
        "Foram identificados percentuais nao aprovados. "
        "Nenhum arquivo foi criado."
    )

# 6. Criacao da coluna derivada com Decimal

tratada = base.copy(deep=True)

tratada[coluna_nova] = pd.Series(
    valores_convertidos,
    index=base.index,
    dtype=object
)

# 7. Testes antes do salvamento

if tratada.shape != (1465, 22):
    raise AssertionError("Dimensoes incorretas.")

pd.testing.assert_frame_equal(
    base,
    tratada[colunas_anteriores],
    check_dtype=True
)

if not tratada[coluna_nova].map(
    lambda valor: isinstance(valor, Decimal)
).all():
    raise AssertionError("Tipo Decimal nao preservado.")

if tratada[coluna_nova].isna().sum() != 0:
    raise AssertionError("Valores ausentes inesperados.")

if not tratada[coluna_nova].map(
    lambda valor: zero <= valor <= um
).all():
    raise AssertionError("Taxas fora da faixa 0 a 1.")

# Verificacao exata da conversao
for percentual_original, taxa in zip(
    base[coluna_original],
    tratada[coluna_nova]
):
    valor_percentual = Decimal(
        percentual_original.strip().removesuffix("%")
    )

    if taxa * cem != valor_percentual:
        raise AssertionError(
            "Conversao percentual divergente."
        )

print("\n=== T4.5 - RESULTADOS EM MEMORIA ===")
print("Registros antes:", len(base))
print("Registros depois:", len(tratada))
print("Colunas antes:", len(base.columns))
print("Colunas depois:", len(tratada.columns))
print(
    "Taxas numericas validas:",
    int(tratada[coluna_nova].notna().sum())
)
print(
    "Valores ausentes:",
    int(tratada[coluna_nova].isna().sum())
)
print("Registros excluidos: 0")
print("Imputacoes realizadas: 0")
print("Correcoes dos descontos: 0")

# 8. Salvamento protegido contra sobrescrita

with saida.open(
    "x", encoding="utf-8", newline=""
) as f:
    tratada.to_csv(
        f,
        index=False,
        na_rep=""
    )

# 9. Reabertura e comparacao

reaberta = pd.read_csv(
    saida,
    dtype=str,
    keep_default_na=False,
    encoding="utf-8"
)

if reaberta.shape != (1465, 22):
    raise AssertionError(
        "Dimensoes divergentes apos reabertura."
    )

pd.testing.assert_frame_equal(
    base,
    reaberta[colunas_anteriores],
    check_dtype=True
)

# Recuperacao exata dos objetos Decimal
recuperadas = reaberta[coluna_nova].map(Decimal)

for esperado, recuperado in zip(
    tratada[coluna_nova],
    recuperadas
):
    if esperado != recuperado:
        raise AssertionError(
            "Valor divergente apos reabertura."
        )

if not recuperadas.map(
    lambda valor: zero <= valor <= um
).all():
    raise AssertionError(
        "Taxas fora da faixa apos reabertura."
    )

# 10. Integridade das versoes anteriores

if calcular_sha256(original) != hash_original:
    raise AssertionError("Base original alterada.")

if calcular_sha256(entrada) != hash_t44:
    raise AssertionError("Arquivo T4.4 alterado.")

print("\n=== T4.5 - VALIDACAO FINAL ===")
print("Arquivo derivado:", saida.name)
print("Registros preservados:", len(reaberta))
print("Colunas anteriores preservadas:", len(colunas_anteriores))
print("Total de colunas:", len(reaberta.columns))
print("Percentuais convertidos:", len(recuperadas))
print("Valores ausentes:", int(recuperadas.isna().sum()))
print("Escala numerica: 0 a 1")
print("Moeda: nao se aplica")
print("SHA-256 original: PRESERVADO")
print("SHA-256 T4.4: PRESERVADO")
print("SHA-256 T4.5:", calcular_sha256(saida))
print("Reabertura e comparacao: APROVADAS")
print("=== FIM DO TRATAMENTO T4.5 ===")
```


### 4.5.1. Resultados, validação e discussão do tratamento T4.5

## 4.6. Diagnóstico de consistência entre preços e descontos

### Objetivo

Avaliar a coerência matemática entre os preços
originais, os preços com desconto e os percentuais
informados no Amazon Sales Dataset.

### Fundamentação metodológica

Os tratamentos T4.4 e T4.5 disponibilizaram
variáveis numéricas de preços e descontos.

Antes de utilizá-las em indicadores comerciais,
é necessário investigar sua consistência,
considerando diferenças decorrentes do
arredondamento dos percentuais.

### Metodologia

A investigação será realizada em Python,
utilizando Decimal e a versão T4.5 da base.

Serão avaliados:

- CD01: preços originais iguais a zero;
- CD02: preços descontados superiores aos originais;
- CD03: correspondência matemática exata;
- CD04: compatibilidade com arredondamento;
- CD05: magnitude das divergências residuais.

Os resultados serão quantificados em registros
e percentuais, com identificação das posições
que necessitam de investigação.

### Critérios de validação

1. Integridade SHA-256 confirmada.
2. Dimensões de 1.465 registros e 22 colunas.
3. Comparações realizadas com Decimal.
4. Divergências quantificadas.
5. Nenhuma alteração nos arquivos de origem.

### Limitações

O arredondamento testado constitui uma hipótese,
não uma confirmação da regra utilizada na
elaboração do dataset.

Diferenças matemáticas não serão classificadas
automaticamente como erros.

Nenhum tratamento será realizado neste procedimento.


```python

# P4.6 - Restabelecimento do acesso ao Google Drive
# Nenhum dado sera alterado.

from google.colab import drive
from pathlib import Path

drive.mount("/content/drive")

raiz = Path(
    "/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON"
)

original = raiz / "01_DADOS_ORIGINAIS" / "01.amazon.csv"

tratada = (
    raiz / "02_DADOS_TRATADOS"
    / "amazon_t4_05_desconto_pct_v01.csv"
)

print("\n=== VERIFICACAO DO ACESSO ===")
print("Pasta do projeto localizada:", raiz.is_dir())
print("Base original localizada:", original.is_file())
print("Base T4.5 localizada:", tratada.is_file())
print("=== FIM DA VERIFICACAO ===")
```

```python

# FIAP - TECH CHALLENGE FASE 04
# P4.6 - Diagnostico de consistencia entre precos e descontos
# SOMENTE LEITURA - nenhuma alteracao sera realizada.

from pathlib import Path
from decimal import Decimal, ROUND_HALF_UP
from collections import Counter
import hashlib
import re
import pandas as pd

raiz = Path("/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON")

original = raiz / "01_DADOS_ORIGINAIS" / "01.amazon.csv"

entrada = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_05_desconto_pct_v01.csv"
)

hash_original = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

hash_t45 = (
    "5EFDD045F3FF6165FDC5865305E744A1"
    "4E7227F82654DD23983613026194EAA6"
)

def calcular_sha256(caminho):
    sha = hashlib.sha256()
    with caminho.open("rb") as f:
        for bloco in iter(lambda: f.read(1024 * 1024), b""):
            sha.update(bloco)
    return sha.hexdigest().upper()

# 1. Verificacao de integridade

for caminho in [original, entrada]:
    if not caminho.is_file():
        raise FileNotFoundError(caminho)

if calcular_sha256(original) != hash_original:
    raise ValueError("SHA-256 original divergente.")

if calcular_sha256(entrada) != hash_t45:
    raise ValueError("SHA-256 T4.5 divergente.")

print("=== P4.6 - CONTROLE DE INTEGRIDADE ===")
print("SHA-256 original: CONFERIDO")
print("SHA-256 T4.5: CONFERIDO")

# 2. Leitura textual, sem conversoes persistidas

base = pd.read_csv(
    entrada,
    dtype=str,
    keep_default_na=False,
    encoding="utf-8"
)

if base.shape != (1465, 22):
    raise ValueError(f"Dimensoes inesperadas: {base.shape}")

campos = [
    "actual_price_inr",
    "discounted_price_inr",
    "discount_percentage",
    "discount_rate"
]

if not all(c in base.columns for c in campos):
    raise ValueError("Colunas necessarias ausentes.")

print("\n=== DIMENSOES ===")
print("Registros:", len(base))
print("Colunas:", len(base.columns))

# 3. Analise com Decimal

padrao_percentual = re.compile(
    r"([0-9]+(?:\.[0-9]+)?)%"
)

resultados = []
inconsistencias_formato = []

for posicao, registro in enumerate(
    base.to_dict("records"), start=1
):
    try:
        preco_original = Decimal(
            registro["actual_price_inr"]
        )
        preco_descontado = Decimal(
            registro["discounted_price_inr"]
        )

        texto_pct = registro["discount_percentage"].strip()
        correspondencia = padrao_percentual.fullmatch(texto_pct)

        if not correspondencia:
            raise ValueError("Percentual original invalido.")

        texto_numero = correspondencia.group(1)
        informado = Decimal(texto_numero)
        taxa = Decimal(registro["discount_rate"])

        if not (
            preco_original.is_finite()
            and preco_descontado.is_finite()
            and informado.is_finite()
            and taxa.is_finite()
        ):
            raise ValueError("Valor nao finito.")

        if taxa * Decimal("100") != informado:
            raise ValueError(
                "Taxa derivada diverge do percentual original."
            )

        if preco_original < 0 or preco_descontado < 0:
            raise ValueError("Preco negativo inesperado.")

        # Casas decimais efetivamente informadas
        casas = (
            len(texto_numero.split(".")[1])
            if "." in texto_numero else 0
        )

        if preco_original == 0:
            categoria = "Preco original zero"
            calculado = None
            diferenca = None
        else:
            calculado = (
                (preco_original - preco_descontado)
                / preco_original * Decimal("100")
            )

            diferenca = abs(calculado - informado)

            unidade = Decimal("1").scaleb(-casas)

            arredondado = calculado.quantize(
                unidade,
                rounding=ROUND_HALF_UP
            )

            if calculado == informado:
                categoria = "Correspondencia exata"
            elif arredondado == informado:
                categoria = "Compativel com arredondamento"
            elif diferenca <= Decimal("1"):
                categoria = "Divergencia ate 1 pp"
            else:
                categoria = "Divergencia superior a 1 pp"

        resultados.append({
            "posicao": posicao,
            "original": preco_original,
            "descontado": preco_descontado,
            "informado": informado,
            "calculado": calculado,
            "diferenca_pp": diferenca,
            "categoria": categoria,
            "descontado_maior": (
                preco_descontado > preco_original
            )
        })

    except Exception as erro:
        inconsistencias_formato.append(
            (posicao, str(erro))
        )

if inconsistencias_formato:
    print("\nOcorrencias com formato inesperado:")
    for item in inconsistencias_formato[:10]:
        print(item)

    raise ValueError(
        "Verificacao interrompida por inconsistencias."
    )

# 4. CD01 e CD02

total = len(resultados)

zeros = sum(
    r["original"] == 0 for r in resultados
)

descontados_maiores = sum(
    r["descontado_maior"] for r in resultados
)

print("\n=== CD01/CD02: VERIFICACAO DOS PRECOS ===")
print("Precos originais iguais a zero:", zeros)
print(
    "Precos descontados superiores aos originais:",
    descontados_maiores
)

# 5. CD03 a CD05: classificacao das diferencas

frequencias = Counter(
    r["categoria"] for r in resultados
)

print("\n=== CD03/CD04/CD05: CONSISTENCIA DOS DESCONTOS ===")

categorias = [
    "Correspondencia exata",
    "Compativel com arredondamento",
    "Divergencia ate 1 pp",
    "Divergencia superior a 1 pp",
    "Preco original zero"
]

for categoria in categorias:
    quantidade = frequencias[categoria]
    percentual = Decimal(quantidade) * 100 / total

    print(
        f"{categoria}: {quantidade} "
        f"({percentual:.2f}%)"
    )

if sum(frequencias.values()) != total:
    raise AssertionError("Classificacao incompleta.")

# 6. Maiores divergencias, sem identificadores pessoais

divergencias = [
    r for r in resultados
    if r["diferenca_pp"] is not None
]

divergencias.sort(
    key=lambda r: r["diferenca_pp"],
    reverse=True
)

print("\n=== MAIORES DIFERENCAS ABSOLUTAS ===")
print("Valores em INR; diferencas em pontos percentuais.")

for r in divergencias[:10]:
    print(
        f"Registro {r['posicao']}: "
        f"original={r['original']:.2f} | "
        f"descontado={r['descontado']:.2f} | "
        f"informado={r['informado']}% | "
        f"calculado={r['calculado']:.4f}% | "
        f"diferenca={r['diferenca_pp']:.4f} pp | "
        f"{r['categoria']}"
    )

# 7. Integridade final

if calcular_sha256(original) != hash_original:
    raise AssertionError("Base original alterada.")

if calcular_sha256(entrada) != hash_t45:
    raise AssertionError("Versao T4.5 alterada.")

print("\n=== FIM DO P4.6 ===")
print("Registros examinados:", total)
print("SHA-256 original: PRESERVADO")
print("SHA-256 T4.5: PRESERVADO")
print("Nenhuma transformacao persistida.")
print("Nenhum arquivo derivado criado.")
```


### 4.6.1. Resultados e discussão da consistência entre preços e descontos

O procedimento P4.6 avaliou a consistência matemática
entre o preço original, o preço com desconto e o
percentual de desconto informado nos 1.465 registros
do Amazon Sales Dataset.

A investigação utilizou a versão derivada T4.5,
com 22 variáveis, e cálculos de precisão decimal
realizados com a classe Decimal do Python.

A integridade dos arquivos original e T4.5 foi
confirmada por comparação dos hashes SHA-256.

#### CD01 e CD02 — Verificação dos preços

| Critério | Registros |
|---|---:|
| Preços originais iguais a zero | 0 |
| Preços com desconto superiores aos originais | 0 |

Não foram encontradas situações que impedissem
o cálculo do desconto por divisão pelo preço
original.

Também não foram identificados registros em que
o preço descontado superasse o preço original.

#### CD03 a CD05 — Consistência dos descontos

O desconto calculado foi obtido pela expressão:

\[
\text{Desconto calculado (\%)} =
\frac{\text{Preço original} - \text{Preço descontado}}
{\text{Preço original}} \times 100
\]

Os resultados foram:

| Classificação | Registros | Percentual |
|---|---:|---:|
| Correspondência exata | 68 | 4,64% |
| Compatível com arredondamento | 1.397 | 95,36% |
| Divergência residual até 1 p.p. | 0 | 0,00% |
| Divergência superior a 1 p.p. | 0 | 0,00% |
| Preço original igual a zero | 0 | 0,00% |
| **Total** | **1.465** | **100,00%** |

Todos os registros apresentaram correspondência
exata ou compatibilidade com o arredondamento
testado.

O método utilizado para testar essa compatibilidade
foi ROUND_HALF_UP, respeitando a quantidade de
casas decimais do percentual informado.

#### Análise das maiores diferenças

A listagem das maiores diferenças absolutas
apresentou valores de até 0,5000 ponto percentual.

Dois registros ilustram esse comportamento:

|---|---:|---:|---:|---:|
| 818 | 600,00 | 249,00 | 59% | 58,5000% |
| 1457 | 1.200,00 | 498,00 | 59% | 58,5000% |

Nos dois casos, o percentual informado de 59%
é compatível com o arredondamento de 58,5%
pelo método ROUND_HALF_UP.

As diferenças observadas não constituíram
divergências residuais após a aplicação
do critério de arredondamento testado.

#### Interpretação dos resultados

Os resultados indicam consistência matemática
entre preços e percentuais informados,
considerando o método de arredondamento avaliado.

A maior parte dos registros apresentou percentuais
compatíveis com arredondamento, enquanto 68
registros apresentaram correspondência exata.

Não foram identificadas divergências que
justificassem correção automática de preços
ou percentuais no escopo deste procedimento.

#### Implicações para a análise de negócio

Os resultados permitem utilizar os preços e
percentuais convertidos nas análises comerciais,
respeitando a escala numérica e a moeda original.

Os preços estão denominados em rupias indianas (INR).

Os descontos informados podem ser utilizados

#### Limitações metodológicas

## 5.1. EDA inicial — Portfólio, preços, descontos e satisfação

### Objetivo

Caracterizar quantitativamente o conjunto de dados,
produzindo estatísticas descritivas iniciais sobre:

1. Quantidade de registros e produtos distintos.
2. Distribuição dos produtos por categoria.
3. Preços originais e com desconto, em INR.
4. Taxas de desconto.
5. Avaliações dos produtos.
6. Contagens informadas de avaliações.

### Fonte

**Arquivo de entrada:**

`02_DADOS_TRATADOS/amazon_t4_05_desconto_pct_v01.csv`

**Dimensões esperadas:**

1.465 registros e 22 colunas.

### Metodologia

A análise será realizada em Python, utilizando
pandas, Decimal e funções estatísticas descritivas.

Serão consideradas duas unidades de análise:

- Registro: cada linha do arquivo CSV.
- Produto distinto: cada valor único de `product_id`.

As estatísticas por produto serão utilizadas
somente quando os atributos envolvidos
apresentarem consistência dentro dos grupos
de identificadores repetidos.

Os preços serão analisados em rupias indianas,
sem conversão cambial.

As contagens de avaliações serão examinadas
sem pressupor que correspondam a vendas,
faturamento ou avaliações individuais únicas.

### Critérios de validação

- Integridade SHA-256 confirmada.
- Dimensões de 1.465 × 22.
- Quantidade de produtos distintos conciliada
  com o diagnóstico P3.5.
- Valores ausentes identificados.
- Denominadores explicitados.
- Nenhuma alteração dos arquivos de origem.

### Limitações

A análise descritiva não estabelece causalidade.

A existência de produtos repetidos poderá
influenciar estatísticas calculadas por registro.

As categorias serão analisadas inicialmente
conforme os valores textuais completos da fonte,
sem redefinição de sua hierarquia.

Não serão realizadas exclusões, correções,
consolidações persistentes ou novos tratamentos.

```python

# FIAP - TECH CHALLENGE FASE 04
# Grupo 165
# E5.1 - Analise exploratoria descritiva inicial
# SOMENTE LEITURA - nenhum arquivo sera criado ou alterado.

from pathlib import Path
from collections import Counter
from decimal import Decimal, InvalidOperation
import hashlib
import pandas as pd

# 1. Fonte e integridade

raiz = Path("/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON")

original = (
    raiz / "01_DADOS_ORIGINAIS" / "01.amazon.csv"
)

entrada = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_05_desconto_pct_v01.csv"
)

hash_original = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

hash_t45 = (
    "5EFDD045F3FF6165FDC5865305E744A1"
    "4E7227F82654DD23983613026194EAA6"
)

def calcular_sha256(caminho):
    sha = hashlib.sha256()
    with caminho.open("rb") as f:
        for bloco in iter(lambda: f.read(1024 * 1024), b""):
            sha.update(bloco)
    return sha.hexdigest().upper()

for caminho in [original, entrada]:
    if not caminho.is_file():
        raise FileNotFoundError(caminho)

if calcular_sha256(original) != hash_original:
    raise ValueError("SHA-256 original divergente.")

if calcular_sha256(entrada) != hash_t45:
    raise ValueError("SHA-256 T4.5 divergente.")

print("=== E5.1 - INTEGRIDADE ===")
print("SHA-256 original: CONFERIDO")
print("SHA-256 T4.5: CONFERIDO")
print("Pandas:", pd.__version__)

# 2. Leitura como texto, sem alterar tipos no arquivo

base = pd.read_csv(
    entrada,
    dtype=str,
    keep_default_na=False,
    encoding="utf-8"
)

if base.shape != (1465, 22):
    raise ValueError(
        f"Dimensoes inesperadas: {base.shape}"
    )

campos = [
    "product_id",
    "category",
    "actual_price_inr",
    "discounted_price_inr",
    "discount_rate",
    "rating_numerico",
    "rating_count_numerico"
]

if not all(c in base.columns for c in campos):
    raise ValueError("Colunas necessarias ausentes.")

print("\n=== DIMENSOES ===")
print("Registros:", len(base))
print("Colunas:", len(base.columns))

# 3. Funcao de leitura numerica temporaria
# Os dados originais e derivados nao sao modificados.

def ler_decimais(serie, nome, ausentes_esperados=0):
    numeros = []
    ausentes = 0

    for posicao, texto in enumerate(serie, start=1):
        if texto == "":
            ausentes += 1
            continue

        try:
            numero = Decimal(texto)
        except InvalidOperation:
            raise ValueError(
                f"{nome}: formato invalido no registro {posicao}"
            )

        if not numero.is_finite():
            raise ValueError(
                f"{nome}: valor nao finito no registro {posicao}"
            )

        numeros.append(numero)

    if ausentes != ausentes_esperados:
        raise ValueError(
            f"{nome}: ausentes inesperados: {ausentes}"
        )

    return numeros

def resumo_estatistico(valores, casas=2):
    if not valores:
        raise ValueError("Serie sem valores validos.")

    ordenados = sorted(valores)
    n = len(ordenados)

    media = sum(
        ordenados, Decimal("0")
    ) / Decimal(n)

    mediana = (
        ordenados[(n - 1) // 2]
        + ordenados[n // 2]
    ) / Decimal("2")

    formato = f",.{casas}f"

    return {
        "n": n,
        "min": format(ordenados[0], formato),
        "media": format(media, formato),
        "mediana": format(mediana, formato),
        "max": format(ordenados[-1], formato)
    }

def imprimir_resumo(nome, valores, casas=2):
    dados = resumo_estatistico(valores, casas)
    print(f"\n{nome}")
    print("Observacoes validas:", dados["n"])
    print("Minimo:", dados["min"])
    print("Media:", dados["media"])
    print("Mediana:", dados["mediana"])
    print("Maximo:", dados["max"])

# 4. EDA01 - Portfolio e produtos distintos

frequencia_produtos = Counter(base["product_id"])

produtos_distintos = len(frequencia_produtos)

grupos_repetidos = sum(
    n > 1 for n in frequencia_produtos.values()
)

excedentes = sum(
    n - 1
    for n in frequencia_produtos.values()
    if n > 1
)

if (
    produtos_distintos != 1351
    or grupos_repetidos != 92
    or excedentes != 114
):
    raise ValueError(
        "Contagem de produtos divergente do P3.5."
    )

print("\n=== EDA01: PORTFOLIO ===")
print("Registros:", len(base))
print("Produtos distintos:", produtos_distintos)
print("Produtos com ID repetido:", grupos_repetidos)
print("Ocorrencias excedentes:", excedentes)

# 5. EDA02 - Categorias por produto distinto
# Verifica consistencia antes de contar produtos.

variacoes_categoria = (
    base.groupby("product_id")["category"]
    .nunique(dropna=False)
)

if (variacoes_categoria > 1).any():
    raise ValueError(
        "Produtos com categorias divergentes. "
        "Analise interrompida."
    )

# Deduplicacao somente em memoria para contagem.
# Nenhum arquivo e modificado ou consolidado.

produtos_unicos = base.drop_duplicates(
    subset=["product_id"]
).copy()

if len(produtos_unicos) != 1351:
    raise AssertionError("Produtos distintos divergentes.")

categorias_produtos = Counter(
    produtos_unicos["category"]
)

categorias_registros = Counter(
    base["category"]
)

if sum(categorias_produtos.values()) != 1351:
    raise AssertionError(
        "Soma de produtos por categoria divergente."
    )

print("\n=== EDA02: CATEGORIAS ===")
print(
    "Categorias textuais distintas:",
    len(categorias_produtos)
)

print("\n10 categorias com mais produtos distintos:")
print("Produtos | Registros | Categoria original completa")

ranking_categorias = sorted(
    categorias_produtos.items(),
    key=lambda item: (-item[1], item[0])
)

for categoria, quantidade in ranking_categorias[:10]:
    print(
        f"{quantidade:7d} | "
        f"{categorias_registros[categoria]:9d} | "
        f"{categoria}"
    )

# 6. EDA03 - Estatisticas monetarias por registro

precos_originais = ler_decimais(
    base["actual_price_inr"],
    "actual_price_inr"
)

precos_descontados = ler_decimais(
    base["discounted_price_inr"],
    "discounted_price_inr"
)

print("\n=== EDA03: PRECOS POR REGISTRO (INR) ===")

imprimir_resumo(
    "Preco original em INR",
    precos_originais
)

imprimir_resumo(
    "Preco com desconto em INR",
    precos_descontados
)

# 7. EDA04 - Taxa de desconto por registro

taxas = ler_decimais(
    base["discount_rate"],
    "discount_rate"
)

if not all(
    Decimal("0") <= taxa <= Decimal("1")
    for taxa in taxas
):
    raise ValueError("Taxa fora da faixa de 0 a 1.")

taxas_percentuais = [
    taxa * Decimal("100")
    for taxa in taxas
]

print("\n=== EDA04: DESCONTOS POR REGISTRO ===")

imprimir_resumo(
    "Desconto informado (%)",
    taxas_percentuais
)

# 8. EDA05 - Satisfacao: registros e produtos
# A utilizacao de produtos unicos exige que
# o rating seja constante dentro de cada product_id.

variacoes_rating = (
    base.groupby("product_id")["rating_numerico"]
    .nunique(dropna=False)
)

if (variacoes_rating > 1).any():
    raise ValueError(
        "Rating divergente dentro do mesmo produto."
    )

ratings_registros = ler_decimais(
    base["rating_numerico"],
    "rating_numerico",
    ausentes_esperados=1
)

ratings_produtos = ler_decimais(
    produtos_unicos["rating_numerico"],
    "rating_numerico por produto",
    ausentes_esperados=1
)

print("\n=== EDA05: SATISFACAO ===")

imprimir_resumo(
    "Rating por registro",
    ratings_registros
)

imprimir_resumo(
    "Rating por produto distinto",
    ratings_produtos
)

print("\nDistribuicao dos ratings por produto distinto:")

frequencias_rating = Counter(ratings_produtos)

for nota, quantidade in sorted(
    frequencias_rating.items()
):
    print(
        f"Rating {nota}: {quantidade} produtos"
    )

print(
    "Produtos sem rating numerico:",
    len(produtos_unicos) - len(ratings_produtos)
)

# 9. EDA06 - Contagens de avaliacoes por registro
# Nao somar contagens, pois a granularidade
# dos produtos repetidos exige cuidado.

contagens = ler_decimais(
    base["rating_count_numerico"],
    "rating_count_numerico",
    ausentes_esperados=2
)

if not all(
    valor >= 0
    and valor == valor.to_integral_value()
    for valor in contagens
):
    raise ValueError(
        "Contagem negativa ou nao inteira."
    )

print("\n=== EDA06: ENGAJAMENTO POR REGISTRO ===")

imprimir_resumo(
    "Contagem informada de avaliacoes",
    contagens
)

print(
    "Registros sem contagem:",
    len(base) - len(contagens)
)

print(
    "Registros com contagem zero:",
    sum(valor == 0 for valor in contagens)
)

print(
    "\nNota: as contagens nao foram somadas."
)
print(
    "Nao foi estimado total de reviews unicos."
)

# 10. Integridade final

if calcular_sha256(original) != hash_original:
    raise AssertionError("Arquivo original alterado.")

if calcular_sha256(entrada) != hash_t45:
    raise AssertionError("Arquivo T4.5 alterado.")

print("\n=== FIM DO E5.1 ===")
print("SHA-256 original: PRESERVADO")
print("SHA-256 T4.5: PRESERVADO")
print("Nenhum tratamento realizado.")
print("Nenhum arquivo derivado criado.")
```

descritivos de portfólio, preços,
descontos, satisfação e engajamento.

Os resultados foram obtidos sem novos
tratamentos e com integridade dos arquivos
de origem confirmada.

As análises fornecem subsídios para o
desenvolvimento do dashboard executivo,
respeitando as limitações de granularidade
e interpretação identificadas.

**Status:** E5.1 executado e validado.

**Próximas análises:** Dependentes de
avaliação metodológica e autorização.



## 5.2. Análise de desempenho por categorias

### Objetivo

Investigar a distribuição do portfólio e comparar
indicadores de preços, descontos, engajamento
e satisfação entre as categorias de produtos.

### Fundamentação

O enunciado do Tech Challenge Amazon Product
Performance & Customer Insights orienta a
identificação de categorias relevantes,
diferenças de satisfação e oportunidades
relacionadas à estratégia comercial.

Fonte: FIAP, Tech Challenge, pp. 3–5.

### Estrutura hierárquica

A coluna `category` apresenta classificações
hierárquicas separadas pelo caractere `|`.

A análise examinará:

1. Categoria principal: primeiro nível.
2. Caminho hierárquico completo: valor original.

A extração do primeiro nível será realizada
apenas em memória, sem modificar o dataset.

### Indicadores

- Número de produtos distintos.
- Número de registros.
- Participação no portfólio de produtos distintos.
- Mediana do preço com desconto, por registro.
- Desconto médio informado, por registro.
- Rating médio simples, por produto distinto.
- Mediana das contagens de avaliações, por registro.

### Critérios metodológicos

As categorias serão verificadas quanto à
consistência dentro de cada `product_id`.

Os preços e descontos permanecerão calculados
por registro, devido à existência de produtos
repetidos com possíveis diferenças comerciais.

A satisfação será calculada por produto distinto,
considerando cada produto com rating válido uma vez.

Para rankings exploratórios de satisfação por
caminho completo, será adotado um mínimo de
dez produtos com rating válido.

### Validação e rastreabilidade

- Conferir SHA-256 da base original e T4.5.
- Preservar os 1.465 registros e 22 colunas.
- Conciliar 1.351 produtos distintos.
- Conciliar 211 caminhos completos.
- Validar a estrutura das categorias.
- Identificar denominadores de cada indicador.
- Não realizar tratamentos persistentes.

### Limitações

Os indicadores comerciais calculados por
registro podem atribuir maior peso a produtos
com identificadores repetidos.

Diferenças entre categorias são descritivas,
não constituindo evidências de causalidade.

As contagens de avaliações não serão somadas,
pois não foi comprovada sua independência.

A análise não contém dados de vendas ou receita.

Nenhuma transformação ou novo arquivo será criado.


```python

# FIAP - TECH CHALLENGE FASE 04
# E5.2 - Analise descritiva por categorias
# SOMENTE LEITURA - nenhum arquivo sera modificado.

from pathlib import Path
from decimal import Decimal, InvalidOperation
import hashlib
import pandas as pd

# 1. Integridade

raiz = Path("/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON")

original = raiz / "01_DADOS_ORIGINAIS" / "01.amazon.csv"

entrada = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_05_desconto_pct_v01.csv"
)

hash_original = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

hash_t45 = (
    "5EFDD045F3FF6165FDC5865305E744A1"
    "4E7227F82654DD23983613026194EAA6"
)

def sha256(arquivo):
    h = hashlib.sha256()
    with arquivo.open("rb") as f:
        for bloco in iter(lambda: f.read(1048576), b""):
            h.update(bloco)
    return h.hexdigest().upper()

for arquivo in [original, entrada]:
    if not arquivo.is_file():
        raise FileNotFoundError(arquivo)

assert sha256(original) == hash_original
assert sha256(entrada) == hash_t45

base = pd.read_csv(
    entrada,
    dtype=str,
    keep_default_na=False,
    encoding="utf-8"
)

assert base.shape == (1465, 22)

necessarias = [
    "product_id",
    "category",
    "actual_price_inr",
    "discounted_price_inr",
    "discount_rate",
    "rating_numerico",
    "rating_count_numerico"
]

if not all(c in base.columns for c in necessarias):
    raise ValueError("Colunas obrigatorias ausentes.")

print("=== E5.2 - CONTROLE INICIAL ===")
print("SHA-256 original: CONFERIDO")
print("SHA-256 T4.5: CONFERIDO")
print("Dimensoes:", base.shape)

# 2. Verificacao da estrutura hierarquica

if base["product_id"].eq("").any():
    raise ValueError("product_id vazio.")

if base["category"].eq("").any():
    raise ValueError("Categoria vazia.")

categorias = base["category"].unique()

for categoria in categorias:
    partes = categoria.split("|")

    if any(parte.strip() == "" for parte in partes):
        raise ValueError(
            f"Hierarquia inesperada: {categoria!r}"
        )

variacao_categoria = (
    base.groupby("product_id")["category"].nunique()
)

if (variacao_categoria > 1).any():
    raise ValueError(
        "Mesmo produto apresenta categorias divergentes."
    )

variacao_rating = (
    base.groupby("product_id")["rating_numerico"]
    .nunique(dropna=False)
)

if (variacao_rating > 1).any():
    raise ValueError(
        "Mesmo produto apresenta ratings divergentes."
    )

# 3. Estrutura analitica temporaria

dados = base.copy(deep=True)

dados["categoria_principal"] = (
    dados["category"]
    .str.split("|", regex=False)
    .str[0]
)

def converter_decimal(texto, permitir_vazio=False):
    if texto == "":
        if permitir_vazio:
            return None
        raise ValueError("Valor numerico vazio inesperado.")

    try:
        valor = Decimal(texto)
    except InvalidOperation:
        raise ValueError(f"Numero invalido: {texto!r}")

    if not valor.is_finite():
        raise ValueError("Valor numerico nao finito.")

    return valor

conversoes = {
    "actual_price_inr": False,
    "discounted_price_inr": False,
    "discount_rate": False,
    "rating_numerico": True,
    "rating_count_numerico": True
}

for coluna, permite_vazio in conversoes.items():
    dados[coluna] = dados[coluna].map(
        lambda x: converter_decimal(x, permite_vazio)
    )

if dados["rating_numerico"].isna().sum() != 1:
    raise ValueError("Quantidade de ratings ausentes divergente.")

if dados["rating_count_numerico"].isna().sum() != 2:
    raise ValueError("Quantidade de contagens ausentes divergente.")

if not dados["discount_rate"].map(
    lambda x: Decimal("0") <= x <= Decimal("1")
).all():
    raise ValueError("Taxa de desconto fora da faixa.")

# Produtos distintos somente na memoria

produtos = dados.drop_duplicates(
    subset=["product_id"]
).copy()

if len(produtos) != 1351:
    raise ValueError("Produtos distintos divergentes.")

if dados["category"].nunique() != 211:
    raise ValueError("Caminhos de categoria divergentes.")

# 4. Funcoes de estatistica Decimal

def valores_validos(serie):
    return [v for v in serie if v is not None]

def media_decimal(valores):
    if not valores:
        return None
    return sum(valores, Decimal("0")) / Decimal(len(valores))

def mediana_decimal(valores):
    if not valores:
        return None

    ordem = sorted(valores)
    n = len(ordem)

    return (
        ordem[(n - 1) // 2] + ordem[n // 2]
    ) / Decimal("2")

def formatar(valor, casas=2):
    if valor is None:
        return "N/D"
    return f"{valor:,.{casas}f}"

# 5. EDA07 - Categorias principais

print("\n=== EDA07: CATEGORIAS PRINCIPAIS ===")
print(
    "Categorias principais distintas:",
    dados["categoria_principal"].nunique()
)
print("Caminhos completos distintos:", 211)
print("Produtos distintos:", len(produtos))

resumo_principal = []

for nome, grupo_registros in dados.groupby(
    "categoria_principal", sort=True
):
    grupo_produtos = produtos[
        produtos["categoria_principal"] == nome
    ]

    n_produtos = len(grupo_produtos)
    n_registros = len(grupo_registros)

    ratings = valores_validos(
        grupo_produtos["rating_numerico"]
    )

    contagens = valores_validos(
        grupo_registros["rating_count_numerico"]
    )

    resumo_principal.append({
        "categoria": nome,
        "produtos": n_produtos,
        "registros": n_registros,
        "participacao": (
            Decimal(n_produtos) * 100 / Decimal(1351)
        ),
        "preco_mediano": mediana_decimal(
            list(grupo_registros["discounted_price_inr"])
        ),
        "desconto_medio": media_decimal(
            [
                x * Decimal("100")
                for x in grupo_registros["discount_rate"]
            ]
        ),
        "rating_medio": media_decimal(ratings),
        "n_rating": len(ratings),
        "engajamento_mediano": mediana_decimal(contagens)
    })

resumo_principal.sort(
    key=lambda r: (-r["produtos"], r["categoria"])
)

print("\nIndicadores por categoria principal:")
print(
    "Categoria | Produtos | Registros | % produtos | "
    "Preco mediano INR | Desconto medio % | "
    "Rating medio | N rating | Mediana rating_count"
)

for r in resumo_principal:
    print(
        f"{r['categoria']} | "
        f"{r['produtos']} | "
        f"{r['registros']} | "
        f"{formatar(r['participacao'])}% | "
        f"{formatar(r['preco_mediano'])} | "
        f"{formatar(r['desconto_medio'])}% | "
        f"{formatar(r['rating_medio'])} | "
        f"{r['n_rating']} | "
        f"{formatar(r['engajamento_mediano'], 0)}"
    )

if sum(r["produtos"] for r in resumo_principal) != 1351:
    raise AssertionError("Soma de produtos divergente.")

if sum(r["registros"] for r in resumo_principal) != 1465:
    raise AssertionError("Soma de registros divergente.")

# 6. EDA08 - Categorias completas mais representadas

print("\n=== EDA08: CAMINHOS COMPLETOS ===")

resumo_detalhado = []

for caminho, grupo_produtos in produtos.groupby(
    "category", sort=True
):
    ratings = valores_validos(
        grupo_produtos["rating_numerico"]
    )

    resumo_detalhado.append({
        "categoria": caminho,
        "produtos": len(grupo_produtos),
        "ratings_validos": len(ratings),
        "rating_medio": media_decimal(ratings)
    })

resumo_detalhado.sort(
    key=lambda r: (-r["produtos"], r["categoria"])
)

print("\n10 caminhos com mais produtos distintos:")

for r in resumo_detalhado[:10]:
    print(
        f"{r['produtos']} produtos | "
        f"rating medio={formatar(r['rating_medio'])} | "
        f"{r['categoria']}"
    )

# 7. EDA09 - Satisfacao por caminho completo
# Corte exploratorio: pelo menos 10 ratings validos.
# Nenhum registro e eliminado do dataset.

elegiveis = [
    r for r in resumo_detalhado
    if r["ratings_validos"] >= 10
]

print("\n=== EDA09: SATISFACAO POR CAMINHO ===")
print("Caminhos completos totais:", len(resumo_detalhado))
print(
    "Caminhos com >= 10 produtos avaliados:",
    len(elegiveis)
)

melhores = sorted(
    elegiveis,
    key=lambda r: (
        -r["rating_medio"],
        -r["ratings_validos"],
        r["categoria"]
    )
)

piores = sorted(
    elegiveis,
    key=lambda r: (
        r["rating_medio"],
        -r["ratings_validos"],
        r["categoria"]
    )
)

print("\n5 maiores ratings medios:")
for r in melhores[:5]:
    print(
        f"{formatar(r['rating_medio'])} | "
        f"{r['ratings_validos']} avaliados | "
        f"{r['categoria']}"
    )

print("\n5 menores ratings medios:")
for r in piores[:5]:
    print(
        f"{formatar(r['rating_medio'])} | "
        f"{r['ratings_validos']} avaliados | "
        f"{r['categoria']}"
    )

# 8. EDA10 - Comparacao comercial
# Rankings por categoria principal.
# Precos e descontos permanecem por registro.

print("\n=== EDA10: COMPARACAO COMERCIAL ===")

print("\n3 maiores precos medianos com desconto (INR):")

for r in sorted(
    resumo_principal,
    key=lambda r: -r["preco_mediano"]
)[:3]:
    print(
        f"{r['categoria']} | "
        f"{formatar(r['preco_mediano'])} INR | "
        f"{r['registros']} registros"
    )

print("\n3 maiores descontos medios informados:")

for r in sorted(
    resumo_principal,
    key=lambda r: -r["desconto_medio"]
)[:3]:
    print(
        f"{r['categoria']} | "
        f"{formatar(r['desconto_medio'])}% | "
        f"{r['registros']} registros"
    )

# 9. Controle final

assert sha256(original) == hash_original
assert sha256(entrada) == hash_t45
assert base.shape == (1465, 22)

print("\n=== FIM DO E5.2 ===")
print("Registros da fonte:", len(base))
print("Produtos distintos:", len(produtos))
print("Categorias completas:", len(resumo_detalhado))
print("SHA-256 original: PRESERVADO")
print("SHA-256 T4.5: PRESERVADO")
print("Nenhum tratamento realizado.")
print("Nenhum arquivo derivado criado.")
```

## 5.3. Análise exploratória de preços, descontos e satisfação

### Objetivo

Investigar possíveis associações entre os preços,
os percentuais de desconto e as avaliações
dos produtos do Amazon Sales Dataset.

### Fundamentação

O enunciado do Tech Challenge Amazon orienta
investigar a relação entre descontos e avaliações,
bem como padrões de comportamento comercial
e satisfação dos consumidores.

Fonte: FIAP, Tech Challenge Amazon, pp. 3–4.

### Unidade de análise

Será utilizada uma observação por produto distinto,
identificado por `product_id`.

Como existem identificadores repetidos,
os preços e descontos de cada produto serão
representados temporariamente por suas medianas.

O rating será considerado uma vez por produto,
após verificação de consistência.

Essa consolidação possui finalidade exclusivamente
analítica e não será persistida em arquivo.

### Procedimentos

- EDA11: verificar a variabilidade comercial
  dos produtos com identificadores repetidos.
- EDA12: comparar a satisfação por faixas
  de descontos.
- EDA13: comparar a satisfação por quartis
  de preços com desconto.
- EDA14: calcular correlações de Spearman
  entre preços, descontos e ratings.
- EDA15: examinar as correlações nas três
  categorias principais mais representativas.

### Critérios metodológicos

A correlação de Spearman será utilizada
como medida descritiva da intensidade e direção
de associações monotônicas.

Não serão realizadas inferências causais.

As faixas de desconto serão:

- 0% a menos de 25%;
- 25% a menos de 50%;
- 50% a menos de 75%;
- 75% a 100%.

Os quartis de preço serão calculados a partir
das medianas dos preços por produto.

### Validações

1. Conferir SHA-256 da fonte original e T4.5.
2. Conciliar 1.465 registros e 1.351 produtos.
3. Confirmar a consistência dos ratings.
4. Preservar a ausência de rating de um produto.
5. Explicitar os denominadores das análises.
6. Não alterar arquivos ou criar novas versões.

### Limitações

A análise é observacional e descritiva.

A mediana por produto resume os registros
repetidos e não representa necessariamente
uma oferta individual específica.

Os resultados não demonstram que descontos
causem mudanças na satisfação.

A análise não inclui dados de vendas,
receita ou histórico temporal.

Nenhum tratamento persistente será realizado.


```python

# FIAP - TECH CHALLENGE FASE 04
# E5.3 - Precos, descontos e satisfacao
# SOMENTE LEITURA - nenhuma alteracao persistente.

from pathlib import Path
from decimal import Decimal, InvalidOperation
from collections import Counter
import hashlib
import pandas as pd

# 1. Fontes e verificacao de integridade

raiz = Path("/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON")

original = raiz / "01_DADOS_ORIGINAIS" / "01.amazon.csv"

entrada = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_05_desconto_pct_v01.csv"
)

hash_original = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

hash_t45 = (
    "5EFDD045F3FF6165FDC5865305E744A1"
    "4E7227F82654DD23983613026194EAA6"
)

def sha256(caminho):
    h = hashlib.sha256()
    with caminho.open("rb") as f:
        for bloco in iter(lambda: f.read(1048576), b""):
            h.update(bloco)
    return h.hexdigest().upper()

for caminho in [original, entrada]:
    if not caminho.is_file():
        raise FileNotFoundError(caminho)

if sha256(original) != hash_original:
    raise ValueError("SHA-256 original divergente.")

if sha256(entrada) != hash_t45:
    raise ValueError("SHA-256 T4.5 divergente.")

base = pd.read_csv(
    entrada,
    dtype=str,
    keep_default_na=False,
    encoding="utf-8"
)

if base.shape != (1465, 22):
    raise ValueError(f"Dimensoes inesperadas: {base.shape}")

campos = [
    "product_id",
    "category",
    "actual_price_inr",
    "discounted_price_inr",
    "discount_rate",
    "rating_numerico"
]

if not all(c in base.columns for c in campos):
    raise ValueError("Colunas necessarias ausentes.")

print("=== E5.3 - CONTROLE INICIAL ===")
print("SHA-256 original: CONFERIDO")
print("SHA-256 T4.5: CONFERIDO")
print("Registros:", len(base))
print("Colunas:", len(base.columns))

# 2. Conversao numerica temporaria

def decimal_seguro(texto, aceita_vazio=False):
    if texto == "":
        if aceita_vazio:
            return None
        raise ValueError("Valor obrigatorio vazio.")

    try:
        numero = Decimal(texto)
    except InvalidOperation:
        raise ValueError(f"Valor invalido: {texto!r}")

    if not numero.is_finite():
        raise ValueError("Numero nao finito.")

    return numero

dados = base.copy(deep=True)

dados["preco_original"] = dados[
    "actual_price_inr"
].map(decimal_seguro)

dados["preco_descontado"] = dados[
    "discounted_price_inr"
].map(decimal_seguro)

dados["taxa_desconto"] = dados[
    "discount_rate"
].map(decimal_seguro)

dados["rating_num"] = dados[
    "rating_numerico"
].map(lambda x: decimal_seguro(x, True))

dados["categoria_principal"] = (
    dados["category"]
    .str.split("|", regex=False)
    .str[0]
)

if dados["rating_num"].isna().sum() != 1:
    raise ValueError("Ausencia de rating divergente.")

for _, linha in dados.iterrows():
    original_preco = linha["preco_original"]
    preco_desc = linha["preco_descontado"]
    taxa = linha["taxa_desconto"]
    rating = linha["rating_num"]

    if not (
        Decimal("0") < original_preco
        and Decimal("0") <= preco_desc <= original_preco
    ):
        raise ValueError("Preco comercial inesperado.")

    if not Decimal("0") <= taxa <= Decimal("1"):
        raise ValueError("Desconto fora da faixa.")

    if rating is not None and not (
        Decimal("0") <= rating <= Decimal("5")
    ):
        raise ValueError("Rating fora da faixa.")

# 3. Funcoes estatisticas

def mediana_decimal(valores):
    valores = sorted(valores)
    n = len(valores)

    if n == 0:
        return None

    return (
        valores[(n - 1) // 2]
        + valores[n // 2]
    ) / Decimal("2")

def media_decimal(valores):
    if not valores:
        return None

    return sum(
        valores, Decimal("0")
    ) / Decimal(len(valores))

def formatar(valor, casas=2):
    if valor is None:
        return "N/D"
    return f"{valor:,.{casas}f}"

# 4. EDA11 - Produtos repetidos e valores comerciais

produtos_analiticos = []
variacao_preco = 0
variacao_desconto = 0
grupos_repetidos = 0

for product_id, grupo in dados.groupby(
    "product_id", sort=False
):
    if not product_id:
        raise ValueError("Identificador de produto vazio.")

    if grupo["category"].nunique() != 1:
        raise ValueError(
            "Categoria divergente para o mesmo produto."
        )

    if grupo["rating_num"].nunique(dropna=False) != 1:
        raise ValueError(
            "Rating divergente para o mesmo produto."
        )

    if len(grupo) > 1:
        grupos_repetidos += 1

        if grupo["preco_descontado"].nunique() > 1:
            variacao_preco += 1

        if grupo["taxa_desconto"].nunique() > 1:
            variacao_desconto += 1

    produtos_analiticos.append({
        "product_id": product_id,
        "categoria_principal":
            grupo["categoria_principal"].iloc[0],
        "n_registros": len(grupo),
        "preco_original": mediana_decimal(
            list(grupo["preco_original"])
        ),
        "preco_descontado": mediana_decimal(
            list(grupo["preco_descontado"])
        ),
        "desconto_pct": mediana_decimal(
            list(grupo["taxa_desconto"])
        ) * Decimal("100"),
        "rating": grupo["rating_num"].iloc[0]
    })

if len(produtos_analiticos) != 1351:
    raise ValueError("Quantidade de produtos divergente.")

if grupos_repetidos != 92:
    raise ValueError("Grupos repetidos divergentes.")

com_rating = [
    p for p in produtos_analiticos
    if p["rating"] is not None
]

if len(com_rating) != 1350:
    raise ValueError("Quantidade de ratings divergente.")

print("\n=== EDA11: VARIABILIDADE POR PRODUTO ===")
print("Produtos distintos:", len(produtos_analiticos))
print("Produtos com ID repetido:", grupos_repetidos)
print(
    "Grupos repetidos com variacao no preco descontado:",
    variacao_preco
)
print(
    "Grupos repetidos com variacao no desconto:",
    variacao_desconto
)
print("Produtos com rating valido:", len(com_rating))
print("Produtos sem rating:", 1)

# 5. EDA12 - Satisfacao por faixa de desconto

def faixa_desconto(percentual):
    if percentual < Decimal("25"):
        return "0% a <25%"
    if percentual < Decimal("50"):
        return "25% a <50%"
    if percentual < Decimal("75"):
        return "50% a <75%"
    return "75% a 100%"

faixas_desconto = [
    "0% a <25%",
    "25% a <50%",
    "50% a <75%",
    "75% a 100%"
]

print("\n=== EDA12: SATISFACAO POR FAIXA DE DESCONTO ===")
print(
    "Faixa | Produtos | Ratings validos | "
    "Rating medio | Preco descontado mediano INR"
)

total_faixas = 0

for faixa in faixas_desconto:
    grupo = [
        p for p in produtos_analiticos
        if faixa_desconto(p["desconto_pct"]) == faixa
    ]

    ratings = [
        p["rating"] for p in grupo
        if p["rating"] is not None
    ]

    precos = [p["preco_descontado"] for p in grupo]

    total_faixas += len(grupo)

    print(
        f"{faixa} | "
        f"{len(grupo)} | "
        f"{len(ratings)} | "
        f"{formatar(media_decimal(ratings))} | "
        f"{formatar(mediana_decimal(precos))}"
    )

if total_faixas != 1351:
    raise AssertionError(
        "Soma das faixas de desconto divergente."
    )

# 6. EDA13 - Quartis de preco com desconto

def quantil_decimal(valores, numerador, denominador=4):
    ordem = sorted(valores)
    n = len(ordem)

    pos = (
        Decimal(n - 1)
        * Decimal(numerador)
        / Decimal(denominador)
    )

    inferior = int(pos)
    superior = min(inferior + 1, n - 1)
    fracao = pos - Decimal(inferior)

    return (
        ordem[inferior]
        + (ordem[superior] - ordem[inferior]) * fracao
    )

todos_precos = [
    p["preco_descontado"]
    for p in produtos_analiticos
]

q1 = quantil_decimal(todos_precos, 1)
q2 = quantil_decimal(todos_precos, 2)
q3 = quantil_decimal(todos_precos, 3)

def faixa_preco(preco):
    if preco <= q1:
        return "Q1 - Menores precos"
    if preco <= q2:
        return "Q2"
    if preco <= q3:
        return "Q3"
    return "Q4 - Maiores precos"

faixas_preco = [
    "Q1 - Menores precos",
    "Q2",
    "Q3",
    "Q4 - Maiores precos"
]

print("\n=== EDA13: SATISFACAO POR QUARTIL DE PRECO ===")
print("Limite Q1 (INR):", formatar(q1))
print("Mediana Q2 (INR):", formatar(q2))
print("Limite Q3 (INR):", formatar(q3))

total_quartis = 0

for faixa in faixas_preco:
    grupo = [
        p for p in produtos_analiticos
        if faixa_preco(p["preco_descontado"]) == faixa
    ]

    ratings = [
        p["rating"] for p in grupo
        if p["rating"] is not None
    ]

    total_quartis += len(grupo)

    print(
        f"{faixa} | "
        f"produtos={len(grupo)} | "
        f"ratings_validos={len(ratings)} | "
        f"rating_medio={formatar(media_decimal(ratings))}"
    )

if total_quartis != 1351:
    raise AssertionError(
        "Soma dos quartis de preco divergente."
    )

# 7. Correlacao de Spearman com postos medios
# para empates, calculados sem converter
# precos e descontos para float.

def postos_medios(valores):
    ordem = sorted(
        range(len(valores)),
        key=lambda i: valores[i]
    )

    postos = [0.0] * len(valores)
    inicio = 0

    while inicio < len(ordem):
        fim = inicio + 1

        while (
            fim < len(ordem)
            and valores[ordem[fim]] == valores[ordem[inicio]]
        ):
            fim += 1

        posto_medio = ((inicio + 1) + fim) / 2

        for j in range(inicio, fim):
            postos[ordem[j]] = posto_medio

        inicio = fim

    return postos

def spearman_produtos(amostra, variavel):
    validos = [
        p for p in amostra
        if p["rating"] is not None
    ]

    n = len(validos)

    if n < 3:
        return n, None

    x = [p[variavel] for p in validos]
    y = [p["rating"] for p in validos]

    if len(set(x)) < 2 or len(set(y)) < 2:
        return n, None

    postos_x = pd.Series(
        postos_medios(x), dtype="float64"
    )

    postos_y = pd.Series(
        postos_medios(y), dtype="float64"
    )

    rho = postos_x.corr(postos_y)

    if pd.isna(rho) or not (-1 <= rho <= 1):
        raise ValueError("Correlacao invalida.")

    return n, float(rho)

def imprimir_correlacao(nome, amostra):
    n_preco, rho_preco = spearman_produtos(
        amostra, "preco_descontado"
    )

    n_desc, rho_desc = spearman_produtos(
        amostra, "desconto_pct"
    )

    print(f"\n{nome}")
    print("Produtos com rating valido:", n_preco)

    print(
        "Spearman preco descontado x rating:",
        f"{rho_preco:.4f}"
        if rho_preco is not None else "N/D"
    )

    print(
        "Spearman desconto x rating:",
        f"{rho_desc:.4f}"
        if rho_desc is not None else "N/D"
    )

    if n_preco != n_desc:
        raise AssertionError("Denominadores divergentes.")

print("\n=== EDA14: CORRELACOES GERAIS ===")
imprimir_correlacao(
    "Todos os produtos distintos",
    produtos_analiticos
)

# 8. EDA15 - Correlacoes nas principais categorias

print("\n=== EDA15: CORRELACOES POR CATEGORIA ===")

contagem_categorias = Counter(
    p["categoria_principal"]
    for p in produtos_analiticos
)

principais = [
    categoria for categoria, _ in
    contagem_categorias.most_common(3)
]

for categoria in principais:
    amostra = [
        p for p in produtos_analiticos
        if p["categoria_principal"] == categoria
    ]

    imprimir_correlacao(categoria, amostra)

# 9. Controles finais

if sha256(original) != hash_original:
    raise AssertionError("Base original alterada.")

if sha256(entrada) != hash_t45:
    raise AssertionError("Base T4.5 alterada.")

print("\n=== FIM DO E5.3 ===")
print("Registros da fonte:", len(base))
print("Produtos distintos:", len(produtos_analiticos))
print("Produtos com rating valido:", len(com_rating))
print("SHA-256 original: PRESERVADO")
print("SHA-256 T4.5: PRESERVADO")
print("Nenhum tratamento realizado.")
print("Nenhum arquivo derivado criado.")
```


### 5.3.1. Resultados e interpretação de negócio

# 6.&nbsp;Customer Insights — Análise dos comentários




## 6.1. Diagnóstico da qualidade textual

### Objetivo

Avaliar as características e limitações dos
comentários presentes em `review_content`,
preparando a investigação qualitativa da
experiência dos consumidores.

### Fundamentação

O enunciado do Tech Challenge Amazon orienta
explorar o conteúdo das avaliações para
identificar opiniões, aspectos valorizados
e problemas relatados pelos consumidores.

Fonte: FIAP, Tech Challenge Amazon, pp. 4–5.

### Procedimento

O diagnóstico examinará:

1. Textos preenchidos e vazios.
2. Comprimento dos comentários.
3. Repetição de textos completos.
4. Repetições entre produtos diferentes.
5. Características estruturais e sinais
   que possam dificultar a análise textual.
6. Amostra para inspeção qualitativa.

### Cuidados metodológicos

Não serão separados comentários concatenados
sem identificação e validação do delimitador.

A ocorrência de vírgulas ou outros caracteres
não comprova a existência de um delimitador.

Os identificadores de usuários e avaliações
não serão interpretados como registros
individuais já separados.

A identificação de idioma não será inferida
apenas pela presença de caracteres ASCII.

### Integridade e rastreabilidade

Fonte de análise:

`02_DADOS_TRATADOS/amazon_t4_05_desconto_pct_v01.csv`

Dimensões esperadas: 1.465 registros e 22 colunas.

Os arquivos original e tratado serão
validados por SHA-256 antes e depois da análise.

Nenhum arquivo será criado ou alterado.

### Resultado esperado

Produzir evidências sobre a qualidade,
a estrutura e as limitações dos comentários,
fundamentando uma análise textual posterior.

Não será realizado tratamento dos dados
nem classificação de sentimentos nesta etapa.


```python

# FIAP - TECH CHALLENGE FASE 04
# E6.1 - Diagnostico dos comentarios
# SOMENTE LEITURA - nenhum arquivo sera alterado.

from pathlib import Path
from collections import Counter
from statistics import median
import hashlib
import random
import re
import pandas as pd

# 1. Caminhos e verificacao de integridade

raiz = Path("/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON")

original = raiz / "01_DADOS_ORIGINAIS" / "01.amazon.csv"

entrada = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_05_desconto_pct_v01.csv"
)

hash_original = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

hash_t45 = (
    "5EFDD045F3FF6165FDC5865305E744A1"
    "4E7227F82654DD23983613026194EAA6"
)

def sha256(caminho):
    h = hashlib.sha256()

    with caminho.open("rb") as arquivo:
        for bloco in iter(
            lambda: arquivo.read(1048576), b""
        ):
            h.update(bloco)

    return h.hexdigest().upper()

for caminho in [original, entrada]:
    if not caminho.is_file():
        raise FileNotFoundError(caminho)

if sha256(original) != hash_original:
    raise ValueError("SHA-256 original divergente.")

if sha256(entrada) != hash_t45:
    raise ValueError("SHA-256 T4.5 divergente.")

base = pd.read_csv(
    entrada,
    dtype=str,
    keep_default_na=False,
    encoding="utf-8"
)

if base.shape != (1465, 22):
    raise ValueError(
        f"Dimensoes inesperadas: {base.shape}"
    )

campos = [
    "product_id",
    "review_id",
    "review_title",
    "review_content"
]

if not all(c in base.columns for c in campos):
    raise ValueError("Colunas necessarias ausentes.")

print("=== E6.1 - CONTROLE INICIAL ===")
print("SHA-256 original: CONFERIDO")
print("SHA-256 T4.5: CONFERIDO")
print("Registros:", len(base))
print("Colunas:", len(base.columns))

# 2. Preparacao temporaria dos textos
# str.strip() e usado apenas para diagnostico.
# Nenhuma coluna do arquivo e modificada.

textos_originais = base["review_content"]

textos = textos_originais.map(
    lambda valor: valor.strip()
)

titulos = base["review_title"].map(
    lambda valor: valor.strip()
)

validos = textos[textos.ne("")]

total = len(base)
n_validos = len(validos)
n_vazios = total - n_validos

percentual = lambda n: 100 * n / total

print("\n=== NLP01: PREENCHIMENTO ===")
print("Total de registros:", total)
print(
    "Comentarios preenchidos:",
    n_validos,
    f"({percentual(n_validos):.2f}%)"
)
print(
    "Comentarios vazios:",
    n_vazios,
    f"({percentual(n_vazios):.2f}%)"
)
print(
    "Valores somente com espacos:",
    int(
        (
            textos_originais.ne("")
            & textos.eq("")
        ).sum()
    )
)
print(
    "Titulos de review preenchidos:",
    int(titulos.ne("").sum())
)
print(
    "Titulos de review vazios:",
    int(titulos.eq("").sum())
)

# 3. Distribuicao dos comprimentos

comprimentos = validos.map(len)

# Numero aproximado de palavras.
# Nao e tokenizacao linguistica definitiva.
palavras = validos.map(
    lambda texto: len(
        re.findall(r"\b\w+\b", texto, flags=re.UNICODE)
    )
)

print("\n=== NLP02: COMPRIMENTO DOS TEXTOS ===")

if n_validos:
    print(
        "Caracteres - minimo:",
        int(comprimentos.min())
    )
    print(
        "Caracteres - media:",
        f"{comprimentos.mean():.2f}"
    )
    print(
        "Caracteres - mediana:",
        f"{comprimentos.median():.2f}"
    )
    print(
        "Caracteres - Q1:",
        f"{comprimentos.quantile(.25):.2f}"
    )
    print(
        "Caracteres - Q3:",
        f"{comprimentos.quantile(.75):.2f}"
    )
    print(
        "Caracteres - maximo:",
        int(comprimentos.max())
    )
    print(
        "Palavras aproximadas - mediana:",
        f"{palavras.median():.2f}"
    )
else:
    print("Nenhum comentario valido para analisar.")

# 4. Repeticao de conteudo completo

frequencias = Counter(validos.tolist())

grupos_repetidos = sum(
    quantidade > 1
    for quantidade in frequencias.values()
)

ocorrencias_excedentes = sum(
    quantidade - 1
    for quantidade in frequencias.values()
    if quantidade > 1
)

textos_com_repeticao = sum(
    quantidade
    for quantidade in frequencias.values()
    if quantidade > 1
)

# Identifica textos completos presentes em
# mais de um product_id, sem atribuir causas.

auxiliar = pd.DataFrame({
    "texto": textos,
    "produto": base["product_id"]
})

auxiliar = auxiliar[auxiliar["texto"].ne("")]

textos_multiplos_produtos = (
    auxiliar.groupby("texto")["produto"]
    .nunique()
)

textos_entre_produtos = int(
    (textos_multiplos_produtos > 1).sum()
)

print("\n=== NLP03: REPETICAO DE TEXTOS ===")
print("Textos completos distintos:", len(frequencias))
print(
    "Grupos de textos completos repetidos:",
    grupos_repetidos
)
print(
    "Registros envolvidos em repeticoes:",
    textos_com_repeticao
)
print(
    "Ocorrencias excedentes:",
    ocorrencias_excedentes
)
print(
    "Textos presentes em mais de um produto:",
    textos_entre_produtos
)

# 5. Indicadores estruturais
# Sao sinais, nao provas de concatenacao.

com_virgula = validos.map(
    lambda texto: "," in texto
).sum()

com_quebra_linha = validos.map(
    lambda texto: "\n" in texto or "\r" in texto
).sum()

com_ponto_virgula = validos.map(
    lambda texto: ";" in texto
).sum()

com_caractere_nao_ascii = validos.map(
    lambda texto: any(ord(c) > 127 for c in texto)
).sum()

com_html_aparente = validos.map(
    lambda texto: bool(
        re.search(r"<[^>]{1,80}>", texto)
    )
).sum()

com_url_aparente = validos.map(
    lambda texto: bool(
        re.search(r"https?://|www\.", texto, re.I)
    )
).sum()

review_ids_com_virgula = int(
    base["review_id"].map(
        lambda texto: "," in texto
    ).sum()
)

print("\n=== NLP04: INDICADORES ESTRUTURAIS ===")
print("Comentarios com virgula:", int(com_virgula))
print(
    "Comentarios com quebra de linha:",
    int(com_quebra_linha)
)
print(
    "Comentarios com ponto e virgula:",
    int(com_ponto_virgula)
)
print(
    "Comentarios com caracteres nao ASCII:",
    int(com_caractere_nao_ascii)
)
print(
    "Comentarios com HTML aparente:",
    int(com_html_aparente)
)
print(
    "Comentarios com URL aparente:",
    int(com_url_aparente)
)
print(
    "Registros com virgula em review_id:",
    review_ids_com_virgula
)

print(
    "Delimitador de comentarios: NAO VALIDADO"
)
print(
    "Idioma dos comentarios: NAO DETERMINADO"
)
print(
    "Nenhum comentario foi separado."
)

# 6. Amostra reprodutivel para inspecao
# Exibe apenas trechos curtos, sem IDs pessoais.
# URLs e emails sao mascarados nos exemplos.

print("\n=== NLP05: AMOSTRA QUALITATIVA ===")

if n_validos:
    indices = validos.index.tolist()

    amostra_aleatoria = random.Random(42).sample(
        indices, min(4, len(indices))
    )

    menor_indice = min(
        indices,
        key=lambda i: len(textos.loc[i])
    )

    maior_indice = max(
        indices,
        key=lambda i: len(textos.loc[i])
    )

    amostra = sorted(set(
        amostra_aleatoria
        + [indices[0], menor_indice, maior_indice]
    ))

    for indice in amostra:
        trecho = textos.loc[indice]

        trecho = re.sub(
            r"https?://\S+|www\.\S+",
            "[URL]",
            trecho,
            flags=re.I
        )

        trecho = re.sub(
            r"[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}",
            "[EMAIL]",
            trecho
        )

        trecho = " ".join(trecho.split())

        print(
            f"Registro {indice + 1} | "
            f"caracteres={len(textos.loc[indice])} | "
            f"trecho={trecho[:130]!r}"
        )

print(
    "\nNota: a amostra e exploratoria; "
    "nao representa validacao automatica do idioma."
)

# 7. Controle final

if sha256(original) != hash_original:
    raise AssertionError("Arquivo original alterado.")

if sha256(entrada) != hash_t45:
    raise AssertionError("Arquivo T4.5 alterado.")

print("\n=== FIM DO E6.1 ===")
print("Registros da fonte:", len(base))
print("Textos preenchidos:", n_validos)
print("SHA-256 original: PRESERVADO")
print("SHA-256 T4.5: PRESERVADO")
print("Nenhum tratamento realizado.")
print("Nenhum arquivo derivado criado.")
```

## 6.2. Text Mining — Mineração de Temas e Percepções

### Objetivo

Identificar assuntos recorrentes nos comentários
dos consumidores e produzir evidências textuais
para a análise de Customer Insights.

### Fundamentação acadêmica

FIAP. Geração de insights.
Professor Edgard Joseph Kiriyama, página 5.

A técnica de Text Mining permite extrair
informações relevantes de textos não estruturados,
incluindo assuntos, sentimentos e entidades.

FIAP. Tech Challenge Amazon, páginas 4 e 5.

O desafio orienta investigar aspectos valorizados,
problemas relatados e padrões presentes nas
percepções dos consumidores.

### Método aprovado

Será utilizada mineração temática exploratória
baseada em expressões e regras transparentes.

Os temas iniciais serão:

1. Qualidade e durabilidade;
2. Funcionamento e desempenho;
3. Preço e custo-benefício;
4. Bateria e carregamento;
5. Compatibilidade e instalação;
6. Entrega, devolução e atendimento.

A identificação temática será acompanhada
de inspeção contextual de amostras.

### Unidade de análise

As frequências serão calculadas separadamente para:

- Registros textuais da base;
- Textos completos distintos;
- Produtos distintos associados a menções.

Essas unidades não equivalem ao número de
consumidores ou avaliações individuais.

Um mesmo texto poderá mencionar vários temas.

### Cuidados metodológicos

Os comentários originais serão preservados.

Não será realizada separação por vírgulas,
pois o delimitador não foi validado.

A identificação de termos indica menção
a um assunto, não necessariamente elogio
ou reclamação.

O vocabulário inicial será em inglês,
com possibilidade de revisão após
inspeção das evidências.

A classificação temática será considerada
provisória até a validação contextual.

### Critérios de validação

- Conferir os hashes da fonte original e T4.5.
- Conciliar 1.465 registros.
- Conciliar 1.212 textos completos distintos.
- Explicitar os denominadores das frequências.
- Examinar exemplos positivos e falsos positivos
  no sentido de correspondência temática.
- Registrar limitações e possíveis omissões.
- Não realizar tratamentos persistentes.

### Limitações

A correspondência de expressões não constitui
compreensão semântica integral.

O método pode produzir falsos positivos,
não reconhecer sinônimos e deixar de identificar
assuntos expressos de maneiras diferentes.

Blocos concatenados podem reunir opiniões
de consumidores distintos.

Não haverá classificação automática
de sentimento por comentário completo.

Os resultados serão inicialmente exploratórios.


```python

# FIAP - TECH CHALLENGE FASE 04
# E6.2 - Text Mining: temas e percepcoes
# DIAGNOSTICO EM MEMORIA - nenhum arquivo sera alterado.

from pathlib import Path
from collections import Counter
import hashlib
import random
import re
import pandas as pd

# 1. Fontes e integridade

raiz = Path("/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON")

original = raiz / "01_DADOS_ORIGINAIS" / "01.amazon.csv"

entrada = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_05_desconto_pct_v01.csv"
)

hash_original = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

hash_t45 = (
    "5EFDD045F3FF6165FDC5865305E744A1"
    "4E7227F82654DD23983613026194EAA6"
)

def sha256(caminho):
    h = hashlib.sha256()
    with caminho.open("rb") as f:
        for bloco in iter(lambda: f.read(1048576), b""):
            h.update(bloco)
    return h.hexdigest().upper()

for caminho in [original, entrada]:
    if not caminho.is_file():
        raise FileNotFoundError(caminho)

if sha256(original) != hash_original:
    raise ValueError("SHA-256 original divergente.")

if sha256(entrada) != hash_t45:
    raise ValueError("SHA-256 T4.5 divergente.")

base = pd.read_csv(
    entrada,
    dtype=str,
    keep_default_na=False,
    encoding="utf-8"
)

if base.shape != (1465, 22):
    raise ValueError("Dimensoes inesperadas.")

necessarias = [
    "product_id",
    "category",
    "review_content"
]

if not all(c in base.columns for c in necessarias):
    raise ValueError("Colunas obrigatorias ausentes.")

print("=== E6.2 - CONTROLE INICIAL ===")
print("SHA-256 original: CONFERIDO")
print("SHA-256 T4.5: CONFERIDO")
print("Registros:", len(base))
print("Colunas:", len(base.columns))

# 2. Corpus original e textos distintos
# strip() apenas em memoria, conforme E6.1.

textos = base["review_content"].map(str.strip).tolist()
produtos = base["product_id"].tolist()

if any(not texto for texto in textos):
    raise ValueError("Texto vazio inesperado.")

if any(not produto for produto in produtos):
    raise ValueError("Produto sem identificador.")

textos_unicos = list(dict.fromkeys(textos))

if len(textos_unicos) != 1212:
    raise ValueError(
        "Quantidade de textos distintos divergente do E6.1."
    )

if len(textos) - len(textos_unicos) != 253:
    raise ValueError(
        "Ocorrencias excedentes divergentes do E6.1."
    )

if len(set(produtos)) != 1351:
    raise ValueError("Produtos distintos divergentes.")

print("\n=== NLP06: UNIDADES DE ANALISE ===")
print("Registros textuais:", len(textos))
print("Textos completos distintos:", len(textos_unicos))
print("Ocorrencias excedentes:", 253)
print("Produtos distintos:", len(set(produtos)))

# 3. Vocabulário inicial de aspectos
# Hipoteses de busca: termos sujeitos a validacao.
# Nenhum termo implica sentimento automaticamente.

vocabulario = {
    "Qualidade e durabilidade": [
        "quality", "build quality", "durable",
        "durability", "scratch", "scratches",
        "damage", "damaged", "broken", "broke",
        "defect", "defective", "fragile", "material"
    ],
    "Funcionamento e desempenho": [
        "work", "works", "working",
        "function", "functions", "functioning",
        "performance", "perform", "performs",
        "speed", "slow", "fast",
        "not working", "stopped working",
        "heating", "cooling"
    ],
    "Preco e custo-beneficio": [
        "price", "priced", "value for money",
        "worth", "cost", "costly",
        "expensive", "cheap", "budget",
        "affordable", "money"
    ],
    "Bateria e carregamento": [
        "battery", "batteries", "charge",
        "charging", "charger", "charged",
        "power bank", "battery backup",
        "battery life"
    ],
    "Compatibilidade e instalacao": [
        "compatible", "compatibility",
        "connect", "connected", "connecting",
        "connection", "install", "installed",
        "installation", "setup", "set up",
        "bluetooth", "wifi", "wi-fi",
        "pairing", "pair", "driver", "port"
    ],
    "Entrega, devolucao e atendimento": [
        "delivery", "delivered", "shipping",
        "packaging", "package", "packed",
        "return", "returned", "refund",
        "replacement", "replace", "courier",
        "customer service", "seller"
    ]
}

def construir_padrao(termos):
    partes = []

    for termo in sorted(
        set(termos), key=len, reverse=True
    ):
        trecho = re.escape(termo)
        trecho = trecho.replace(r"\ ", r"\s+")
        partes.append(trecho)

    expressao = (
        r"(?<!\w)(?:" + "|".join(partes) + r")(?!\w)"
    )

    return re.compile(expressao, re.IGNORECASE)

padroes = {
    tema: construir_padrao(termos)
    for tema, termos in vocabulario.items()
}

# Espacos internos normalizados somente para busca.
textos_busca = {
    texto: re.sub(r"\s+", " ", texto)
    for texto in textos_unicos
}

# Classificacao de mencoes por bloco textual
# completo. Um bloco pode receber varios temas.

marcacoes = {}

for texto in textos_unicos:
    conteudo = textos_busca[texto]

    marcacoes[texto] = {
        tema
        for tema, padrao in padroes.items()
        if padrao.search(conteudo)
    }

# 4. Contagens sem tratar blocos como
# avaliacoes individuais.

print("\n=== NLP07: FREQUENCIA DE TEMAS ===")
print(
    "Tema | Registros | % registros | "
    "Textos distintos | % textos distintos | "
    "Produtos associados"
)

for tema in vocabulario:
    n_registros = sum(
        tema in marcacoes[texto]
        for texto in textos
    )

    n_unicos = sum(
        tema in marcacoes[texto]
        for texto in textos_unicos
    )

    produtos_associados = {
        produto
        for produto, texto in zip(produtos, textos)
        if tema in marcacoes[texto]
    }

    print(
        f"{tema} | "
        f"{n_registros} | "
        f"{100*n_registros/1465:.2f}% | "
        f"{n_unicos} | "
        f"{100*n_unicos/1212:.2f}% | "
        f"{len(produtos_associados)}"
    )

sem_tema = sum(
    not marcacoes[texto]
    for texto in textos_unicos
)

com_multiplos = sum(
    len(marcacoes[texto]) >= 2
    for texto in textos_unicos
)

print("\nTextos distintos sem tema identificado:", sem_tema)
print(
    "Textos distintos com dois ou mais temas:",
    com_multiplos
)

print(
    "\nObservacao: os temas nao sao exclusivos."
)
print(
    "A soma das frequencias pode superar "
    "o numero de textos."
)

# 5. Categorias principais e temas
# Produto associado = product_id com pelo
# menos um registro contendo o tema.

categorias_principais = [
    categoria.split("|")[0]
    for categoria in base["category"]
]

categorias_por_produto = {}

for produto, categoria in zip(
    produtos, categorias_principais
):
    if (
        produto in categorias_por_produto
        and categorias_por_produto[produto] != categoria
    ):
        raise ValueError(
            "Categoria principal divergente por produto."
        )

    categorias_por_produto[produto] = categoria

contagem_categorias = Counter(
    categorias_por_produto.values()
)

principais = [
    nome for nome, _ in
    contagem_categorias.most_common(3)
]

rotulos_pt = {
    "Electronics": "Eletronicos",
    "Home&Kitchen": "Casa e Cozinha",
    "Computers&Accessories":
        "Informatica e Acessorios"
}

print("\n=== NLP08: TEMAS NAS PRINCIPAIS CATEGORIAS ===")
print(
    "Categoria | Tema | Produtos associados | "
    "Total de produtos | % produtos"
)

for categoria in principais:
    total_produtos = contagem_categorias[categoria]

    for tema in vocabulario:
        associados = {
            produto
            for produto, texto, cat in zip(
                produtos, textos, categorias_principais
            )
            if (
                cat == categoria
                and tema in marcacoes[texto]
            )
        }

        print(
            f"{rotulos_pt.get(categoria, categoria)} | "
            f"{tema} | "
            f"{len(associados)} | "
            f"{total_produtos} | "
            f"{100*len(associados)/total_produtos:.2f}%"
        )

# 6. Evidencias contextuais para validacao humana
# Duas amostras por tema, semente fixa.
# Nao exibir identificadores pessoais.

def mascarar_trecho(texto):
    texto = re.sub(
        r"https?://\S+|www\.\S+",
        "[URL]",
        texto,
        flags=re.IGNORECASE
    )

    texto = re.sub(
        r"[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}",
        "[EMAIL]",
        texto
    )

    texto = re.sub(
        r"(?<!\w)\+?\d[\d().\s-]{8,}\d(?!\w)",
        "[NUMERO]",
        texto
    )

    return texto

gerador = random.Random(42)

print("\n=== NLP09: AMOSTRAS CONTEXTUAIS ===")
print(
    "Amostras de mencoes para avaliacao de "
    "relevancia e falsos positivos."
)

for tema, padrao in padroes.items():
    candidatos = [
        texto for texto in textos_unicos
        if tema in marcacoes[texto]
    ]

    amostra = gerador.sample(
        candidatos, min(2, len(candidatos))
    )

    print(f"\nTEMA: {tema}")
    print("Textos distintos identificados:", len(candidatos))

    for indice, texto in enumerate(amostra, start=1):
        conteudo = textos_busca[texto]
        achado = padrao.search(conteudo)

        if achado is None:
            raise AssertionError(
                "Correspondencia tematica nao recuperada."
            )

        inicio = max(0, achado.start() - 70)
        fim = min(
            len(conteudo),
            achado.end() + 100
        )

        trecho = mascarar_trecho(
            conteudo[inicio:fim]
        )

        print(
            f"  Exemplo {indice}: "
            f"termo={achado.group(0)!r} | "
            f"contexto={trecho!r}"
        )

# Inspecao adicional dos blocos sem temas
nao_classificados = [
    texto for texto in textos_unicos
    if not marcacoes[texto]
]

print("\n=== NLP10: TEXTOS SEM TEMA IDENTIFICADO ===")
print(
    "Total de textos distintos sem tema:",
    len(nao_classificados)
)

for i, texto in enumerate(
    gerador.sample(
        nao_classificados,
        min(4, len(nao_classificados))
    ),
    start=1
):
    trecho = mascarar_trecho(
        textos_busca[texto][:160]
    )
    print(f"Exemplo {i}: {trecho!r}")

# 7. Verificacoes finais

if sha256(original) != hash_original:
    raise AssertionError("Base original alterada.")

if sha256(entrada) != hash_t45:
    raise AssertionError("Base T4.5 alterada.")

print("\n=== FIM DO E6.2 ===")
print("Registros:", len(base))
print("Textos distintos:", len(textos_unicos))
print("Produtos distintos:", len(set(produtos)))
print("SHA-256 original: PRESERVADO")
print("SHA-256 T4.5: PRESERVADO")
print("Nenhum tratamento realizado.")
print("Nenhum arquivo derivado criado.")
print("Classificacao tematica: PROVISORIA")
print("Validacao contextual: PENDENTE")
```


### 6.2.1. Revisão metodológica E6.2-V1

#### Fundamentação

A execução inicial do E6.2 identificou
1.212 textos completos distintos e classificou
menções a seis temas da experiência do consumidor.

A inspeção contextual revelou ambiguidades,
incluindo a classificação da expressão
"return value of spent money" como relacionada
a devoluções.

Também foram identificados termos genéricos
que podem produzir correspondências indevidas,
como "work", "fast", "pair" e "port".

#### Objetivo da revisão

Revisar o vocabulário inicial, comparar
as frequências anteriores e posteriores
e examinar amostras de correspondências.

#### Procedimentos

1. Reproduzir as frequências do E6.2.
2. Aplicar alterações explícitas ao vocabulário.
3. Mascarar endereços e contatos nos textos
   temporários antes da busca por expressões.
4. Executar testes com frases de controle.
5. Comparar as frequências temáticas.
6. Examinar amostras de textos mantidos,
   retirados e acrescentados.
7. Inspecionar textos sem tema identificado.

#### Critérios de validação

- Integridade SHA-256 preservada.
- 1.465 registros e 22 variáveis.
- 1.212 textos completos distintos.
- Reprodução dos resultados iniciais.
- Registro das diferenças nas frequências.
- Ausência de transformações persistentes.

#### Limitações

A revisão não constitui treinamento de
modelo supervisionado.

A avaliação contextual será exploratória,
não permitindo afirmar precisão estatística
sem uma amostra rotulada e validada.

As contagens representam menções temáticas
em blocos de texto, não sentimentos,
reclamações ou consumidores individuais.

A revisão não separará comentários concatenados.

Os novos resultados permanecerão provisórios
até avaliação das evidências produzidas.


```python

# FIAP - TECH CHALLENGE FASE 04
# E6.2-V1 - Revisao controlada do vocabulario
# SOMENTE MEMORIA - nenhum arquivo sera modificado.

from pathlib import Path
from collections import Counter
import hashlib
import random
import re
import pandas as pd

# 1. Integridade

raiz = Path("/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON")

original = raiz / "01_DADOS_ORIGINAIS" / "01.amazon.csv"

entrada = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_05_desconto_pct_v01.csv"
)

hash_original = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

hash_t45 = (
    "5EFDD045F3FF6165FDC5865305E744A1"
    "4E7227F82654DD23983613026194EAA6"
)

def sha256(caminho):
    h = hashlib.sha256()

    with caminho.open("rb") as f:
        for bloco in iter(
            lambda: f.read(1048576), b""
        ):
            h.update(bloco)

    return h.hexdigest().upper()

for caminho in [original, entrada]:
    if not caminho.is_file():
        raise FileNotFoundError(caminho)

if sha256(original) != hash_original:
    raise ValueError("SHA-256 original divergente.")

if sha256(entrada) != hash_t45:
    raise ValueError("SHA-256 T4.5 divergente.")

base = pd.read_csv(
    entrada,
    dtype=str,
    keep_default_na=False,
    encoding="utf-8"
)

if base.shape != (1465, 22):
    raise ValueError("Dimensoes inesperadas.")

for coluna in ["product_id", "review_content"]:
    if coluna not in base.columns:
        raise ValueError(f"Coluna ausente: {coluna}")

textos = base["review_content"].map(str.strip).tolist()
produtos = base["product_id"].tolist()

if any(not texto for texto in textos):
    raise ValueError("Texto vazio inesperado.")

textos_unicos = list(dict.fromkeys(textos))

if len(textos_unicos) != 1212:
    raise ValueError("Textos distintos divergentes.")

if len(set(produtos)) != 1351:
    raise ValueError("Produtos distintos divergentes.")

print("=== E6.2-V1 - CONTROLE INICIAL ===")
print("SHA-256 original: CONFERIDO")
print("SHA-256 T4.5: CONFERIDO")
print("Registros:", len(base))
print("Textos distintos:", len(textos_unicos))
print("Produtos distintos:", len(set(produtos)))

# 2. Reproducao exata do vocabulario inicial

vocabulario_original = {
    "Qualidade e durabilidade": [
        "quality", "build quality", "durable",
        "durability", "scratch", "scratches",
        "damage", "damaged", "broken", "broke",
        "defect", "defective", "fragile", "material"
    ],
    "Funcionamento e desempenho": [
        "work", "works", "working",
        "function", "functions", "functioning",
        "performance", "perform", "performs",
        "speed", "slow", "fast",
        "not working", "stopped working",
        "heating", "cooling"
    ],
    "Preco e custo-beneficio": [
        "price", "priced", "value for money",
        "worth", "cost", "costly",
        "expensive", "cheap", "budget",
        "affordable", "money"
    ],
    "Bateria e carregamento": [
        "battery", "batteries", "charge",
        "charging", "charger", "charged",
        "power bank", "battery backup",
        "battery life"
    ],
    "Compatibilidade e instalacao": [
        "compatible", "compatibility",
        "connect", "connected", "connecting",
        "connection", "install", "installed",
        "installation", "setup", "set up",
        "bluetooth", "wifi", "wi-fi",
        "pairing", "pair", "driver", "port"
    ],
    "Entrega, devolucao e atendimento": [
        "delivery", "delivered", "shipping",
        "packaging", "package", "packed",
        "return", "returned", "refund",
        "replacement", "replace", "courier",
        "customer service", "seller"
    ]
}

def construir_padroes(vocabulario):
    padroes = {}

    for tema, termos in vocabulario.items():
        partes = []

        for termo in sorted(
            set(termos), key=len, reverse=True
        ):
            expressao = re.escape(termo)
            expressao = expressao.replace(
                r"\ ", r"\s+"
            )

            partes.append(expressao)

        padroes[tema] = re.compile(
            r"(?<!\w)(?:" +
            "|".join(partes) +
            r")(?!\w)",
            re.IGNORECASE
        )

    return padroes

padroes_originais = construir_padroes(
    vocabulario_original
)

def normalizar_espacos(texto):
    return re.sub(r"\s+", " ", texto)

conteudos_originais = {
    texto: normalizar_espacos(texto)
    for texto in textos_unicos
}

def classificar(texto, padroes):
    return {
        tema
        for tema, padrao in padroes.items()
        if padrao.search(texto)
    }

marcacoes_originais = {
    texto: classificar(
        conteudos_originais[texto],
        padroes_originais
    )
    for texto in textos_unicos
}

# Valores apresentados na execucao inicial E6.2

contagens_anteriores = {
    "Qualidade e durabilidade": 883,
    "Funcionamento e desempenho": 917,
    "Preco e custo-beneficio": 944,
    "Bateria e carregamento": 411,
    "Compatibilidade e instalacao": 413,
    "Entrega, devolucao e atendimento": 442
}

for tema, esperado in contagens_anteriores.items():
    obtido = sum(
        tema in marcacoes_originais[texto]
        for texto in textos_unicos
    )

    if obtido != esperado:
        raise ValueError(
            f"Reproducao divergente: {tema}. "
            f"Esperado={esperado}, obtido={obtido}"
        )

print("\n=== V1-01: REPRODUCAO DO E6.2 ===")
print("Seis frequencias anteriores: CONFERIDAS")

# 3. Revisao documentada do vocabulario

alteracoes = {
    "Qualidade e durabilidade": {
        "remover": [],
        "adicionar": [
            "sturdy", "sturdiness",
            "reliable", "reliability",
            "long lasting", "wear and tear"
        ]
    },
    "Funcionamento e desempenho": {
        "remover": [
            "work", "fast", "slow"
        ],
        "adicionar": [
            "does not work", "doesn't work",
            "did not work", "not work",
            "works well", "works fine",
            "working properly", "work properly",
            "slow performance", "slow response",
            "slow speed", "fast performance"
        ]
    },
    "Preco e custo-beneficio": {
        "remover": ["money"],
        "adicionar": [
            "overpriced", "reasonably priced",
            "reasonable price", "good value",
            "worth the money", "waste of money"
        ]
    },
    "Bateria e carregamento": {
        "remover": [],
        "adicionar": [
            "battery drains",
            "battery draining",
            "poor battery backup",
            "backup time",
            "battery endurance"
        ]
    },
    "Compatibilidade e instalacao": {
        "remover": [
            "pair", "driver", "port"
        ],
        "adicionar": [
            "usb port", "hdmi port",
            "device driver", "pair with",
            "paired", "driver installation"
        ]
    },
    "Entrega, devolucao e atendimento": {
        "remover": [
            "return", "replace"
        ],
        "adicionar": [
            "return request", "return policy",
            "return process", "return pickup",
            "return initiated", "initiated return",
            "returning the product",
            "product return", "replacement service",
            "replacement request",
            "replace the product",
            "late delivery", "delivery delay"
        ]
    }
}

vocabulario_revisado = {}

for tema, termos in vocabulario_original.items():
    remover = set(alteracoes[tema]["remover"])
    adicionar = alteracoes[tema]["adicionar"]

    if not remover.issubset(set(termos)):
        raise ValueError(
            f"Termo inexistente para remocao: {tema}"
        )

    atualizados = [
        termo for termo in termos
        if termo not in remover
    ]

    vocabulario_revisado[tema] = list(
        dict.fromkeys(atualizados + adicionar)
    )

padroes_revisados = construir_padroes(
    vocabulario_revisado
)

print("\n=== V1-02: ALTERACOES DO VOCABULARIO ===")

for tema, ajuste in alteracoes.items():
    print(f"\n{tema}")
    print("Removidos:", ajuste["remover"])
    print("Acrescentados:", ajuste["adicionar"])

# 4. Protecao textual temporaria
# Mascaramento ANTES do recorte e da busca.
# Os textos originais permanecem intactos.

padrao_email = re.compile(
    r"[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}"
)

padrao_url = re.compile(
    r"(?:https?://|www\.)\S+",
    re.IGNORECASE
)

padrao_dominio = re.compile(
    r"\b(?:[a-z0-9.-]+\.)+"
    r"(?:com|org|net|in|br|io|edu|gov)"
    r"/\S*",
    re.IGNORECASE
)

padrao_telefone = re.compile(
    r"(?<!\w)\+?\d[\d().\s-]{8,}\d(?!\w)"
)

def preparar_texto(texto):
    texto = padrao_email.sub("[EMAIL]", texto)
    texto = padrao_url.sub("[URL]", texto)
    texto = padrao_dominio.sub("[URL]", texto)
    texto = padrao_telefone.sub("[NUMERO]", texto)

    return normalizar_espacos(texto)

conteudos_revisados = {
    texto: preparar_texto(texto)
    for texto in textos_unicos
}

marcacoes_revisadas = {
    texto: classificar(
        conteudos_revisados[texto],
        padroes_revisados
    )
    for texto in textos_unicos
}

# 5. Frases de controle com resultados esperados
# Sao testes da regra, nao avaliacoes reais da base.

tema_func = "Funcionamento e desempenho"
tema_preco = "Preco e custo-beneficio"
tema_comp = "Compatibilidade e instalacao"
tema_entrega = "Entrega, devolucao e atendimento"

testes = [
    (
        tema_entrega,
        "Return value of spent money",
        False
    ),
    (
        tema_entrega,
        "I initiated a return request.",
        True
    ),
    (
        tema_func,
        "It does not seem to affect the work.",
        False
    ),
    (
        tema_func,
        "This product stopped working.",
        True
    ),
    (
        tema_func,
        "Fast delivery of this item.",
        False
    ),
    (
        tema_entrega,
        "Fast delivery of this item.",
        True
    ),
    (
        tema_preco,
        "Good value for money.",
        True
    ),
    (
        tema_comp,
        "I bought a pair of socks.",
        False
    ),
    (
        tema_comp,
        "The device uses an HDMI port.",
        True
    )
]

for tema, frase, esperado in testes:
    encontrado = bool(
        padroes_revisados[tema].search(
            preparar_texto(frase)
        )
    )

    if encontrado != esperado:
        raise AssertionError(
            f"Falha no teste: {tema} | {frase!r}"
        )

print("\n=== V1-03: FRASES DE CONTROLE ===")
print("Testes aprovados:", len(testes))
print("Falhas:", 0)

# 6. Comparacao das frequencias
# A unidade principal e o texto completo distinto.
# Um texto pode receber varios temas.

print("\n=== V1-04: FREQUENCIAS ANTES E DEPOIS ===")
print(
    "Tema | Antes | Depois | Diferenca | "
    "% dos 1212 textos | Registros | Produtos"
)

for tema in vocabulario_original:
    antes = contagens_anteriores[tema]

    depois = sum(
        tema in marcacoes_revisadas[texto]
        for texto in textos_unicos
    )

    registros = sum(
        tema in marcacoes_revisadas[texto]
        for texto in textos
    )

    associados = {
        produto
        for produto, texto in zip(produtos, textos)
        if tema in marcacoes_revisadas[texto]
    }

    print(
        f"{tema} | "
        f"{antes} | "
        f"{depois} | "
        f"{depois - antes:+d} | "
        f"{100 * depois / 1212:.2f}% | "
        f"{registros} | "
        f"{len(associados)}"
    )

sem_tema = sum(
    len(marcacoes_revisadas[texto]) == 0
    for texto in textos_unicos
)

multiplos_temas = sum(
    len(marcacoes_revisadas[texto]) >= 2
    for texto in textos_unicos
)

print("\nTextos distintos sem tema:", sem_tema)
print(
    "Textos distintos com dois ou mais temas:",
    multiplos_temas
)

# 7. Mudancas individuais de classificacao
# Diferenca liquida nao representa quantidade
# de textos efetivamente reclassificados.

print("\n=== V1-05: MUDANCAS DE CLASSIFICACAO ===")
print(
    "Tema | Mantidos | Retirados | Acrescentados"
)

for tema in vocabulario_original:
    mantidos = 0
    retirados = 0
    acrescentados = 0

    for texto in textos_unicos:
        antigo = tema in marcacoes_originais[texto]
        novo = tema in marcacoes_revisadas[texto]

        if antigo and novo:
            mantidos += 1
        elif antigo and not novo:
            retirados += 1
        elif novo and not antigo:
            acrescentados += 1

    print(
        f"{tema} | "
        f"{mantidos} | "
        f"{retirados} | "
        f"{acrescentados}"
    )

# 8. Amostras contextuais deterministicas
# A revisao humana deve examinar:
# - correspondencias mantidas
# - correspondencias retiradas
# - correspondencias acrescentadas

gerador = random.Random(42)

def escolher_amostra(candidatos, limite):
    return gerador.sample(
        candidatos,
        min(limite, len(candidatos))
    )

def exibir_contexto(texto, tema, versao):
    if versao == "anterior":
        conteudo = preparar_texto(
            conteudos_originais[texto]
        )
        padrao = padroes_originais[tema]
    else:
        conteudo = conteudos_revisados[texto]
        padrao = padroes_revisados[tema]

    achado = padrao.search(conteudo)

    if achado is None:
        # Em textos retirados, a busca antiga pode
        # estar dentro de URL mascarada.
        print(
            "  Correspondencia antiga removida "
            "pelo mascaramento."
        )
        return

    inicio = max(0, achado.start() - 85)
    fim = min(len(conteudo), achado.end() + 115)

    trecho = conteudo[inicio:fim]

    print(
        f"  termo={achado.group(0)!r} | "
        f"contexto={trecho!r}"
    )

print("\n=== V1-06: VALIDACAO CONTEXTUAL ===")
print(
    "Amostra exploratoria, nao estimativa "
    "estatistica da precisao."
)

for tema in vocabulario_original:
    mantidos = [
        texto for texto in textos_unicos
        if (
            tema in marcacoes_originais[texto]
            and tema in marcacoes_revisadas[texto]
        )
    ]

    retirados = [
        texto for texto in textos_unicos
        if (
            tema in marcacoes_originais[texto]
            and tema not in marcacoes_revisadas[texto]
        )
    ]

    acrescentados = [
        texto for texto in textos_unicos
        if (
            tema not in marcacoes_originais[texto]
            and tema in marcacoes_revisadas[texto]
        )
    ]

    print(f"\nTEMA: {tema}")

    for tipo, candidatos, limite, versao in [
        ("MANTIDOS", mantidos, 3, "revisada"),
        ("RETIRADOS", retirados, 2, "anterior"),
        ("ACRESCENTADOS", acrescentados, 2, "revisada")
    ]:
        print(
            f"\n{tipo}: {len(candidatos)} textos"
        )

        amostra = escolher_amostra(
            candidatos, limite
        )

        for numero, texto in enumerate(
            amostra, start=1
        ):
            print(f"Exemplo {numero}:")
            exibir_contexto(texto, tema, versao)

# 9. Amostra dos textos sem tema

nao_classificados = [
    texto
    for texto in textos_unicos
    if not marcacoes_revisadas[texto]
]

print("\n=== V1-07: TEXTOS SEM TEMA ===")
print("Total:", len(nao_classificados))

for numero, texto in enumerate(
    escolher_amostra(nao_classificados, 5),
    start=1
):
    trecho = conteudos_revisados[texto][:180]
    print(f"Exemplo {numero}: {trecho!r}")

# 10. Integridade final

if sha256(original) != hash_original:
    raise AssertionError("Arquivo original alterado.")

if sha256(entrada) != hash_t45:
    raise AssertionError("Arquivo T4.5 alterado.")

print("\n=== FIM DO E6.2-V1 ===")
print("Registros preservados:", len(base))
print("Textos distintos:", len(textos_unicos))
print("SHA-256 original: PRESERVADO")
print("SHA-256 T4.5: PRESERVADO")
print("Nenhum tratamento persistente.")
print("Nenhum arquivo derivado criado.")
print("Resultados revisados: PROVISORIOS")
print("Avaliacao humana: PENDENTE")
```


### 6.2.2. Revisão final E6.2-V2 — Validação temática

#### Objetivo

Revisar pontualmente as regras de identificação
temática, com base nos falsos positivos e nas
omissões identificados no procedimento E6.2-V1.

#### Fundamentação

A inspeção contextual da versão V1 identificou
correções adequadas, mas também a perda de
menções legítimas sobre devolução e custo-benefício.

A revisão V2 busca melhorar a especificidade
das correspondências sem excluir referências
relevantes aos assuntos investigados.

#### Alterações propostas

1. Recuperar expressões contextuais de devolução.
2. Recuperar expressões de custo-benefício.
3. Reduzir ambiguidades do termo "package".
4. Preservar os demais temas da versão V1.
5. Comparar quantitativamente V1 e V2.
6. Examinar amostras de correspondências
   mantidas, removidas e acrescentadas.
7. Inspecionar exemplos não classificados.

#### Metodologia

A análise utilizará expressões regulares,
vocabulário temático explícito e inspeção
qualitativa contextual.

Serão consideradas três unidades descritivas:

- Registros textuais;
- Textos completos distintos;
- Produtos distintos associados às menções.

Os indicadores não representarão o número
de consumidores individuais ou reclamações.

As alterações serão executadas exclusivamente
em memória, preservando os textos originais.

#### Critérios de validação

- SHA-256 original e T4.5 conferidos.
- 1.465 registros preservados.
- 1.212 textos completos distintos.
- Reprodução das frequências V1.
- Testes de controle aprovados.
- Comparação V1 versus V2.
- Inspeção contextual das alterações.
- Ausência de transformações persistentes.

#### Limitações

A identificação de temas por regras
não substitui a interpretação humana.

Expressões legítimas podem não ser reconhecidas,
e determinadas correspondências podem apresentar
ambiguidade contextual.

A análise não constitui classificação
supervisionada de sentimentos.

Amostras exploratórias não permitem estimar
a precisão estatística do classificador.

Os resultados permanecerão provisórios
até a avaliação das evidências da execução.


```python

# FIAP - TECH CHALLENGE FASE 04
# E6.2-V2 - Revisao final das regras de Text Mining
# SOMENTE LEITURA - nenhum arquivo sera criado ou alterado.

from pathlib import Path
from collections import Counter
import hashlib
import random
import re
import pandas as pd

# 1. Caminhos e integridade

raiz = Path("/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON")

original = raiz / "01_DADOS_ORIGINAIS" / "01.amazon.csv"

entrada = (
    raiz / "02_DADOS_TRATADOS" /
    "amazon_t4_05_desconto_pct_v01.csv"
)

hash_original = (
    "4BA126C4BA8EDD35E62E94CE1C853073"
    "D832DE380184AB4F5F97D1E314882C77"
)

hash_t45 = (
    "5EFDD045F3FF6165FDC5865305E744A1"
    "4E7227F82654DD23983613026194EAA6"
)

def sha256(caminho):
    h = hashlib.sha256()
    with caminho.open("rb") as f:
        for bloco in iter(lambda: f.read(1048576), b""):
            h.update(bloco)
    return h.hexdigest().upper()

for caminho in [original, entrada]:
    if not caminho.is_file():
        raise FileNotFoundError(caminho)

if sha256(original) != hash_original:
    raise ValueError("SHA-256 original divergente.")

if sha256(entrada) != hash_t45:
    raise ValueError("SHA-256 T4.5 divergente.")

base = pd.read_csv(
    entrada,
    dtype=str,
    keep_default_na=False,
    encoding="utf-8"
)

if base.shape != (1465, 22):
    raise ValueError("Dimensoes inesperadas.")

for campo in ["product_id", "category", "review_content"]:
    if campo not in base.columns:
        raise ValueError(f"Coluna ausente: {campo}")

textos = base["review_content"].map(str.strip).tolist()
produtos = base["product_id"].tolist()
categorias = base["category"].tolist()

if any(not t for t in textos):
    raise ValueError("Comentario vazio inesperado.")

textos_unicos = list(dict.fromkeys(textos))

if len(textos_unicos) != 1212:
    raise ValueError("Textos distintos divergentes.")

if len(set(produtos)) != 1351:
    raise ValueError("Produtos distintos divergentes.")

print("=== E6.2-V2 - CONTROLE INICIAL ===")
print("SHA-256 original: CONFERIDO")
print("SHA-256 T4.5: CONFERIDO")
print("Registros:", len(base))
print("Textos distintos:", len(textos_unicos))
print("Produtos distintos:", len(set(produtos)))

# 2. Reconstituicao EXATA do vocabulario V1

v1 = {
    "Qualidade e durabilidade": [
        "quality", "build quality", "durable",
        "durability", "scratch", "scratches",
        "damage", "damaged", "broken", "broke",
        "defect", "defective", "fragile", "material",
        "sturdy", "sturdiness", "reliable",
        "reliability", "long lasting", "wear and tear"
    ],
    "Funcionamento e desempenho": [
        "works", "working", "function", "functions",
        "functioning", "performance", "perform",
        "performs", "speed", "not working",
        "stopped working", "heating", "cooling",
        "does not work", "doesn't work",
        "did not work", "not work",
        "works well", "works fine",
        "working properly", "work properly",
        "slow performance", "slow response",
        "slow speed", "fast performance"
    ],
    "Preco e custo-beneficio": [
        "price", "priced", "value for money",
        "worth", "cost", "costly", "expensive",
        "cheap", "budget", "affordable",
        "overpriced", "reasonably priced",
        "reasonable price", "good value",
        "worth the money", "waste of money"
    ],
    "Bateria e carregamento": [
        "battery", "batteries", "charge",
        "charging", "charger", "charged",
        "power bank", "battery backup",
        "battery life", "battery drains",
        "battery draining", "poor battery backup",
        "backup time", "battery endurance"
    ],
    "Compatibilidade e instalacao": [
        "compatible", "compatibility", "connect",
        "connected", "connecting", "connection",
        "install", "installed", "installation",
        "setup", "set up", "bluetooth", "wifi",
        "wi-fi", "pairing", "usb port", "hdmi port",
        "device driver", "pair with", "paired",
        "driver installation"
    ],
    "Entrega, devolucao e atendimento": [
        "delivery", "delivered", "shipping",
        "packaging", "package", "packed",
        "returned", "refund", "replacement",
        "courier", "customer service", "seller",
        "return request", "return policy",
        "return process", "return pickup",
        "return initiated", "initiated return",
        "returning the product", "product return",
        "replacement service", "replacement request",
        "replace the product", "late delivery",
        "delivery delay"
    ]
}

def compilar(vocabulario):
    padroes = {}

    for tema, termos in vocabulario.items():
        partes = []

        for termo in sorted(
            set(termos), key=len, reverse=True
        ):
            parte = re.escape(termo)
            parte = parte.replace(r"\ ", r"\s+")
            partes.append(parte)

        padroes[tema] = re.compile(
            r"(?<!\w)(?:" + "|".join(partes) + r")(?!\w)",
            re.IGNORECASE
        )

    return padroes

# 3. Mascaramento identico ao procedimento V1

padrao_email = re.compile(
    r"[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}"
)

padrao_url = re.compile(
    r"(?:https?://|www\.)\S+",
    re.IGNORECASE
)

padrao_dominio = re.compile(
    r"\b(?:[a-z0-9.-]+\.)+"
    r"(?:com|org|net|in|br|io|edu|gov)"
    r"/\S*",
    re.IGNORECASE
)

padrao_telefone = re.compile(
    r"(?<!\w)\+?\d[\d().\s-]{8,}\d(?!\w)"
)

def preparar_texto(texto):
    texto = padrao_email.sub("[EMAIL]", texto)
    texto = padrao_url.sub("[URL]", texto)
    texto = padrao_dominio.sub("[URL]", texto)
    texto = padrao_telefone.sub("[NUMERO]", texto)
    return re.sub(r"\s+", " ", texto)

conteudos = {
    texto: preparar_texto(texto)
    for texto in textos_unicos
}

def identificar(conteudo, padroes):
    return {
        tema
        for tema, padrao in padroes.items()
        if padrao.search(conteudo)
    }

padroes_v1 = compilar(v1)

marcacoes_v1 = {
    texto: identificar(conteudos[texto], padroes_v1)
    for texto in textos_unicos
}

esperados_v1 = {
    "Qualidade e durabilidade": 918,
    "Funcionamento e desempenho": 836,
    "Preco e custo-beneficio": 923,
    "Bateria e carregamento": 411,
    "Compatibilidade e instalacao": 388,
    "Entrega, devolucao e atendimento": 403
}

for tema, esperado in esperados_v1.items():
    observado = sum(
        tema in marcacoes_v1[t]
        for t in textos_unicos
    )

    if observado != esperado:
        raise ValueError(
            f"V1 divergente em {tema}: "
            f"{observado} versus {esperado}"
        )

print("\n=== V2-01: REPRODUCAO DA V1 ===")
print("Seis frequencias V1: CONFERIDAS")

# 4. Alteracoes V2 - exclusivamente pontuais

v2 = {
    tema: termos.copy()
    for tema, termos in v1.items()
}

novos_termos = {
    "Preco e custo-beneficio": [
        "money waste",
        "waste money",
        "wasted my money",
        "waste my money",
        "money well spent",
        "for the money",
        "money spent",
        "money's worth"
    ],
    "Entrega, devolucao e atendimento": [
        "too late to return",
        "cannot return",
        "can't return",
        "unable to return",
        "want to return",
        "need to return",
        "have to return",
        "return it",
        "return this",
        "return the",
        "return now",
        "return window",
        "return period",
        "return option",
        "no return",
        "not able to return",
        "well packed",
        "poorly packed",
        "damaged package",
        "package arrived",
        "package was damaged",
        "package was open",
        "sealed package",
        "open package"
    ]
}

termos_removidos = {
    "Entrega, devolucao e atendimento": [
        "package", "packed"
    ]
}

for tema, remover in termos_removidos.items():
    v2[tema] = [
        termo for termo in v2[tema]
        if termo not in remover
    ]

for tema, adicionar in novos_termos.items():
    v2[tema] = list(
        dict.fromkeys(v2[tema] + adicionar)
    )

padroes_v2 = compilar(v2)

marcacoes_v2 = {
    texto: identificar(conteudos[texto], padroes_v2)
    for texto in textos_unicos
}

print("\n=== V2-02: ALTERACOES CONTROLADAS ===")
for tema in v1:
    rem = sorted(set(v1[tema]) - set(v2[tema]))
    adc = sorted(set(v2[tema]) - set(v1[tema]))

    print(f"\n{tema}")
    print("Removidos:", rem)
    print("Adicionados:", adc)

# 5. Frases artificiais de controle
# Testam regras, nao acuracia em dados reais.

testes = [
    (
        "Entrega, devolucao e atendimento",
        "Return value of spent money",
        False
    ),
    (
        "Entrega, devolucao e atendimento",
        "It's too late to return now",
        True
    ),
    (
        "Entrega, devolucao e atendimento",
        "I cannot return this item",
        True
    ),
    (
        "Entrega, devolucao e atendimento",
        "I initiated a return request",
        True
    ),
    (
        "Entrega, devolucao e atendimento",
        "The product is a good package",
        False
    ),
    (
        "Entrega, devolucao e atendimento",
        "Packed with useful features",
        False
    ),
    (
        "Entrega, devolucao e atendimento",
        "The item was well packed",
        True
    ),
    (
        "Preco e custo-beneficio",
        "This item is money waste",
        True
    ),
    (
        "Preco e custo-beneficio",
        "Excellent for the money",
        True
    ),
    (
        "Preco e custo-beneficio",
        "I wasted my money",
        True
    ),
    (
        "Compatibilidade e instalacao",
        "A pair of screen guards",
        False
    ),
    (
        "Compatibilidade e instalacao",
        "The HDMI port is compatible",
        True
    ),
    (
        "Funcionamento e desempenho",
        "Fast delivery arrived today",
        False
    ),
    (
        "Bateria e carregamento",
        "Fast charging is supported",
        True
    ),
    (
        "Qualidade e durabilidade",
        "The product is sturdy",
        True
    )
]

for tema, frase, esperado in testes:
    obtido = bool(
        padroes_v2[tema].search(preparar_texto(frase))
    )

    if obtido != esperado:
        raise AssertionError(
            f"Teste falhou: {tema} | {frase!r}"
        )

print("\n=== V2-03: TESTES DE CONTROLE ===")
print("Testes aprovados:", len(testes))
print("Falhas:", 0)

# 6. Frequencias revisadas e conciliacao

print("\n=== V2-04: FREQUENCIAS V1 VERSUS V2 ===")
print(
    "Tema | V1 textos | V2 textos | Diferenca | "
    "% textos V2 | Registros V2 | Produtos V2"
)

for tema in v1:
    antes = esperados_v1[tema]

    depois = sum(
        tema in marcacoes_v2[t]
        for t in textos_unicos
    )

    registros = sum(
        tema in marcacoes_v2[t]
        for t in textos
    )

    associados = {
        produto
        for produto, texto in zip(produtos, textos)
        if tema in marcacoes_v2[texto]
    }

    print(
        f"{tema} | "
        f"{antes} | "
        f"{depois} | "
        f"{depois - antes:+d} | "
        f"{100 * depois / 1212:.2f}% | "
        f"{registros} | "
        f"{len(associados)}"
    )

sem_tema = sum(
    not marcacoes_v2[t]
    for t in textos_unicos
)

multiplos = sum(
    len(marcacoes_v2[t]) >= 2
    for t in textos_unicos
)

print("\nTextos distintos sem tema:", sem_tema)
print("Textos com dois ou mais temas:", multiplos)
print("Temas nao exclusivos: SIM")

# 7. Matriz de mudancas por tema

print("\n=== V2-05: MUDANCAS DE CLASSIFICACAO ===")
print("Tema | Mantidos | Retirados | Acrescentados")

for tema in v1:
    mantidos = sum(
        tema in marcacoes_v1[t]
        and tema in marcacoes_v2[t]
        for t in textos_unicos
    )

    retirados = sum(
        tema in marcacoes_v1[t]
        and tema not in marcacoes_v2[t]
        for t in textos_unicos
    )

    adicionados = sum(
        tema not in marcacoes_v1[t]
        and tema in marcacoes_v2[t]
        for t in textos_unicos
    )

    if (
        mantidos + retirados != esperados_v1[tema]
    ):
        raise AssertionError(
            f"Conciliacao V1 incorreta: {tema}"
        )

    print(
        f"{tema} | "
        f"{mantidos} | "
        f"{retirados} | "
        f"{adicionados}"
    )

# 8. Temas por categoria principal
# Uma contagem por product_id associado.

categoria_por_produto = {}

for produto, caminho in zip(produtos, categorias):
    categoria = caminho.split("|")[0]

    if (
        produto in categoria_por_produto
        and categoria_por_produto[produto] != categoria
    ):
        raise ValueError(
            "Categoria principal divergente por produto."
        )

    categoria_por_produto[produto] = categoria

totais_categoria = Counter(
    categoria_por_produto.values()
)

principais = [
    "Electronics",
    "Home&Kitchen",
    "Computers&Accessories"
]

rotulos = {
    "Electronics": "Eletronicos",
    "Home&Kitchen": "Casa e Cozinha",
    "Computers&Accessories":
        "Informatica e Acessorios"
}

print("\n=== V2-06: TEMAS POR CATEGORIA ===")
print(
    "Categoria | Tema | Produtos associados | "
    "Total de produtos | % produtos"
)

for categoria in principais:
    total_categoria = totais_categoria[categoria]

    for tema in v2:
        associados = {
            produto
            for produto, texto in zip(produtos, textos)
            if (
                categoria_por_produto[produto] == categoria
                and tema in marcacoes_v2[texto]
            )
        }

        print(
            f"{rotulos[categoria]} | "
            f"{tema} | "
            f"{len(associados)} | "
            f"{total_categoria} | "
            f"{100 * len(associados) / total_categoria:.2f}%"
        )

# 9. Amostras para avaliacao humana
# Selecionadas com semente fixa.
# Textos mascarados antes da exibicao.

gerador = random.Random(42)

primeira_posicao = {}

for posicao, texto in enumerate(textos, start=1):
    primeira_posicao.setdefault(texto, posicao)

def selecionar(candidatos, limite=2):
    return gerador.sample(
        candidatos, min(limite, len(candidatos))
    )

def mostrar_evidencia(texto, tema, versao):
    conteudo = conteudos[texto]

    padrao = (
        padroes_v1[tema]
        if versao == "V1"
        else padroes_v2[tema]
    )

    achado = padrao.search(conteudo)

    if achado is None:
        print("  Correspondencia nao recuperada.")
        return

    inicio = max(0, achado.start() - 80)
    fim = min(len(conteudo), achado.end() + 110)

    print(
        f"  Registro {primeira_posicao[texto]} | "
        f"termo={achado.group(0)!r} | "
        f"contexto={conteudo[inicio:fim]!r}"
    )

print("\n=== V2-07: AMOSTRAS CONTEXTUAIS ===")
print(
    "Objetivo: identificar acertos, falsos positivos "
    "e possiveis omissoes."
)

for tema in v2:
    mantidos = [
        t for t in textos_unicos
        if (
            tema in marcacoes_v1[t]
            and tema in marcacoes_v2[t]
        )
    ]

    retirados = [
        t for t in textos_unicos
        if (
            tema in marcacoes_v1[t]
            and tema not in marcacoes_v2[t]
        )
    ]

    adicionados = [
        t for t in textos_unicos
        if (
            tema not in marcacoes_v1[t]
            and tema in marcacoes_v2[t]
        )
    ]

    nao_marcados = [
        t for t in textos_unicos
        if tema not in marcacoes_v2[t]
    ]

    print(f"\nTEMA: {tema}")

    for grupo, candidatos, versao in [
        ("MANTIDOS", mantidos, "V2"),
        ("RETIRADOS", retirados, "V1"),
        ("ACRESCENTADOS", adicionados, "V2")
    ]:
        print(
            f"\n{grupo}: {len(candidatos)} textos"
        )

        for texto in selecionar(candidatos):
            mostrar_evidencia(texto, tema, versao)

    print(
        f"\nNAO MARCADOS: {len(nao_marcados)} textos"
    )

    for texto in selecionar(nao_marcados):
        conteudo = conteudos[texto]

        print(
            f"  Registro {primeira_posicao[texto]} | "
            f"trecho={conteudo[:180]!r}"
        )

# 10. Textos sem nenhum tema

sem_classificacao = [
    t for t in textos_unicos
    if not marcacoes_v2[t]
]

print("\n=== V2-08: TEXTOS SEM TEMA ===")
print("Total:", len(sem_classificacao))

for texto in selecionar(sem_classificacao, 5):
    print(
        f"Registro {primeira_posicao[texto]} | "
        f"trecho={conteudos[texto][:180]!r}"
    )

# 11. Integridade final

if sha256(original) != hash_original:
    raise AssertionError("Base original alterada.")

if sha256(entrada) != hash_t45:
    raise AssertionError("Base T4.5 alterada.")

print("\n=== FIM DO E6.2-V2 ===")
print("Registros:", len(base))
print("Textos distintos:", len(textos_unicos))
print("Produtos distintos:", len(set(produtos)))
print("SHA-256 original: PRESERVADO")
print("SHA-256 T4.5: PRESERVADO")
print("Nenhum tratamento persistente.")
print("Nenhum arquivo derivado criado.")
print("Classificacao tematica: PROVISORIA")
print("Validacao humana: PENDENTE")
```


### 6.2.3. Resultados finais e limitações do Text Mining

#### Objetivo

Consolidar os resultados da mineração temática
exploratória dos comentários do Amazon Sales
Dataset, considerando a versão inicial E6.2
e as revisões metodológicas V1 e V2.

#### Fundamentação acadêmica

A metodologia foi fundamentada na técnica de
Text Mining, apresentada no material FIAP
"Geração de insights", página 5, e nas
orientações do Tech Challenge Amazon sobre
a exploração das percepções dos consumidores.

A abordagem adotada consiste na identificação
de temas por meio de vocabulário explícito,
expressões regulares e inspeção contextual.

Não foi desenvolvido classificador supervisionado
de sentimentos.

#### Unidade de análise

| Indicador | Resultado |
|---|---:|
| Registros textuais | 1.465 |
| Textos completos distintos | 1.212 |
| Ocorrências excedentes de textos repetidos | 253 |
| Produtos distintos | 1.351 |

A unidade principal da mineração temática
foi o texto completo distinto.

Um texto pode conter vários assuntos
ou opiniões concatenadas.

Assim, a frequência temática não equivale
ao número de consumidores individuais.

#### Histórico das revisões

| Tema | E6.2 inicial | V1 | V2 |
|---|---:|---:|---:|
| Qualidade e durabilidade | 883 | 918 | 918 |

| Bateria e carregamento | 411 | 411 | 411 |
| Compatibilidade e instalação | 413 | 388 | 388 |
| Entrega, devolução e atendimento | 442 | 403 | 392 |

As alterações decorreram da revisão de termos
genéricos, da inclusão de expressões contextuais
e da identificação de ambiguidades.

Reduções ou aumentos nas frequências não foram
interpretados automaticamente como melhoria
de qualidade.

#### Resultados temáticos da versão V2

| Tema | Textos distintos | Percentual |
|---|---:|---:|

| Entrega, devolução e atendimento | 392 | 32,34% |
| Compatibilidade e instalação | 388 | 32,01% |

Os percentuais utilizam como denominador
os 1.212 textos completos distintos.

Os temas não são mutuamente exclusivos.

Um mesmo bloco textual pode contribuir
para a frequência de vários temas.

Foram identificados 1.099 textos distintos
com dois ou mais temas e 19 sem qualquer
tema reconhecido pelo vocabulário V2.

#### Comparação entre categorias principais

Os percentuais abaixo representam produtos
distintos associados a pelo menos um bloco
textual contendo a menção temática.

| Tema | Eletrônicos | Casa e Cozinha | Informática e Acessórios |

Denominadores por categoria:

- Eletrônicos: 490 produtos distintos.

As diferenças entre categorias podem refletir
as características dos produtos e a composição
dos respectivos segmentos.

Não devem ser interpretadas automaticamente
como diferenças nas preferências individuais
dos consumidores.

#### Validação das regras

A versão V2 reproduziu as seis frequências
da versão V1 e aprovou 15 testes com frases
de controle.

A inspeção contextual identificou ajustes
adequados, incluindo a recuperação de
referências legítimas a devoluções e
a experiências negativas de custo-benefício.

Entretanto, também foram observadas
limitações residuais, incluindo a possível
perda de menções legítimas à embalagem.

A inspeção realizada foi exploratória.

Não foi produzida uma amostra rotulada
independentemente que permitisse estimar
precisão, recall ou F1 da classificação.

Portanto, os indicadores não devem ser
apresentados como classificações temáticas
perfeitamente validadas.

#### Interpretação de negócio

Os temas mais frequentemente identificados
pela versão V2 foram:

1. Preço e custo-benefício;
2. Qualidade e durabilidade;
3. Funcionamento e desempenho.

Esses assuntos constituem possíveis eixos
para investigação da experiência dos clientes.

Os resultados não demonstram, isoladamente,
se as menções foram positivas ou negativas.

Também não permitem afirmar que determinado
tema seja mais valorizado ou mais criticado
pelos consumidores.

A análise deverá ser apresentada como
identificação exploratória de assuntos
mencionados nos blocos textuais.

#### Aplicação no Power BI

Os resultados poderão fundamentar a página
Customer Insights, mediante:

- Frequência de menções temáticas;
- Comparação temática por categoria;
- Distribuição de produtos associados
  a determinados assuntos;
- Exemplos contextualizados e revisados.

Os indicadores deverão explicitar sua
unidade de análise e seu caráter exploratório.

Os rótulos apresentados no dashboard
serão traduzidos para português,
preservando os valores originais e a
rastreabilidade das classificações.

A criação de estruturas derivadas para
o Power BI dependerá de procedimento
específico, com validação e autorização.

#### Limitações metodológicas

- Os comentários podem estar concatenados.
- Não foi validado delimitador de separação.
- Os textos repetidos exigem cautela.
- O vocabulário utiliza predominantemente inglês.
- Não foi realizada identificação sistemática
  do idioma de todos os textos.
- O método depende de expressões predefinidas.
- Podem ocorrer falsos positivos e negativos.
- Os testes de controle não estimam acurácia.
- Não foi realizada classificação validada
  de sentimentos por aspectos.
- Os resultados não representam consumidores
  ou avaliações individuais únicos.
- Não é possível estabelecer causalidade.

#### Integridade

A execução confirmou os hashes SHA-256
da fonte original e da versão T4.5.

As dimensões e as contagens fundamentais
foram preservadas.

Nenhum arquivo derivado foi criado.

Nenhum tratamento persistente foi realizado.

#### Conclusão

O procedimento E6.2, incluindo as revisões
V1 e V2, produziu uma caracterização
exploratória dos assuntos encontrados nos
comentários dos produtos Amazon.

A abordagem identificou recorrência de
menções a preço, qualidade, funcionamento,
bateria, compatibilidade e experiência
de entrega e atendimento.

Os resultados são adequados como evidência
exploratória, desde que acompanhados das
ressalvas sobre concatenação, repetições
e limitações da classificação baseada
em regras.

```python

# PBI-38 — Auditoria cruzada de preços e descontos por categoria
# Apenas leitura e validação: não altera os arquivos.

from pathlib import Path
import pandas as pd
import hashlib

arquivo = Path(
    "/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON/"
    "02_DADOS_TRATADOS/amazon_t4_05_desconto_pct_v01.csv"
)

sha_esperado = (
    "5EFDD045F3FF6165FDC5865305E744A14"
    "E7227F82654DD23983613026194EAA6"
)

assert arquivo.is_file(), "Arquivo T4.5 não localizado."

sha_atual = hashlib.sha256(arquivo.read_bytes()).hexdigest().upper()
assert sha_atual == sha_esperado, "SHA-256 divergente."

df = pd.read_csv(
    arquivo,
    dtype={"product_id": "string", "category": "string"}
)

assert len(df) == 1465

# Categoria principal extraída sem alterar a coluna original
df["categoria_auditoria"] = (
    df["category"].str.split("|", regex=False).str[0]
)

for coluna in [
    "actual_price_inr",
    "discounted_price_inr",
    "discount_rate"
]:
    df[coluna] = pd.to_numeric(df[coluna], errors="raise")

resultado = (
    df.groupby("categoria_auditoria")
      .agg(
          registros=("product_id", "size"),
          produtos_distintos=("product_id", "nunique"),
          preco_original_mediano=("actual_price_inr", "median"),
          preco_descontado_mediano=("discounted_price_inr", "median"),
          desconto_medio=("discount_rate", "mean")
      )
)

resultado["desconto_medio_pct"] = (
    resultado["desconto_medio"] * 100
)

# Referências observadas nos gráficos Power BI
referencias = {
    "Electronics": (3199.50, 1399.00, 50.83),
    "Home&Kitchen": (2000.00, 1199.00, 40.12),
    "Computers&Accessories": (999.00, 349.00, 54.02)
}

print("SHA-256 validado:", sha_atual)
print("Registros:", len(df))
print("Categorias principais:", len(resultado))
print("\nCONCILIAÇÃO COLAB × POWER BI")

for categoria, esperado in referencias.items():
    linha = resultado.loc[categoria]

    obtido = (
        round(float(linha["preco_original_mediano"]), 2),
        round(float(linha["preco_descontado_mediano"]), 2),
        round(float(linha["desconto_medio_pct"]), 2)
    )

    status = "OK" if obtido == esperado else "DIVERGÊNCIA"

    print(f"\n{categoria}: {status}")
    print("  Colab:", obtido)
    print("  Power BI:", esperado)

print("\nRESULTADOS DAS NOVE CATEGORIAS")
display(
    resultado[
        [
            "registros",
            "produtos_distintos",
            "preco_original_mediano",
            "preco_descontado_mediano",
            "desconto_medio_pct"
        ]
    ].round(2)
)
```

```python

# PBI-39 — Auditoria independente das faixas de desconto
# Somente leitura. Não altera o CSV nem o Power BI.

from pathlib import Path
import hashlib
import pandas as pd

arquivo = Path(
    "/content/drive/MyDrive/FIAP_TC_FASE04_AMAZON/"
    "02_DADOS_TRATADOS/amazon_t4_05_desconto_pct_v01.csv"
)

sha_esperado = (
    "5EFDD045F3FF6165FDC5865305E744A14"
    "E7227F82654DD23983613026194EAA6"
)

assert arquivo.is_file(), "Arquivo T4.5 não localizado."

sha_atual = hashlib.sha256(arquivo.read_bytes()).hexdigest().upper()
assert sha_atual == sha_esperado, "Divergência no SHA-256."

df = pd.read_csv(arquivo)

taxa = pd.to_numeric(df["discount_rate"], errors="raise")

assert len(df) == 1465
assert taxa.notna().all()
assert taxa.between(0, 1).all()

faixas = {
    "0% a <25%": ((taxa >= 0) & (taxa < 0.25)).sum(),
    "25% a <50%": ((taxa >= 0.25) & (taxa < 0.50)).sum(),
    "50% a <75%": ((taxa >= 0.50) & (taxa < 0.75)).sum(),
    "75% a 100%": ((taxa >= 0.75) & (taxa <= 1)).sum()
}

esperado_bi = [234, 480, 572, 179]

resultado = pd.DataFrame({
    "faixa": list(faixas.keys()),
    "colab": list(faixas.values()),
    "power_bi": esperado_bi
})

resultado["status"] = resultado.apply(
    lambda linha: "OK" if linha["colab"] == linha["power_bi"]
    else "DIVERGÊNCIA",
    axis=1
)

resultado["participacao_pct"] = (
    resultado["colab"] / len(df) * 100
).round(2)

print("Registros:", len(df))
print("Total classificado:", resultado["colab"].sum())
print("SHA-256 validado:", sha_atual)

display(resultado)

print(
    "Registros com desconto >= 50%:",
    int((taxa >= 0.50).sum())
)

print(
    "Participação com desconto >= 50%:",
    f"{(taxa >= 0.50).mean() * 100:.2f}%"
)

assert resultado["status"].eq("OK").all(), (
    "Há divergência entre Colab e Power BI."
)

assert resultado["colab"].sum() == 1465

print("PBI-39: CONCILIAÇÃO APROVADA")
```
